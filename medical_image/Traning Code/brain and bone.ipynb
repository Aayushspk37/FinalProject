```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 1 — Setup (UNCHANGED)                                                   ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

!pip install -q timm einops albumentations opencv-python-headless

import os, time, random, copy, warnings
from pathlib import Path
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image
import cv2
from tqdm.auto import tqdm
from scipy.ndimage import zoom

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
import torchvision.models as models

import albumentations as A
from albumentations.pytorch import ToTensorV2
from einops import rearrange

from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, roc_curve, confusion_matrix, classification_report
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize

from PIL import ImageFile
ImageFile.LOAD_TRUNCATED_IMAGES = True 

warnings.filterwarnings('ignore')

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
CKDIR = OUTDIR / 'checkpoints'
PLTDIR = OUTDIR / 'plots'
for d in [CKDIR, PLTDIR]: d.mkdir(exist_ok=True)

MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.JPG', '.JPEG', '.PNG'}
print(f"✅ Setup complete on {DEVICE}")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 2 — Hyperparameters (OPTIMIZED)                                         ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

HPARAMS = {
    'img_size': 224,           # ✅ 224 is 3x faster than 384
    'batch_size': 32,          # ✅ Doubled batch size
    'num_workers': 4,          # ✅ Parallel data loading
    'num_epochs': 40,
    'lr_backbone': 5e-6, 'lr_head': 1e-4, 'weight_decay': 1e-4,
    'warmup_epochs': 3, 'freeze_epochs': 5, 'patience': 10,
    'aug_level': 'light',      # ✅ Lighter augmentations = faster
    'd_model': 384,            # ✅ Halved from 768
    'nhead': 6,                # ✅ Halved from 12
    'n_layers': 4,             # ✅ Reduced from 6
    'dim_ffn': 1536,           # ✅ Halved from 3072
    'head_dropout': 0.3, 'label_smoothing': 0.1,
    'mixup_alpha': 0.2, 'cutmix_alpha': 1.0,
    'use_mixup': True, 'use_cutmix': True, 'mixup_prob': 0.3,
    'backbone': 'convnext_base',  # ✅ Base instead of Large
    'num_classes': 4,
    'class_names': ['notumor', 'glioma', 'meningioma', 'pituitary']
}
print("✅ Hyperparameters ready (OPTIMIZED)")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 3 & 4 — MRI Augmentations (UNCHANGED)                                   ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def get_mri_transforms(split, img_size=224, level='light'):
    norm = [A.Normalize(mean=MEAN, std=STD), ToTensorV2()]
    if split != 'train': 
        return A.Compose([A.Resize(height=img_size, width=img_size)] + norm)
    
    base = [
        A.Resize(height=img_size, width=img_size),
        A.HorizontalFlip(p=0.5),
        A.VerticalFlip(p=0.3),
        A.Rotate(limit=15, p=0.5, border_mode=cv2.BORDER_CONSTANT, value=0),
    ]
    light = [
        A.RandomBrightnessContrast(brightness_limit=0.15, contrast_limit=0.15, p=0.4),
        A.GaussNoise(var_limit=(10.0, 30.0), p=0.15),
    ]
    medium = [
        A.ShiftScaleRotate(shift_limit=0.1, scale_limit=0.1, rotate_limit=10, p=0.4, border_mode=cv2.BORDER_CONSTANT, value=0),
        A.GaussianBlur(blur_limit=(3, 5), p=0.15),
    ]
    heavy = [
        A.ElasticTransform(alpha=30, sigma=3, p=0.1),
        A.CoarseDropout(max_holes=6, max_height=int(img_size*0.08), max_width=int(img_size*0.08), p=0.15, fill_value=0)
    ]
    
    if level == 'light': aug = base + light
    elif level == 'medium': aug = base + light + medium
    else: aug = base + light + medium + heavy
    
    return A.Compose(aug + norm)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 5 & 6 — Loss & Model Architecture (LIGHTWEIGHT + PARAMETERIZED)         ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class SmoothCELoss(nn.Module):
    def __init__(self, label_smoothing=0.1):
        super().__init__()
        self.criterion = nn.CrossEntropyLoss(label_smoothing=label_smoothing)
    def forward(self, inputs, targets): 
        return self.criterion(inputs, targets)

class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): 
        super().__init__()
        self.drop_prob = drop_prob
        
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training:
            return x
        keep_prob = float(1.0 - self.drop_prob)
        random_mask = torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * keep_prob
        return x * random_mask

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3)
        self.proj = nn.Linear(d_model, d_model)
        self.attn_drop = nn.Dropout(dropout)
        self.proj_drop = nn.Dropout(dropout)
        self.attn_weights = None
        
    def forward(self, x, H, W):
        B, N, C = x.shape
        x = x.reshape(B, H, W, C)
        
        pad_h = (self.window_size - H % self.window_size) % self.window_size
        pad_w = (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0: 
            x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C)
        x = x.permute(0, 1, 3, 2, 4, 5).reshape(-1, self.window_size**2, C)
        
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2, 0, 3, 1, 4)
        attn = (qkv[0] @ qkv[1].transpose(-2, -1)) * self.scale
        attn = attn.softmax(dim=-1)
        self.attn_weights = attn.detach()
        
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1, 2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C)
        x = x.permute(0, 1, 3, 2, 4, 5).reshape(B, Hp, Wp, C)
        
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1 = nn.LayerNorm(d_model)
        self.norm2 = nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model * 2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop),
            nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop)
        )
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights = None
        self.window_attn_weights = None
        
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        
        window_out = self.window_attn(x_norm[:, 1:, :], H, W)
        self.window_attn_weights = self.window_attn.attn_weights
        
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1 - gate) * window_out], dim=1)
        
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self, backbone_name='convnext_base'):
        super().__init__()
        if backbone_name == 'convnext_large':
            self.backbone = models.convnext_large(weights=None)
        else:  # convnext_base - faster
            self.backbone = models.convnext_base(weights=None)
        self.feature_extractor = self.backbone.features
        
    def forward(self, x):
        return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes=4, img_size=224, d_model=384, nhead=6, n_layers=4, 
                 dim_ffn=1536, backbone_name='convnext_base'):
        super().__init__()
        self.d_model = d_model
        self.backbone = ImprovedBackbone(backbone_name)
        
        # ✅ FIX: Use actual img_size instead of hardcoded 384
        with torch.no_grad():
            _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        
        self.spatial_h, self.spatial_w = h, w
        self.img_size = img_size  # Store for heatmap generation
        
        # Projection layers
        self.proj = nn.Sequential(
            nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(),
            nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU()
        )
        self.norm_proj = nn.LayerNorm(d_model)
        self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h * w + 1, d_model))
        
        nn.init.trunc_normal_(self.cls_token, std=0.02)
        nn.init.trunc_normal_(self.pos_embed, std=0.02)
        
        # ✅ FIX: Use parameters instead of hardcoded values
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([
            TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i]) 
            for i in range(n_layers)
        ])
        
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(
            nn.LayerNorm(d_model),
            nn.Linear(d_model, d_model // 2), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(d_model // 2, d_model // 4), nn.GELU(), nn.Dropout(0.15),
            nn.Linear(d_model // 4, num_classes)
        )
        
    def forward(self, x):
        B = x.size(0)
        tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens)
        tokens = torch.cat([self.cls_token.expand(B, -1, -1), tokens], dim=1) + self.pos_embed
        
        for block in self.encoder:
            tokens = block(tokens, self.spatial_h, self.spatial_w)
            
        return self.head(self.norm(tokens)[:, 0])
    
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w
    
    def unfreeze_backbone_partial(self, n=3):
        for p in self.backbone.feature_extractor.parameters(): 
            p.requires_grad_(False)
        for child in list(self.backbone.feature_extractor.children())[-n:]:
            for p in child.parameters(): 
                p.requires_grad_(True)

print("✅ Model Architecture ready (LIGHTWEIGHT + PARAMETERIZED)")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 7, 8, 9 — Data Loader, Mixup, Trainer (OPTIMIZED)                       ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class MRIDataset(Dataset):
    def __init__(self, samples, class_names, transform=None):
        self.samples = samples
        self.transform = transform
        self.class_names = class_names
        self.class_to_idx = {n: i for i, n in enumerate(class_names)}
        
    def __len__(self): 
        return len(self.samples)
    
    def __getitem__(self, idx):
        p, l = self.samples[idx]
        l = self.class_to_idx.get(l, l)
        
        img = cv2.imread(p)
        if img is None:
            img = np.array(Image.open(p).convert('RGB'))
        else:
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            
        if self.transform:
            img = self.transform(image=img)['image']
        elif not isinstance(img, torch.Tensor):
            if img.max() > 1: img = img / 255.0
            img = torch.from_numpy(img).float().permute(2, 0, 1)
            
        return img.float(), l
    
    def class_weights(self):
        labels = [self.class_to_idx.get(l, l) for _, l in self.samples]
        counts = torch.bincount(torch.tensor(labels), minlength=len(self.class_names)).float()
        beta = 0.9999
        effective_num = 1.0 - torch.pow(beta, counts)
        return ((1.0 - beta) / (effective_num + 1e-8)) / len(self.class_names) * len(self.class_names)

def load_brain_tumor():
    samples, class_names = [], HPARAMS['class_names']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Brain Tumor MRI Dataset/Brain Tumor MRI Dataset')
    
    for split in ['Training', 'Testing']:
        split_path = base / split
        if not split_path.exists(): continue
        
        for f in split_path.iterdir():
            if not f.is_dir(): continue
            fn = f.name.lower().replace('_', '')
            
            if 'notumor' in fn or 'no' in fn: cname = 'notumor'
            elif 'glioma' in fn: cname = 'glioma'
            elif 'meningioma' in fn: cname = 'meningioma'
            elif 'pituitary' in fn: cname = 'pituitary'
            else: continue
            
            count = 0
            for ext in VALID_EXTS:
                for p in f.rglob(f'*{ext}'):
                    samples.append((str(p), cname))
                    count += 1
            if count > 0: 
                print(f"   {split}/{cname}: {count} images")
    
    return samples, class_names

def mixup_data(x, y, alpha=0.2):
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(x.size(0)).to(x.device)
    return lam * x + (1 - lam) * x[idx], y, y[idx], lam

def cutmix_data(x, y, alpha=1.0):
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(x.size(0)).to(x.device)
    W, H = x.size(2), x.size(3)
    r = np.sqrt(1.0 - lam)
    cw, ch = int(W * r), int(H * r)
    cx, cy = np.random.randint(W), np.random.randint(H)
    x1, y1 = np.clip(cx - cw // 2, 0, W), np.clip(cy - ch // 2, 0, H)
    x2, y2 = np.clip(cx + cw // 2, 0, W), np.clip(cy + ch // 2, 0, H)
    mixed = x.clone()
    mixed[:, :, x1:x2, y1:y2] = x[idx, :, x1:x2, y1:y2]
    return mixed, y, y[idx], 1 - (x2 - x1) * (y2 - y1) / (W * H)

class Trainer:
    def __init__(self, model, loaders, hparams):
        self.model = model
        self.loaders = loaders
        self.hparams = hparams
        self.criterion = SmoothCELoss(hparams['label_smoothing'])
        
        # Separate parameter groups
        bp = [p for n, p in model.named_parameters() if p.requires_grad and 'backbone' in n]
        hp = [p for n, p in model.named_parameters() if p.requires_grad and 'backbone' not in n]
        
        self.optimizer = optim.AdamW([
            {'params': bp, 'lr': hparams['lr_backbone']},
            {'params': hp, 'lr': hparams['lr_head']}
        ], weight_decay=hparams['weight_decay'])
        
        self.scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(
            self.optimizer, T_0=hparams['num_epochs'] // 3, T_mult=2, eta_min=1e-7
        )
        
        self.scaler = GradScaler()
        self.history = {k: [] for k in ['train_loss', 'val_loss', 'train_acc', 'val_acc', 'val_f1', 'lr']}
        self.best_val_f1 = 0
        self.best_state = None
        self.patience_counter = 0
        self.unfrozen = False

    def train_one_epoch(self, epoch):
        self.model.train()
        total_loss, preds, labs = 0, [], []
        
        for images, labels in tqdm(self.loaders['train'], desc=f'Ep {epoch+1}', leave=False):
            images = images.to(DEVICE, non_blocking=True)  # ✅ non_blocking
            labels = labels.to(DEVICE, non_blocking=True)
            
            self.optimizer.zero_grad(set_to_none=True)  # ✅ Faster than zero_grad()
            
            mixed = False
            if np.random.random() < self.hparams['mixup_prob']:
                if np.random.random() < 0.5:
                    images, la, lb, lam = mixup_data(images, labels, self.hparams['mixup_alpha'])
                else:
                    images, la, lb, lam = cutmix_data(images, labels, self.hparams['cutmix_alpha'])
                mixed = True
            
            with autocast():  # ✅ AMP
                out = self.model(images)
                if mixed:
                    loss = lam * self.criterion(out, la) + (1 - lam) * self.criterion(out, lb)
                else:
                    loss = self.criterion(out, labels)
            
            self.scaler.scale(loss).backward()
            self.scaler.unscale_(self.optimizer)
            torch.nn.utils.clip_grad_norm_(self.model.parameters(), max_norm=1.0)
            self.scaler.step(self.optimizer)
            self.scaler.update()
            
            total_loss += loss.item()
            if not mixed:
                preds.extend(out.argmax(1).cpu().numpy())
                labs.extend(labels.cpu().numpy())
        
        n_batches = len(self.loaders['train'])
        avg_loss = total_loss / n_batches
        acc = accuracy_score(labs, preds) if labs else 0
        f1 = f1_score(labs, preds, average='weighted', zero_division=0) if labs else 0
        return avg_loss, acc, f1

    @torch.no_grad()
    def validate(self, loader):
        self.model.eval()
        total_loss, preds, labs, probs = 0, [], [], []
        
        for images, labels in tqdm(loader, desc='Val', leave=False):
            images = images.to(DEVICE, non_blocking=True)
            labels = labels.to(DEVICE, non_blocking=True)
            
            with autocast():  # ✅ AMP for validation too
                out = self.model(images)
                loss = self.criterion(out, labels)
            
            total_loss += loss.item()
            p = F.softmax(out, dim=1)
            preds.extend(out.argmax(1).cpu().numpy())
            labs.extend(labels.cpu().numpy())
            probs.extend(p.cpu().numpy())
        
        n_batches = len(loader)
        avg_loss = total_loss / n_batches
        acc = accuracy_score(labs, preds)
        f1 = f1_score(labs, preds, average='weighted', zero_division=0)
        return avg_loss, acc, f1, 0, preds, labs, probs

    def train(self):
        print(f"\n{'='*60}")
        print(f"Starting training for {self.hparams['num_epochs']} epochs")
        print(f"Img size: {self.hparams['img_size']}, Batch: {self.hparams['batch_size']}")
        print(f"{'='*60}\n")
        
        for epoch in range(self.hparams['num_epochs']):
            # Unfreeze backbone after freeze_epochs
            if epoch == self.hparams['freeze_epochs'] and not self.unfrozen:
                print("\n🔄 Unfreezing backbone (last 3 blocks)")
                self.model.unfreeze_backbone_partial(3)
                self.unfrozen = True
            
            tl, ta, tf = self.train_one_epoch(epoch)
            vl, va, vf, _, _, _, _ = self.validate(self.loaders['val'])
            self.scheduler.step()
            
            self.history['train_loss'].append(tl)
            self.history['val_loss'].append(vl)
            self.history['train_acc'].append(ta)
            self.history['val_acc'].append(va)
            self.history['val_f1'].append(vf)
            self.history['lr'].append(self.optimizer.param_groups[0]['lr'])
            
            print(f"Ep {epoch+1:2d}: T_Loss={tl:.3f} T_Acc={ta:.3f} | V_Loss={vl:.3f} V_Acc={va:.3f} V_F1={vf:.3f} | LR={self.optimizer.param_groups[0]['lr']:.2e}")
            
            if vf > self.best_val_f1:
                self.best_val_f1 = vf
                self.best_state = copy.deepcopy(self.model.state_dict())
                self.patience_counter = 0
                torch.save({'model': self.model.state_dict(), 'f1': vf, 'acc': va}, CKDIR / 'best_model.pth')
                print(f"   ✅ New best F1: {vf:.4f}")
            else:
                self.patience_counter += 1
                if self.patience_counter >= self.hparams['patience']:
                    print(f"\n⏹️ Early stopping at epoch {epoch+1}")
                    break
        
        return self.history

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 10, 11, 12 — Plots, Heatmaps, Main (FIXED 384 → img_size)              ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def plot_history(h):
    fig, ax = plt.subplots(2, 2, figsize=(14, 10))
    ax[0,0].plot(h['train_loss'], label='Train'); ax[0,0].plot(h['val_loss'], label='Val')
    ax[0,0].set_title('Loss'); ax[0,0].legend(); ax[0,0].grid(True, alpha=0.3)
    ax[0,1].plot(h['train_acc'], label='Train'); ax[0,1].plot(h['val_acc'], label='Val')
    ax[0,1].set_title('Accuracy'); ax[0,1].legend(); ax[0,1].grid(True, alpha=0.3)
    ax[0,1].set_ylim(0, 1)
    ax[1,0].plot(h['val_f1'], label='Val F1', color='green')
    ax[1,0].set_title('Validation F1 Score'); ax[1,0].legend(); ax[1,0].grid(True, alpha=0.3)
    ax[1,0].set_ylim(0, 1)
    ax[1,1].plot(h['lr']); ax[1,1].set_title('Learning Rate'); ax[1,1].set_yscale('log'); ax[1,1].grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(PLTDIR / 'training_history.png')
    plt.show()

def plot_roc_curve(labels, probs, names):
    from sklearn.metrics import auc
    labels_bin = label_binarize(labels, classes=list(range(len(names))))
    fpr, tpr, roc_auc = {}, {}, {}
    for i in range(len(names)):
        fpr[i], tpr[i], _ = roc_curve(labels_bin[:, i], [p[i] for p in probs])
        roc_auc[i] = auc(fpr[i], tpr[i])
    
    fig, ax = plt.subplots(figsize=(10, 8))
    for i, name in enumerate(names):
        ax.plot(fpr[i], tpr[i], label=f'{name} (AUC = {roc_auc[i]:.3f})', linewidth=2)
    ax.plot([0, 1], [0, 1], 'k--', linewidth=1)
    ax.set_xlabel('False Positive Rate', fontsize=12)
    ax.set_ylabel('True Positive Rate', fontsize=12)
    ax.set_title('Multi-class ROC Curve', fontsize=14)
    ax.legend(loc='lower right')
    ax.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(PLTDIR / 'roc_curve.png')
    plt.show()

def isolate_tumor(heatmap, percentile=40):
    if heatmap is None: return None
    threshold = np.percentile(heatmap, percentile)
    masked = np.where(heatmap > threshold, heatmap, 0.0)
    if masked.max() > masked.min():
        return (masked - masked.min()) / (masked.max() - masked.min() + 1e-8)
    return np.zeros_like(heatmap)

class HeatmapGenerator:
    def __init__(self, model):
        self.model = model.to(DEVICE)
        self.model.eval()
        self.cnn_f, self.cnn_g = None, None
        self.img_size = getattr(model, 'img_size', 224)  # ✅ Get actual img_size
        
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        
        if last_conv:
            last_conv.register_forward_hook(lambda m, i, o: setattr(self, 'cnn_f', o.detach()))
            last_conv.register_backward_hook(lambda m, gi, go: setattr(self, 'cnn_g', (go[0] if isinstance(go, (tuple, list)) else go).detach()))
    
    def get_cnn(self, x, t):
        out = self.model(x)
        self.model.zero_grad()
        oh = torch.zeros_like(out)
        oh[0, t] = 1
        out.backward(gradient=oh, retain_graph=True)
        
        if self.cnn_f is None: return None
        w = self.cnn_g.mean(dim=(2, 3), keepdim=True)
        cam = F.relu((w * self.cnn_f).sum(1)).squeeze().cpu().numpy()
        
        if cam.max() > cam.min:
            return (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)
        return cam
    
    def get_attn(self, x):
        with torch.no_grad():
            self.model(x)
        g, _ = self.model.get_attentions()
        if not g: return None
        
        a = g[-1]
        a = a[0, 0, 1:].cpu().numpy() if a.dim() == 3 else a[0, :, 0, 1:].mean(0).cpu().numpy()
        a = a.reshape(self.model.spatial_h, self.model.spatial_w)
        
        if a.max() > a.min():
            return (a - a.min()) / (a.max() - a.min() + 1e-8)
        return a
    
    def draw(self, img_path, target, names, save_path):
        img = cv2.imread(str(img_path))
        rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        sz = rgb.shape[:2]
        
        # ✅ FIX: Use actual img_size instead of hardcoded 384
        t = torch.from_numpy(cv2.resize(rgb, (self.img_size, self.img_size))).float().permute(2, 0, 1) / 255.0
        t = ((t - torch.tensor(MEAN).view(3, 1, 1)) / torch.tensor(STD).view(3, 1, 1)).unsqueeze(0).to(DEVICE)
        
        c = self.get_cnn(t.clone(), target)
        a = self.get_attn(t.clone())
        comb = None
        
        if c is not None and a is not None:
            if a.shape != c.shape:
                a = zoom(a, (c.shape[0] / a.shape[0], c.shape[1] / a.shape[1]), order=1)
            comb = 0.5 * c + 0.5 * a
            if comb.max() > comb.min():
                comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8)
        
        c, a, comb = isolate_tumor(c, 40), isolate_tumor(a, 35), isolate_tumor(comb, 35)
        
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        axes[0].imshow(rgb)
        axes[0].set_title(f'True: {names[target]}')
        axes[0].axis('off')
        
        for i, (hmap, title) in enumerate(zip([c, a, comb], ['CNN', 'Attention', 'Combined'])):
            if hmap is not None:
                hr = zoom(hmap, (sz[0] / hmap.shape[0], sz[1] / hmap.shape[1]), order=1)
                colored = (plt.cm.jet(hr)[:, :, :3] * 255).astype(np.uint8)
                axes[i + 1].imshow(cv2.addWeighted(rgb, 0.6, colored, 0.4, 0))
            axes[i + 1].set_title(title)
            axes[i + 1].axis('off')
        
        plt.tight_layout()
        plt.savefig(save_path, dpi=150, bbox_inches='tight')
        plt.show()

def main():
    # ✅ Get img_size from HPARAMS
    IMG_SIZE = HPARAMS['img_size']
    BATCH_SIZE = HPARAMS['batch_size']
    NUM_WORKERS = HPARAMS['num_workers']
    
    print("📦 Loading Brain Tumor Dataset...")
    samples, class_names = load_brain_tumor()
    labels = [class_names.index(l) for l in [s[1] for s in samples]]
    
    train_s, temp_s = train_test_split(samples, test_size=0.3, stratify=labels, random_state=42)
    val_s, test_s = train_test_split(temp_s, test_size=0.5, 
                                      stratify=[class_names.index(l) for l in [s[1] for s in temp_s]], 
                                      random_state=42)
    
    print(f"\n📊 Dataset split: Train={len(train_s)}, Val={len(val_s)}, Test={len(test_s)}")
    print(f"📐 Image size: {IMG_SIZE}, Batch size: {BATCH_SIZE}, Workers: {NUM_WORKERS}")
    
    # ✅ FIX: Use IMG_SIZE from HPARAMS
    train_ds = MRIDataset(train_s, class_names, get_mri_transforms('train', IMG_SIZE, HPARAMS['aug_level']))
    val_ds = MRIDataset(val_s, class_names, get_mri_transforms('val', IMG_SIZE))
    test_ds = MRIDataset(test_s, class_names, get_mri_transforms('test', IMG_SIZE))
    
    sw = [train_ds.class_weights()[class_names.index(l)] for _, l in train_s]
    
    # ✅ FIX: Use BATCH_SIZE and NUM_WORKERS from HPARAMS
    loaders = {
        'train': DataLoader(train_ds, batch_size=BATCH_SIZE, 
                           sampler=WeightedRandomSampler(sw, len(sw), replacement=True),
                           num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=True),
        'val': DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, 
                         num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=True),
        'test': DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, 
                          num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=True)
    }
    
    print("🏗️ Building 4-Class Model...")
    # ✅ FIX: Pass all parameters from HPARAMS
    model = EnhancedHViT(
        num_classes=HPARAMS['num_classes'],
        img_size=IMG_SIZE,
        d_model=HPARAMS['d_model'],
        nhead=HPARAMS['nhead'],
        n_layers=HPARAMS['n_layers'],
        dim_ffn=HPARAMS['dim_ffn'],
        backbone_name=HPARAMS['backbone']
    ).to(DEVICE)
    
    # Print model size
    n_params = sum(p.numel() for p in model.parameters())
    n_trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"📊 Model params: {n_params/1e6:.1f}M (trainable: {n_trainable/1e6:.1f}M)")
    
    trainer = Trainer(model, loaders, HPARAMS)
    history = trainer.train()
    
    print("\n📊 Final Evaluation...")
    model.load_state_dict(torch.load(CKDIR / 'best_model.pth', map_location=DEVICE, weights_only=False)['model'])
    model.eval()
    
    all_p, all_l, all_pr = [], [], []
    with torch.no_grad():
        for im, l in tqdm(loaders['test'], desc='Final Test'):
            with autocast():
                out = model(im.to(DEVICE))
            p = F.softmax(out, dim=1)
            all_p.extend(out.argmax(1).cpu().numpy())
            all_l.extend(l.numpy())
            all_pr.extend(p.cpu().numpy())
    
    test_acc = accuracy_score(all_l, all_p)
    test_f1 = f1_score(all_l, all_p, average='weighted')
    cm = confusion_matrix(all_l, all_p)
    
    print(f"\n{'='*50}")
    print(f"🎯 FINAL TEST RESULTS")
    print(f"   Accuracy: {test_acc:.4f}")
    print(f"   F1 Score: {test_f1:.4f}")
    print(f"{'='*50}")
    print(classification_report(all_l, all_p, target_names=class_names))

    print("\n📈 Generating Graphs...")
    plot_history(history)
    
    fig, ax = plt.subplots(figsize=(10, 8))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=ax)
    ax.set_xlabel('Predicted'); ax.set_ylabel('True')
    plt.tight_layout()
    plt.savefig(PLTDIR / 'confusion_matrix.png')
    plt.show()
    
    plot_roc_curve(all_l, all_pr, class_names)
    
    print("\n🔥 Generating Tumor Heatmaps...")
    hm = HeatmapGenerator(model)
    for cls_idx in range(1, 4):
        corr = [i for i, (p, l) in enumerate(zip(all_p, all_l)) if p == l and l == cls_idx][:2]
        for idx in corr:
            hm.draw(test_s[idx][0], cls_idx, class_names, PLTDIR / f'heatmap_{class_names[cls_idx]}_{idx}.png')
    
    print("\n✅ All done!")

if __name__ == '__main__':
    main()
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 13 — Upload Image & Predict (Interactive)                                ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

import ipywidgets as widgets
from IPython.display import display, clear_output, FileLink
import io
from PIL import Image as PILImage

# Ensure model is loaded and ready
model.load_state_dict(torch.load(CKDIR /'best_model.pth', map_location=DEVICE, weights_only=False)['model'])
model.eval()

CLASS_NAMES = HPARAMS['class_names']
IMG_SIZE = HPARAMS['img_size']

def predict_single_image(image_array):
    """Run inference on a single numpy RGB image."""
    # Resize
    img_resized = cv2.resize(image_array, (IMG_SIZE, IMG_SIZE))
    
    # Convert to tensor with normalization
    img_tensor = torch.from_numpy(img_resized).float().permute(2, 0, 1) / 255.0
    img_tensor = (img_tensor - torch.tensor(MEAN).view(3, 1, 1)) / torch.tensor(STD).view(3, 1, 1)
    img_tensor = img_tensor.unsqueeze(0).to(DEVICE)
    
    # Predict
    with torch.no_grad():
        with autocast():
            logits = model(img_tensor)
            probs = F.softmax(logits, dim=1)
            conf, pred_idx = probs.max(dim=1)
    
    return pred_idx.item(), conf.item(), probs.cpu().numpy()[0]

def generate_prediction_plot(image_rgb, pred_idx, confidence, probs):
    """Create a nice prediction visualization."""
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    
    # Left: Image with prediction
    axes[0].imshow(image_rgb)
    color = 'green' if confidence > 0.8 else ('orange' if confidence > 0.5 else 'red')
    axes[0].set_title(f"Prediction: {CLASS_NAMES[pred_idx]}\nConfidence: {confidence:.2%}", 
                      fontsize=14, fontweight='bold', color=color)
    axes[0].axis('off')
    
    # Right: Bar chart of all class probabilities
    colors = ['#ff6b6b' if i == pred_idx else '#4ecdc4' for i in range(len(CLASS_NAMES))]
    bars = axes[1].barh(CLASS_NAMES, probs * 100, color=colors, edgecolor='black', linewidth=0.5)
    axes[1].set_xlabel('Probability (%)', fontsize=12)
    axes[1].set_title('Class Probabilities', fontsize=14, fontweight='bold')
    axes[1].set_xlim(0, 105)
    
    # Add percentage labels on bars
    for bar, prob in zip(bars, probs):
        axes[1].text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2, 
                     f'{prob:.1f}%', va='center', fontsize=11, fontweight='bold')
    
    axes[1].grid(axis='x', alpha=0.3)
    plt.tight_layout()
    return fig

# ─── Widget Setup ───────────────────────────────────────────────────────────────

upload_widget = widgets.FileUpload(
    accept='.jpg,.jpeg,.png,.bmp,.tiff',
    multiple=False,
    description='📤 Upload MRI Image',
    button_style='info',
    style={'button_width': '300px', 'description_width': '200px'}
)

path_input = widgets.Text(
    value='',
    placeholder='/kaggle/input/.../image.jpg',
    description='📁 Or enter path:',
    layout={'width': '600px'}
)

predict_button = widgets.Button(
    description='🔍 Predict',
    button_style='success',
    layout={'width': '200px'}
)

output_area = widgets.Output()

def on_predict_clicked(b):
    with output_area:
        clear_output(wait=True)
        
        image_rgb = None
        source = ""
        
        # Try file upload first
        if upload_widget.value:
            uploaded_file = list(upload_widget.value.values())[0]
            content = uploaded_file['content']
            image_rgb = np.array(PILImage.open(io.BytesIO(content)).convert('RGB'))
            source = f"Uploaded: {uploaded_file['metadata']['name']}"
        
        # Try path input
        elif path_input.value.strip():
            p = Path(path_input.value.strip())
            if p.exists():
                img = cv2.imread(str(p))
                if img is not None:
                    image_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    source = f"Path: {p.name}"
            else:
                print(f"❌ File not found: {p}")
                return
        
        else:
            print("⚠️ Please upload an image OR enter a file path")
            return
        
        if image_rgb is None:
            print("❌ Could not read image")
            return
        
        # Run prediction
        print(f"🔮 Analyzing {source}...")
        pred_idx, confidence, probs = predict_single_image(image_rgb)
        
        # Display results
        print(f"\n{'='*50}")
        print(f"   PREDICTION: {CLASS_NAMES[pred_idx].upper()}")
        print(f"   CONFIDENCE: {confidence:.2%}")
        print(f"{'='*50}")
        print("\nAll probabilities:")
        for name, prob in zip(CLASS_NAMES, probs):
            marker = " ◀" if name == CLASS_NAMES[pred_idx] else ""
            print(f"   {name:15s}: {prob:.4f} ({prob*100:.1f}%){marker}")
        
        # Show plot
        fig = generate_prediction_plot(image_rgb, pred_idx, confidence, probs)
        plt.show()
        
        # Save prediction plot
        pred_path = PLTDIR / f'prediction_{CLASS_NAMES[pred_idx]}_{confidence:.0%}.png'
        fig.savefig(pred_path, dpi=150, bbox_inches='tight')
        print(f"\n💾 Saved prediction plot to: {pred_path}")

predict_button.on_click(on_predict_clicked)

# Display the interface
print("╔══════════════════════════════════════════════════════════════╗")
print("║        🧠 BRAIN TUMOR CLASSIFICATION - INFERENCE           ║")
print("╠══════════════════════════════════════════════════════════════╣")
print("║  Upload an MRI image or enter a file path, then click       ║")
print("║  'Predict' to see the classification result.                ║")
print("╚══════════════════════════════════════════════════════════════╝\n")

display(widgets.VBox([
    upload_widget,
    widgets.Label("── OR ──"),
    path_input,
    widgets.Label(""),
    predict_button,
    widgets.Label(""),
    output_area
]))


# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 14 — Batch Predict on Test Folder                                         ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def batch_predict(folder_path, save_csv=True):
    """Predict on all images in a folder and optionally save results to CSV."""
    folder = Path(folder_path)
    if not folder.exists():
        print(f"❌ Folder not found: {folder}")
        return None
    
    image_files = []
    for ext in VALID_EXTS:
        image_files.extend(folder.rglob(f'*{ext}'))
    
    if not image_files:
        print(f"❌ No images found in {folder}")
        return None
    
    print(f"📁 Found {len(image_files)} images in {folder}")
    
    results = []
    correct = 0
    total = 0
    
    for img_path in tqdm(image_files, desc='Predicting'):
        # Try to infer true label from folder name
        true_label = None
        for cls_name in CLASS_NAMES:
            if cls_name in img_path.parent.name.lower().replace('_', ''):
                true_label = cls_name
                break
        
        # Read and predict
        img = cv2.imread(str(img_path))
        if img is None:
            continue
        img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        
        pred_idx, confidence, probs = predict_single_image(img_rgb)
        pred_label = CLASS_NAMES[pred_idx]
        
        if true_label is not None:
            total += 1
            if true_label == pred_label:
                correct += 1
        
        results.append({
            'path': str(img_path),
            'filename': img_path.name,
            'true_label': true_label,
            'predicted': pred_label,
            'confidence': confidence,
            **{f'prob_{cls}': p for cls, p in zip(CLASS_NAMES, probs)}
        })
    
    df = pd.DataFrame(results)
    
    if total > 0:
        acc = correct / total
        print(f"\n{'='*50}")
        print(f"📊 BATCH RESULTS")
        print(f"   Total: {total} images with known labels")
        print(f"   Correct: {correct}")
        print(f"   Accuracy: {acc:.2%}")
        print(f"{'='*50}")
        
        # Per-class accuracy
        print("\nPer-class accuracy:")
        for cls in CLASS_NAMES:
            cls_df = df[df['true_label'] == cls]
            if len(cls_df) > 0:
                cls_acc = (cls_df['true_label'] == cls_df['predicted']).mean()
                print(f"   {cls:15s}: {cls_acc:.2%} ({len(cls_df)} images)")
    
    if save_csv:
        csv_path = OUTDIR / 'predictions.csv'
        df.to_csv(csv_path, index=False)
        print(f"\n💾 Saved predictions to: {csv_path}")
    
    return df

# Example usage (uncomment and change path):
# results_df = batch_predict('/kaggle/input/datasets/aayush4204/daataaset/Brain Tumor MRI Dataset/Brain Tumor MRI Dataset/Testing')

```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  STANDALONE HEATMAP GENERATOR                         ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

import torch, torch.nn as nn, torch.nn.functional as F
import numpy as np, cv2, matplotlib.pyplot as plt
from pathlib import Path
from einops import rearrange
from scipy.ndimage import zoom
import torchvision.models as models
import warnings
warnings.filterwarnings('ignore')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
CKDIR = OUTDIR / 'checkpoints'
PLTDIR = OUTDIR / 'plots'
PLTDIR.mkdir(exist_ok=True)

MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)
class_names = ['notumor', 'glioma', 'meningioma', 'pituitary']

# --- 1. Rebuild Model Architecture (Must match exactly) ---
class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): super().__init__(); self.drop_prob = drop_prob
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training: return x
        keep_prob = float(1.0 - self.drop_prob)
        return x * (torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * keep_prob)

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3); self.proj = nn.Linear(d_model, d_model)
        self.attn_drop, self.proj_drop, self.attn_weights = nn.Dropout(dropout), nn.Dropout(dropout), None
    def forward(self, x, H, W):
        B, N, C = x.shape; x = x.reshape(B, H, W, C)
        pad_h, pad_w = (self.window_size - H % self.window_size) % self.window_size, (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0: x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(-1, self.window_size**2, C)
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2,0,3,1,4)
        attn = (qkv[0] @ qkv[1].transpose(-2,-1)) * self.scale; attn = attn.softmax(dim=-1); self.attn_weights = attn.detach()
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1,2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(B, Hp, Wp, C)
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model*2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop), nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop))
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights, self.window_attn_weights = None, None
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        window_out = self.window_attn(x_norm[:, 1:, :], H, W)
        self.window_attn_weights = self.window_attn.attn_weights
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1-gate) * window_out], dim=1)
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self): super().__init__(); self.backbone = models.convnext_base(weights=None); self.feature_extractor = self.backbone.features
    def forward(self, x): return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes=4, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536):
        super().__init__(); self.d_model = d_model; self.img_size = img_size; self.backbone = ImprovedBackbone()
        with torch.no_grad(): _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        self.spatial_h, self.spatial_w = h, w
        self.proj = nn.Sequential(nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(), nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU())
        self.norm_proj = nn.LayerNorm(d_model); self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h*w + 1, d_model))
        nn.init.trunc_normal_(self.cls_token, std=0.02); nn.init.trunc_normal_(self.pos_embed, std=0.02)
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i]) for i in range(n_layers)])
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, d_model//2), nn.GELU(), nn.Dropout(0.3), nn.Linear(d_model//2, d_model//4), nn.GELU(), nn.Dropout(0.15), nn.Linear(d_model//4, num_classes))
    def forward(self, x):
        B = x.size(0); tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens); tokens = torch.cat([self.cls_token.expand(B,-1,-1), tokens], dim=1) + self.pos_embed
        for block in self.encoder: tokens = block(tokens, self.spatial_h, self.spatial_w)
        return self.head(self.norm(tokens)[:, 0])
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w

# --- 2. Load Model ---
print("📦 Rebuilding model and loading weights...")
model = EnhancedHViT(num_classes=4, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536).to(DEVICE)
model.load_state_dict(torch.load(CKDIR/'best_model.pth', map_location=DEVICE, weights_only=False)['model'])
model.eval()
print("✅ Model loaded successfully!")

# --- 3. Heatmap Logic (Fixed) ---
def isolate_tumor(heatmap, percentile=40):
    if heatmap is None: return None
    threshold = np.percentile(heatmap, percentile)
    masked = np.where(heatmap > threshold, heatmap, 0.0)
    return (masked - masked.min()) / (masked.max() - masked.min() + 1e-8) if masked.max() > masked.min() else np.zeros_like(heatmap)

class HeatmapGenerator:
    def __init__(self, model):
        self.model = model.to(DEVICE); self.model.eval(); self.cnn_f, self.cnn_g = None, None
        self.img_size = getattr(model, 'img_size', 224)
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        if last_conv:
            last_conv.register_forward_hook(lambda m,i,o: setattr(self, 'cnn_f', o.detach()))
            last_conv.register_backward_hook(lambda m,gi,go: setattr(self, 'cnn_g', (go[0] if isinstance(go, (tuple, list)) else go).detach()))
    def get_cnn(self, x, t):
        out = self.model(x); self.model.zero_grad(); oh=torch.zeros_like(out); oh[0,t]=1; out.backward(gradient=oh, retain_graph=True)
        if self.cnn_f is None: return None
        w = self.cnn_g.mean(dim=(2, 3), keepdim=True); cam = F.relu((w * self.cnn_f).sum(1)).squeeze().cpu().numpy()
        if cam.max() > cam.min(): return (cam - cam.min()) / (cam.max() - cam.min() + 1e-8) # FIXED ()
        return cam
    def get_attn(self, x):
        with torch.no_grad(): self.model(x)
        g, _ = self.model.get_attentions()
        if not g: return None
        a = g[-1]
        a = a[0, 0, 1:].cpu().numpy() if a.dim()==3 else a[0, :, 0, 1:].mean(0).cpu().numpy()
        a = a.reshape(self.model.spatial_h, self.model.spatial_w)
        if a.max() > a.min(): return (a - a.min()) / (a.max() - a.min() + 1e-8) # FIXED ()
        return a
    def draw(self, img_path, target, names, save_path):
        img = cv2.imread(str(img_path)); rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB); sz = rgb.shape[:2]
        t = torch.from_numpy(cv2.resize(rgb, (self.img_size, self.img_size))).float().permute(2,0,1)/255.0
        t = ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)
        c, a, comb = self.get_cnn(t.clone(), target), self.get_attn(t.clone()), None
        if c is not None and a is not None:
            if a.shape != c.shape: a = zoom(a, (c.shape[0]/a.shape[0], c.shape[1]/a.shape[1]), order=1)
            comb = 0.5*c + 0.5*a
            if comb.max() > comb.min(): comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8)
        c, a, comb = isolate_tumor(c, 40), isolate_tumor(a, 35), isolate_tumor(comb, 35)
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        axes[0].imshow(rgb); axes[0].set_title(f'True: {names[target]}'); axes[0].axis('off')
        for i, (hmap, title) in enumerate(zip([c, a, comb], ['CNN', 'Attention', 'Combined'])):
            if hmap is not None:
                hr = zoom(hmap, (sz[0]/hmap.shape[0], sz[1]/hmap.shape[1]), order=1)
                colored = (plt.cm.jet(hr)[:,:,:3]*255).astype(np.uint8)
                axes[i+1].imshow(cv2.addWeighted(rgb, 0.6, colored, 0.4, 0))
            axes[i+1].set_title(title); axes[i+1].axis('off')
        plt.tight_layout(); plt.savefig(save_path, dpi=150, bbox_inches='tight'); plt.show()

# --- 4. Grab Test Samples & Generate ---
from sklearn.model_selection import train_test_split
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.JPG', '.JPEG', '.PNG'}

# Quick reload test split to find correct images
print("📂 Reloading test set paths...")
samples = []
base = Path('/kaggle/input/datasets/aayush4204/daataaset/Brain Tumor MRI Dataset/Brain Tumor MRI Dataset')
for split in ['Training', 'Testing']:
    split_path = base / split
    if not split_path.exists(): continue
    for f in split_path.iterdir():
        if not f.is_dir(): continue
        fn = f.name.lower().replace('_', '')
        if 'notumor' in fn or 'no' in fn: cname = 'notumor'
        elif 'glioma' in fn: cname = 'glioma'
        elif 'meningioma' in fn: cname = 'meningioma'
        elif 'pituitary' in fn: cname = 'pituitary'
        else: continue
        for ext in VALID_EXTS:
            for p in f.rglob(f'*{ext}'): samples.append((str(p), cname))

labels = [class_names.index(l) for l in [s[1] for s in samples]]
_, temp_s = train_test_split(samples, test_size=0.3, stratify=labels, random_state=42)
_, test_s = train_test_split(temp_s, test_size=0.5, stratify=[class_names.index(l) for l in [s[1] for s in temp_s]], random_state=42)

# Run model on test set to find correct predictions
print("🔍 Finding correct predictions for heatmap...")
all_p, all_l = [], []
hm = HeatmapGenerator(model)

for p, l in test_s:
    img = cv2.imread(p)
    if img is None: continue
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    t = torch.from_numpy(cv2.resize(rgb, (224, 224))).float().permute(2,0,1)/255.0
    t = ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)
    with torch.no_grad(): out = model(t)
    all_p.append(out.argmax(1).item())
    all_l.append(class_names.index(l))

print("\n🔥 Generating Tumor Heatmaps...")
for cls_idx in range(1, 4):
    corr = [i for i,(p,l) in enumerate(zip(all_p, all_l)) if p==l and l==cls_idx][:2]
    for idx in corr: 
        hm.draw(test_s[idx][0], cls_idx, class_names, PLTDIR/f'heatmap_{class_names[cls_idx]}_{idx}.png')

print("\n✅ All heatmaps generated successfully!")
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════════╗
# ║  ULTIMATE MEGA CELL: ALL 20 PLOTS (100% Standalone, PyTorch Shape Fixed)           ║
# ╚══════════════════════════════════════════════════════════════════════════════════════╝

import os, time, random, copy, warnings
from pathlib import Path
from collections import Counter
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns, cv2
from tqdm.auto import tqdm
from scipy.ndimage import zoom
import torch, torch.nn as nn, torch.nn.functional as F
import torchvision.models as models
from sklearn.metrics import (accuracy_score, f1_score, roc_curve, auc, confusion_matrix, 
                             classification_report, precision_recall_curve, average_precision_score)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE
from torch.cuda.amp import autocast
from einops import rearrange

warnings.filterwarnings('ignore')
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
PLTDIR = OUTDIR / 'plots'; PLTDIR.mkdir(exist_ok=True)
CKDIR = OUTDIR / 'checkpoints'
MEAN, STD = (0.485, 0.456, 0.406), (0.229, 0.224, 0.225)
class_names = ['notumor', 'glioma', 'meningioma', 'pituitary']
colors = ['#2ecc71', '#e74c3c', '#3498db', '#f39c12']
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.JPG', '.JPEG', '.PNG'}

# ─── 1. MODEL ARCHITECTURE ────────────────────────────────────────────────────────
class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): super().__init__(); self.drop_prob = drop_prob
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training: return x
        return x * (torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * float(1.0 - self.drop_prob))

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3); self.proj = nn.Linear(d_model, d_model)
        self.attn_drop, self.proj_drop, self.attn_weights = nn.Dropout(dropout), nn.Dropout(dropout), None
    def forward(self, x, H, W):
        B, N, C = x.shape; x = x.reshape(B, H, W, C)
        pad_h, pad_w = (self.window_size - H % self.window_size) % self.window_size, (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0: x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(-1, self.window_size**2, C)
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2,0,3,1,4)
        attn = (qkv[0] @ qkv[1].transpose(-2,-1)) * self.scale; attn = attn.softmax(dim=-1); self.attn_weights = attn.detach()
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1,2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(B, Hp, Wp, C)
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model*2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop), nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop))
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights, self.window_attn_weights = None, None
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        window_out = self.window_attn(x_norm[:, 1:, :], H, W); self.window_attn_weights = self.window_attn.attn_weights
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1-gate) * window_out], dim=1)
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self): super().__init__(); self.backbone = models.convnext_base(weights=None); self.feature_extractor = self.backbone.features
    def forward(self, x): return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes=4, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536):
        super().__init__(); self.d_model = d_model; self.img_size = img_size; self.backbone = ImprovedBackbone()
        with torch.no_grad(): _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        self.spatial_h, self.spatial_w = h, w
        self.proj = nn.Sequential(nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(), nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU())
        self.norm_proj = nn.LayerNorm(d_model); self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h*w + 1, d_model))
        nn.init.trunc_normal_(self.cls_token, std=0.02); nn.init.trunc_normal_(self.pos_embed, std=0.02)
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i]) for i in range(n_layers)])
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, d_model//2), nn.GELU(), nn.Dropout(0.3), nn.Linear(d_model//2, d_model//4), nn.GELU(), nn.Dropout(0.15), nn.Linear(d_model//4, num_classes))
    def forward(self, x):
        B = x.size(0); tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens); tokens = torch.cat([self.cls_token.expand(B,-1,-1), tokens], dim=1) + self.pos_embed
        for block in self.encoder: tokens = block(tokens, self.spatial_h, self.spatial_w)
        return self.head(self.norm(tokens)[:, 0])
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w

# ─── 2. LOAD WEIGHTS & SPLIT DATA ────────────────────────────────────────────────
print("📦 Rebuilding model and loading weights...")
model = EnhancedHViT(num_classes=4, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536).to(DEVICE)
model.load_state_dict(torch.load(CKDIR/'best_model.pth', map_location=DEVICE, weights_only=False)['model'])
model.eval()

print("📂 Scanning dataset and creating splits...")
samples = []
base = Path('/kaggle/input/datasets/aayush4204/daataaset/Brain Tumor MRI Dataset/Brain Tumor MRI Dataset')
for split in ['Training', 'Testing']:
    split_path = base / split
    if not split_path.exists(): continue
    for f in split_path.iterdir():
        if not f.is_dir(): continue
        fn = f.name.lower().replace('_', '')
        if 'notumor' in fn or 'no' in fn: cname = 'notumor'
        elif 'glioma' in fn: cname = 'glioma'
        elif 'meningioma' in fn: cname = 'meningioma'
        elif 'pituitary' in fn: cname = 'pituitary'
        else: continue
        for ext in VALID_EXTS:
            for p in f.rglob(f'*{ext}'): samples.append((str(p), cname))

labels = [class_names.index(l) for l in [s[1] for s in samples]]
train_s, temp_s = train_test_split(samples, test_size=0.3, stratify=labels, random_state=42)
val_s, test_s = train_test_split(temp_s, test_size=0.5, stratify=[class_names.index(l) for l in [s[1] for s in temp_s]], random_state=42)
df_all = pd.DataFrame(samples, columns=['path', 'label'])

def get_tensor(img_rgb):
    t = torch.from_numpy(cv2.resize(img_rgb, (224, 224))).float().permute(2,0,1)/255.0
    return ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)

print("\n" + "="*60)
print("📊 COMPREHENSIVE DATASET ANALYSIS REPORT")
print("="*60)
print(f"\n🔹 Total Images: {len(df_all)} | Classes: {len(class_names)}")
print(f"🔹 Split: 70% Train ({len(train_s)}) / 15% Val ({len(val_s)}) / 15% Test ({len(test_s)})")
for cls in class_names:
    count = len(df_all[df_all['label'] == cls])
    print(f"   - {cls.capitalize():12s}: {count:5d} images ({count/len(df_all)*100:.1f}%)")
print("="*60 + "\n")

# ─── 3. RUN INFERENCE ───────────────────────────────────────────────────────────
print("🧠 Running inference on Test Set...")
all_p, all_l, all_pr = [], [], []
for p, l in tqdm(test_s, desc="Predicting"):
    img = cv2.imread(p)
    if img is None: continue
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    with torch.no_grad(): out = model(get_tensor(rgb)); pr = F.softmax(out, dim=1)
    all_p.append(out.argmax(1).item()); all_l.append(class_names.index(l)); all_pr.append(pr.cpu().numpy()[0])

all_p, all_l, all_pr = np.array(all_p), np.array(all_l), np.array(all_pr)
max_probs = all_pr.max(axis=1); correct_mask = (all_p == all_l)
labels_bin = label_binarize(all_l, classes=list(range(len(class_names))))

# ─── PLOTS 1 to 16 ──────────────────────────────────────────────────────────────
try:
    if len(history['train_loss']) > 0:
        fig, ax = plt.subplots(2, 2, figsize=(14, 10)); fig.suptitle('Model Training Dynamics', fontsize=16, fontweight='bold', y=1.02)
        ax[0,0].plot(history['train_loss'], label='Train', color='#e74c3c', lw=2); ax[0,0].plot(history['val_loss'], label='Val', color='#3498db', lw=2); ax[0,0].set_title('Loss'); ax[0,0].legend(); ax[0,0].grid(True, alpha=0.3)
        ax[0,1].plot(history['train_acc'], label='Train', color='#e74c3c', lw=2); ax[0,1].plot(history['val_acc'], label='Val', color='#3498db', lw=2); ax[0,1].set_title('Accuracy'); ax[0,1].set_ylim(0,1.05); ax[0,1].legend(); ax[0,1].grid(True, alpha=0.3)
        ax[1,0].plot(history['val_f1'], label='Val F1', color='#2ecc71', lw=2); ax[1,0].set_title('F1 Score'); ax[1,0].set_ylim(0,1.05); ax[1,0].legend(); ax[1,0].grid(True, alpha=0.3)
        ax[1,1].plot(history['lr'], color='#9b59b6', lw=2); ax[1,1].set_title('Learning Rate'); ax[1,1].set_yscale('log'); ax[1,1].grid(True, alpha=0.3)
        plt.tight_layout(); plt.savefig(PLTDIR / '1_training_history.png', dpi=300, bbox_inches='tight'); plt.show()
except: pass

cm = confusion_matrix(all_l, all_p); cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
fig, axes = plt.subplots(1, 2, figsize=(16, 6)); fig.suptitle('Confusion Matrix Analysis', fontsize=16, fontweight='bold')
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[0], cbar=False); axes[0].set_title('Absolute Counts'); axes[0].set_ylabel('True'); axes[0].set_xlabel('Predicted')
sns.heatmap(cm_norm, annot=True, fmt='.2%', cmap='Greens', xticklabels=class_names, yticklabels=class_names, ax=axes[1], cbar=False); axes[1].set_title('Normalized (Recall)'); axes[1].set_ylabel('True'); axes[1].set_xlabel('Predicted')
plt.tight_layout(); plt.savefig(PLTDIR / '2_confusion_matrices.png', dpi=300, bbox_inches='tight'); plt.show()

fpr, tpr, roc_auc = {}, {}, {}
fig, ax = plt.subplots(figsize=(10, 8))
for i in range(len(class_names)):
    fpr[i], tpr[i], _ = roc_curve(labels_bin[:, i], all_pr[:, i]); roc_auc[i] = auc(fpr[i], tpr[i])
    ax.plot(fpr[i], tpr[i], color=colors[i], lw=2, label=f'{class_names[i]} (AUC = {roc_auc[i]:.3f})')
all_fpr = np.unique(np.concatenate([fpr[i] for i in range(len(class_names))]))
mean_tpr = np.zeros_like(all_fpr)
for i in range(len(class_names)): mean_tpr += np.interp(all_fpr, fpr[i], tpr[i])
mean_tpr /= len(class_names)
ax.plot(all_fpr, mean_tpr, color='black', lw=3, linestyle='--', label=f'Macro-Avg (AUC = {auc(all_fpr, mean_tpr):.3f})')
ax.plot([0, 1], [0, 1], 'k:', lw=1, alpha=0.5); ax.set_title('Multi-class ROC Curves', fontsize=14, fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '3_roc_curves.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(10, 8))
for i in range(len(class_names)):
    prec, rec, _ = precision_recall_curve(labels_bin[:, i], all_pr[:, i])
    ax.plot(rec, prec, color=colors[i], lw=2, label=f'{class_names[i]} (AP = {average_precision_score(labels_bin[:, i], all_pr[:, i]):.3f})')
ax.set_title('Precision-Recall Curves', fontsize=14, fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3); ax.set_xlim([0,1]); ax.set_ylim([0,1.05])
plt.tight_layout(); plt.savefig(PLTDIR / '4_precision_recall.png', dpi=300, bbox_inches='tight'); plt.show()

report = classification_report(all_l, all_p, target_names=class_names, output_dict=True)
metrics_df = pd.DataFrame(report).T.iloc[:-3, :3]
fig, ax = plt.subplots(figsize=(10, 6)); x = np.arange(len(class_names)); width = 0.25
bars1 = ax.bar(x - width, metrics_df['precision'], width, label='Precision', color='#3498db', edgecolor='black')
bars2 = ax.bar(x, metrics_df['recall'], width, label='Recall', color='#2ecc71', edgecolor='black')
bars3 = ax.bar(x + width, metrics_df['f1-score'], width, label='F1-Score', color='#e74c3c', edgecolor='black')
ax.set_title('Per-Class Performance Metrics', fontsize=14, fontweight='bold'); ax.set_xticks(x); ax.set_xticklabels([n.capitalize() for n in class_names]); ax.legend(); ax.set_ylim(0,1.15); ax.grid(axis='y', alpha=0.3)
for bars in [bars1, bars2, bars3]:
    for bar in bars: ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.01, f'{bar.get_height():.2f}', ha='center', va='bottom', fontsize=9, fontweight='bold')
plt.tight_layout(); plt.savefig(PLTDIR / '5_per_class_metrics.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(max_probs[correct_mask], bins=20, alpha=0.7, label='Correct', color='#2ecc71', edgecolor='black')
ax.hist(max_probs[~correct_mask], bins=20, alpha=0.7, label='Incorrect', color='#e74c3c', edgecolor='black')
ax.set_title('Model Confidence Distribution', fontsize=14, fontweight='bold'); ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '6_confidence_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

def calibration_curve(y_true, y_prob, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1); accs, confs = [], []
    for i in range(n_bins):
        mask = (y_prob > bins[i]) & (y_prob <= bins[i+1])
        if i == n_bins - 1: mask = (y_prob >= bins[i]) & (y_prob <= bins[i+1])
        if mask.sum() > 0: accs.append(y_true[mask].mean()); confs.append(y_prob[mask].mean())
    return np.array(confs), np.array(accs)

fig, ax = plt.subplots(figsize=(8, 8))
for i, cls in enumerate(class_names):
    conf, acc = calibration_curve((all_l == i).astype(int), all_pr[:, i])
    ax.plot(conf, acc, marker='o', color=colors[i], lw=2, label=cls.capitalize(), markersize=5)
ax.plot([0, 1], [0, 1], 'k--', lw=1, label='Perfectly Calibrated'); ax.set_title('Reliability Diagram', fontsize=14, fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3); ax.set_xlim([0,1]); ax.set_ylim([0,1])
plt.tight_layout(); plt.savefig(PLTDIR / '7_calibration_curve.png', dpi=300, bbox_inches='tight'); plt.show()

correct_idxs, incorrect_idxs = np.where(correct_mask)[0], np.where(~correct_mask)[0]
np.random.seed(42)
samp_corr = np.random.choice(correct_idxs, min(9, len(correct_idxs)), replace=False)
samp_incorr = np.random.choice(incorrect_idxs, min(9, len(incorrect_idxs)), replace=False)
fig, axes = plt.subplots(2, 9, figsize=(25, 6))
for i, idx in enumerate(samp_corr):
    img = cv2.cvtColor(cv2.imread(test_s[idx][0]), cv2.COLOR_BGR2RGB)
    axes[0, i].imshow(img); axes[0, i].set_title(f"T:{class_names[all_l[idx]].split()[0]}\nP:{class_names[all_p[idx]].split()[0]}\n{max_probs[idx]:.1%}", fontsize=8, color='green'); axes[0, i].axis('off')
for i, idx in enumerate(samp_incorr):
    img = cv2.cvtColor(cv2.imread(test_s[idx][0]), cv2.COLOR_BGR2RGB)
    axes[1, i].imshow(img); axes[1, i].set_title(f"T:{class_names[all_l[idx]].split()[0]}\nP:{class_names[all_p[idx]].split()[0]}\n{max_probs[idx]:.1%}", fontsize=8, color='red'); axes[1, i].axis('off')
axes[0, 0].set_ylabel('CORRECT', fontsize=12, fontweight='bold', color='green', rotation=0, labelpad=60, va='center')
axes[1, 0].set_ylabel('INCORRECT', fontsize=12, fontweight='bold', color='red', rotation=0, labelpad=60, va='center')
fig.suptitle('Visual Inspection (T=True, P=Predicted)', fontsize=14, fontweight='bold', y=1.05); plt.tight_layout(); plt.savefig(PLTDIR / '8_sample_predictions.png', dpi=300, bbox_inches='tight'); plt.show()

plot_data = [{'True Class': class_names[i].capitalize(), 'Assigned Probability': prob} for i in range(len(class_names)) for prob in all_pr[all_l == i, i]]
df_probs = pd.DataFrame(plot_data)
fig, ax = plt.subplots(figsize=(10, 6))
sns.violinplot(data=df_probs, x='True Class', y='Assigned Probability', palette=colors, inner='quartile', ax=ax)
ax.set_title('Probability Density for True Class', fontsize=14, fontweight='bold'); ax.set_ylim(0,1.05); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '9_probability_violin.png', dpi=300, bbox_inches='tight'); plt.show()

print("⏳ Extracting t-SNE features...")
feature_tensor = None
def hook_fn(module, inp, out): global feature_tensor; feature_tensor = out
handle = model.norm.register_forward_hook(hook_fn)
tsne_indices = np.random.choice(len(test_s), min(600, len(test_s)), replace=False)
features_list, labels_tsne = [], []
for idx in tsne_indices:
    p, l = test_s[idx]; img = cv2.imread(p)
    if img is None: continue
    rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    with torch.no_grad(): _ = model(get_tensor(rgb))
    features_list.append(feature_tensor[:, 0, :].cpu().numpy()[0]); labels_tsne.append(class_names.index(l))
handle.remove()
tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000)
tsne_results = tsne.fit_transform(np.array(features_list))
fig, ax = plt.subplots(figsize=(10, 8))
for i, cls in enumerate(class_names):
    mask = np.array(labels_tsne) == i
    ax.scatter(tsne_results[mask, 0], tsne_results[mask, 1], c=colors[i], label=cls.capitalize(), alpha=0.6, s=20, edgecolors='white', linewidth=0.5)
ax.set_title('t-SNE Visualization of Learned Features', fontsize=14, fontweight='bold'); ax.legend(markerscale=2); ax.grid(True, alpha=0.2); ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.savefig(PLTDIR / '10_tsne_features.png', dpi=300, bbox_inches='tight'); plt.show()

errors = [{'True': class_names[all_l[i]], 'Predicted': class_names[all_p[i]]} for i in range(len(all_l)) if all_l[i] != all_p[i]]
if errors:
    error_df = pd.DataFrame(errors); error_counts = error_df.groupby(['True', 'Predicted']).size().reset_index(name='Count').sort_values(by='Count', ascending=False)
    fig, ax = plt.subplots(figsize=(10, 6)); error_counts['Pair'] = error_counts['True'] + ' → ' + error_counts['Predicted']
    sns.barplot(data=error_counts, x='Count', y='Pair', palette='Reds_r', ax=ax); ax.set_title('Top Misclassification Pairs', fontsize=14, fontweight='bold')
    for i, v in enumerate(error_counts['Count']): ax.text(v + 0.2, i, str(v), color='black', va='center', fontweight='bold')
    plt.tight_layout(); plt.savefig(PLTDIR / '11_error_analysis.png', dpi=300, bbox_inches='tight'); plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 6)); fig.suptitle('Overall Dataset Class Distribution', fontsize=16, fontweight='bold')
counts = df_all['label'].value_counts().reindex(class_names)
sns.barplot(x=counts.index, y=counts.values, ax=axes[0], palette=colors, edgecolor='black'); axes[0].set_title('Image Count per Class', fontsize=13)
for i, v in enumerate(counts.values): axes[0].text(i, v + 30, str(v), ha='center', fontweight='bold', fontsize=11)
axes[1].pie(counts.values, labels=[c.capitalize() for c in class_names], autopct='%1.1f%%', colors=colors, startangle=90, wedgeprops={'edgecolor': 'black', 'linewidth': 1.5}, textprops={'fontsize': 11, 'fontweight': 'bold'}); axes[1].set_title('Percentage Distribution', fontsize=13)
plt.tight_layout(); plt.savefig(PLTDIR / '12_class_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

train_s_df = pd.DataFrame(train_s, columns=['path', 'label']); train_s_df['split'] = 'Train'
val_s_df = pd.DataFrame(val_s, columns=['path', 'label']); val_s_df['split'] = 'Validation'
test_s_df = pd.DataFrame(test_s, columns=['path', 'label']); test_s_df['split'] = 'Test'
df_splits = pd.concat([train_s_df, val_s_df, test_s_df])
split_counts = df_splits.groupby(['split', 'label']).size().unstack(fill_value=0)[class_names]
fig, ax = plt.subplots(figsize=(10, 6)); fig.suptitle('Class Distribution Across Data Splits', fontsize=16, fontweight='bold')
split_counts.plot(kind='barh', stacked=True, ax=ax, color=colors, edgecolor='black', linewidth=0.5)
ax.set_xlabel('Number of Images'); ax.set_ylabel(''); ax.set_yticklabels(ax.get_yticklabels(), rotation=0, fontsize=12); ax.legend(title='Class', bbox_to_anchor=(1.05, 1)); ax.grid(axis='x', alpha=0.3)
for i, split in enumerate(split_counts.index): ax.text(split_counts.loc[split].sum() + 10, i, f'Total: {split_counts.loc[split].sum()}', va='center', fontweight='bold', fontsize=11)
plt.tight_layout(); plt.savefig(PLTDIR / '13_split_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

print("⏳ Computing average MRI per class...")
fig, axes = plt.subplots(1, 4, figsize=(16, 4)); fig.suptitle('Average MRI Intensity per Class', fontsize=14, fontweight='bold')
for i, cls in enumerate(class_names):
    cls_paths = df_all[df_all['label'] == cls]['path'].sample(n=min(200, len(df_all[df_all['label']==cls])), random_state=42).tolist()
    avg_img = np.zeros((224, 224, 3), dtype=np.float32)
    for p in cls_paths:
        img = cv2.imread(p)
        if img is not None: avg_img += cv2.resize(cv2.cvtColor(img, cv2.COLOR_BGR2RGB), (224, 224))
    axes[i].imshow((avg_img / len(cls_paths)).astype(np.uint8)); axes[i].set_title(f'{cls.capitalize()}'); axes[i].axis('off')
plt.tight_layout(); plt.savefig(PLTDIR / '14_average_mri.png', dpi=300, bbox_inches='tight'); plt.show()

print("⏳ Sampling pixel intensities...")
fig, ax = plt.subplots(figsize=(10, 6))
for i, cls in enumerate(class_names):
    cls_paths = df_all[df_all['label'] == cls]['path'].sample(n=50, random_state=42).tolist(); pixels = []
    for p in cls_paths:
        img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if img is not None: pixels.extend(np.random.choice(img.flatten(), size=1000, replace=False))
    sns.kdeplot(pixels, fill=True, color=colors[i], label=cls.capitalize(), ax=ax, alpha=0.3, linewidth=2)
ax.set_title('Pixel Intensity Distribution by Class', fontsize=14, fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '15_pixel_intensity.png', dpi=300, bbox_inches='tight'); plt.show()

print("⏳ Checking image dimensions...")
dim_data = []
for p in df_all['path'].sample(n=500, random_state=42):
    img = cv2.imread(p)
    if img is not None:
        h, w = img.shape[:2]; lbl = df_all.loc[df_all['path']==p, 'label'].values[0]
        dim_data.append({'Height': h, 'Width': w, 'Label': lbl.capitalize()})
dim_df = pd.DataFrame(dim_data)
fig, axes = plt.subplots(1, 2, figsize=(14, 6)); fig.suptitle('Original Image Dimension Analysis', fontsize=14, fontweight='bold')
sns.boxplot(data=dim_df, x='Label', y='Height', palette=colors, ax=axes[0]); axes[0].set_title('Image Height Distribution')
sns.boxplot(data=dim_df, x='Label', y='Width', palette=colors, ax=axes[1]); axes[1].set_title('Image Width Distribution')
plt.tight_layout(); plt.savefig(PLTDIR / '16_image_dimensions.png', dpi=300, bbox_inches='tight'); plt.show()


# ─── PLOT 17: GRAD-CAM + ATTENTION (FIXED FOR PYTORCH SHAPE) ────────────────────
print("\n🔥 Generating Grad-CAM + Attention Overlays...")

class AdvancedXAI:
    def __init__(self, model):
        self.model = model; self.model.eval()
        self.grads, self.activations = None, None
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        last_conv.register_forward_hook(self._act_hook)
        last_conv.register_backward_hook(self._grad_hook)

    def _act_hook(self, m, i, o): self.activations = o.detach()
    def _grad_hook(self, m, gi, go): self.grads = go[0].detach()

    def generate(self, img_path, target_class):
        img = cv2.imread(str(img_path)); rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        orig_h, orig_w = rgb.shape[:2]
        
        x = get_tensor(rgb).float().requires_grad_(True)
        out = self.model(x)
        self.model.zero_grad()
        out[0, target_class].backward(retain_graph=True)
        
        weights = self.grads.mean(dim=(2,3), keepdim=True)
        cam = F.relu((weights * self.activations).sum(1)).squeeze().cpu().numpy()
        cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)
        
        with torch.no_grad(): _ = self.model(x)
        g_attn, _ = self.model.get_attentions()
        
        # ✅ FIX: Safely handle both 3D (averaged) and 4D (separated heads) tensors
        attn_weights = g_attn[-1]
        if attn_weights.dim() == 4:
            attn = attn_weights[0, :, 0, 1:].mean(0).cpu().numpy() # Mean over heads
        else:
            attn = attn_weights[0, 0, 1:].cpu().numpy()           # Directly slice CLS to patches
            
        attn = attn.reshape(self.model.spatial_h, self.model.spatial_w)
        attn = (attn - attn.min()) / (attn.max() - attn.min() + 1e-8)
        
        def overlay_heatmap(hmap):
            hmap_resized = cv2.resize(hmap, (orig_w, orig_h))
            heatmap_color = cv2.applyColorMap((hmap_resized * 255).astype(np.uint8), cv2.COLORMAP_JET)
            return cv2.addWeighted(rgb, 0.6, heatmap_color, 0.4, 0)

        cam_rgb = overlay_heatmap(cam)
        attn_rgb = overlay_heatmap(attn)
        comb = 0.5*cam + 0.5*attn
        comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8)
        comb_rgb = overlay_heatmap(comb)
        
        return rgb, cam_rgb, attn_rgb, comb_rgb

xai = AdvancedXAI(model)
fig, axes = plt.subplots(3, 4, figsize=(20, 15))
fig.suptitle('Model Explainability: Grad-CAM vs. Attention vs. Combined', fontsize=18, fontweight='bold', y=1.01)
titles = ['Original MRI', 'Grad-CAM (CNN Focus)', 'Attention (ViT Focus)', 'Fused Explanation']

for row_idx, cls_idx in enumerate([1, 2, 3]):
    mask = (all_p == cls_idx) & (all_l == cls_idx) & (max_probs > 0.95)
    if not mask.any(): mask = (all_p == cls_idx) & (all_l == cls_idx)
    idx = np.where(mask)[0][0]
    
    imgs = xai.generate(test_s[idx][0], cls_idx)
    for col_idx, img in enumerate(imgs):
        axes[row_idx, col_idx].imshow(img)
        axes[row_idx, col_idx].set_title(f"{titles[col_idx]}\n{class_names[cls_idx].capitalize()} ({max_probs[idx]:.1%})", fontsize=12)
        axes[row_idx, col_idx].axis('off')

plt.tight_layout(); plt.savefig(PLTDIR / '17_xai_gradcam_attention.png', dpi=300, bbox_inches='tight'); plt.show()


# ─── PLOT 18 & 19: DEPLOYMENT THRESHOLDS ────────────────────────────────────────
print("\n🚀 Analyzing Class-wise Confidence Thresholds...")
thresholds = np.arange(0.5, 1.0, 0.05)
metrics_data = []
for thresh in thresholds:
    mask = max_probs >= thresh
    if mask.sum() == 0: continue
    p_thresh, l_thresh = all_p[mask], all_l[mask]
    acc = accuracy_score(l_thresh, p_thresh)
    f1 = f1_score(l_thresh, p_thresh, average='weighted', zero_division=0)
    coverage = mask.sum() / len(all_p)
    metrics_data.append({'Threshold': thresh, 'Accuracy': acc, 'F1-Score': f1, 'Coverage': coverage * 100})

df_thresh = pd.DataFrame(metrics_data)
fig, ax1 = plt.subplots(figsize=(10, 6))
fig.suptitle('Deployment Strategy: Confidence vs. Performance/Coverage', fontsize=14, fontweight='bold')
color1, color2 = '#3498db', '#e74c3c'
ax1.set_xlabel('Minimum Confidence Threshold'); ax1.set_ylabel('Metric Score', color=color1)
ax1.plot(df_thresh['Threshold'], df_thresh['Accuracy'], color=color1, marker='o', lw=2, label='Accuracy (Accepted)')
ax1.plot(df_thresh['Threshold'], df_thresh['F1-Score'], color='#2ecc71', marker='s', lw=2, label='F1-Score (Accepted)')
ax1.tick_params(axis='y', labelcolor=color1); ax1.grid(True, alpha=0.3)
ax2 = ax1.twinx(); ax2.set_ylabel('Dataset Coverage (%)', color=color2)
ax2.plot(df_thresh['Threshold'], df_thresh['Coverage'], color=color2, marker='^', lw=2, linestyle='--', label='Coverage')
ax2.tick_params(axis='y', labelcolor=color2)
fig.legend(loc='lower left', bbox_to_anchor=(0.1, -0.1), ncol=3); plt.tight_layout(); plt.savefig(PLTDIR / '18_deployment_thresholds.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
for i, cls in enumerate(class_names[1:]):
    cls_f1s = []; cls_threshs = np.arange(0.4, 1.0, 0.05)
    for t in cls_threshs:
        mask = (all_p == (i+1)) & (max_probs >= t)
        cls_f1s.append(f1_score(all_l[mask] == (i+1), all_p[mask] == (i+1), zero_division=0) if mask.sum() > 0 else 0)
    opt_idx = np.argmax(cls_f1s)
    ax.plot(cls_threshs, cls_f1s, color=colors[i+1], marker='o', lw=2, label=f'{cls.capitalize()} (Opt @ {cls_threshs[opt_idx]:.2f})')
    ax.axvline(x=cls_threshs[opt_idx], color=colors[i+1], linestyle='--', alpha=0.5)
ax.set_title('Per-Tumor Class: Precision-F1 vs. Confidence Threshold', fontsize=14, fontweight='bold'); ax.set_xlabel('Confidence Threshold'); ax.set_ylabel('F1-Score for Class'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '19_class_specific_thresholds.png', dpi=300, bbox_inches='tight'); plt.show()

# ─── PLOT 20: DEEP FEATURE MAPS (SAFE FALLBACK) ────────────────────────────────
print("\n🧠 Extracting Deep Feature Maps...")

# ✅ FIX: Safe fallback if no >95% confidence glioma exists in this specific split
mask = (all_l == 1) & (all_p == 1) & (max_probs > 0.90)
if not mask.any(): mask = (all_l == 1) & (all_p == 1) & (max_probs > 0.50)
if not mask.any(): mask = (all_l == 1) & (all_p == 1) # Any correct glioma
if not mask.any(): mask = (all_l == 1)               # Any glioma at all

glioma_idx = np.where(mask)[0][0]
test_rgb = cv2.cvtColor(cv2.imread(test_s[glioma_idx][0]), cv2.COLOR_BGR2RGB)
x_feat = get_tensor(test_rgb)

feature_maps = {}
def get_hook(name):
    def hook(module, input, output): feature_maps[name] = output.detach().cpu()
    return hook

model.backbone.feature_extractor[1].register_forward_hook(get_hook('Early_Layer'))
model.backbone.feature_extractor[4].register_forward_hook(get_hook('Mid_Layer'))
model.backbone.feature_extractor[7].register_forward_hook(get_hook('Late_Layer'))

with torch.no_grad(): _ = model(x_feat)

fig, axes = plt.subplots(3, 9, figsize=(20, 7))
fig.suptitle('Hierarchical Feature Extraction (Glioma MRI)', fontsize=16, fontweight='bold')
layers = ['Early_Layer', 'Mid_Layer', 'Late_Layer']
layer_names = ['Early Features (Edges/Textures)', 'Mid Features (Structures)', 'Late Features (Semantic/Tumor)']

for row_idx, (layer, title) in enumerate(zip(layers, layer_names)):
    fm = feature_maps[layer].squeeze(); n_channels = fm.shape[0]
    channel_idxs = np.linspace(0, n_channels - 1, 8, dtype=int)
    for col_idx, c_idx in enumerate(channel_idxs):
        channel_img = fm[c_idx].numpy()
        channel_img = (channel_img - channel_img.min()) / (channel_img.max() - channel_img.min() + 1e-8)
        axes[row_idx, col_idx].imshow(channel_img, cmap='magma'); axes[row_idx, col_idx].set_title(f'Ch {c_idx}', fontsize=9); axes[row_idx, col_idx].axis('off')
    axes[row_idx, -1].axis('off')
    axes[row_idx, -1].text(1.1, 0.5, title, transform=axes[row_idx, -1].transAxes, fontsize=12, fontweight='bold', va='center', rotation=270)

plt.tight_layout(); plt.savefig(PLTDIR / '20_feature_maps_hierarchy.png', dpi=300, bbox_inches='tight'); plt.show()

print("\n" + "="*60)
print("✅ ALL 20 PLOTS GENERATED SUCCESSFULLY!")
print(f"📁 Location: {PLTDIR}")
print("="*60)
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 15 — Download Entire Kaggle Working Space (FIXED)                        ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

import shutil
from IPython.display import FileLink

def download_workspace(output_name='brain_tumor.zip'):
    """Zip the entire /kaggle/working directory and provide download link."""
    
    zip_path = OUTDIR / output_name
    
    # Remove old zip if exists
    if zip_path.exists():
        zip_path.unlink()
    
    print("📦 Scanning /kaggle/working for files to archive...")
    
    total_size = 0
    for item in OUTDIR.rglob('*'):
        if item.is_file() and item != zip_path:
            size_mb = item.stat().st_size / (1024 * 1024)
            total_size += size_mb
            if size_mb > 0.1:
                rel_path = item.relative_to(OUTDIR)
                print(f"      📄 {rel_path}: {size_mb:.2f} MB")
    
    print(f"\n   Total uncompressed size: {total_size:.2f} MB")
    
    # ✅ FIXED: Removed 'with' statement
    print("\n🔒 Compressing... (this may take a minute for large models)")
    archive_name = str(zip_path.with_suffix(''))
    shutil.make_archive(archive_name, 'zip', OUTDIR)
    
    # Re-find the zip in case of naming quirks
    final_zip = Path(archive_name + '.zip')
    final_size = final_zip.stat().st_size / (1024 * 1024)
    
    print(f"✅ Archive created successfully!")
    print(f"   File: {final_zip.name}")
    print(f"   Size: {final_size:.2f} MB")
    
    # Provide download link
    print("\n⬇️ Click the link below to download:")
    display(FileLink(str(final_zip), result_html_prefix="🔗 Download Link: "))
    
    return str(final_zip)

# Run it
zip_file = download_workspace()
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 1 — Setup, Imports, and Global Configuration                             ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
!pip install -q timm einops albumentations opencv-python-headless

import os, time, random, copy, warnings
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageFile
import cv2
from tqdm.auto import tqdm
from scipy.ndimage import zoom

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
import torchvision.models as models

import albumentations as A
from albumentations.pytorch import ToTensorV2
from einops import rearrange

from sklearn.metrics import (accuracy_score, f1_score, roc_auc_score, roc_curve, 
                             confusion_matrix, classification_report, precision_recall_curve,
                             average_precision_score, matthews_corrcoef, cohen_kappa_score,
                             log_loss, brier_score_loss)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE

ImageFile.LOAD_TRUNCATED_IMAGES = True
warnings.filterwarnings('ignore')

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
CKDIR = OUTDIR / 'checkpoints'
PLTDIR = OUTDIR / 'plots'
for d in [CKDIR, PLTDIR]: d.mkdir(exist_ok=True)

MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff'}

# ✅ FRACTURE CONFIG - BINARY CLASSIFICATION
class_names = ['Not Fractured', 'Fractured']
NUM_CLASSES = 2
COLORS = ['#2ecc71', '#e74c3c'] # Green for Normal, Red for Fractured

print(f"✅ Setup complete on {DEVICE} | Classes: {class_names}")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 2 — Hyperparameters                                                       ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
HPARAMS = {
    'img_size': 224, 'batch_size': 32, 'num_epochs': 40,
    'lr_backbone': 1e-5, 'lr_head': 1e-4, 'weight_decay': 1e-4,
    'freeze_epochs': 8, 'patience': 10,
    'd_model': 384, 'nhead': 6, 'n_layers': 4, 'dim_ffn': 1536,
    'label_smoothing': 0.05, 'mixup_alpha': 0.2, 'cutmix_alpha': 1.0, 'mixup_prob': 0.3,
}

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 3 — Data Augmentations (X-Ray Safe: NO Vertical Flip)                   ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
def get_xray_transforms(split, img_size=224):
    norm = [A.Normalize(mean=MEAN, std=STD), ToTensorV2()]
    if split != 'train':
        return A.Compose([A.Resize(height=img_size, width=img_size)] + norm)
    return A.Compose([
        A.Resize(height=img_size, width=img_size),
        A.HorizontalFlip(p=0.5), # NO Vertical Flip for X-rays
        A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.1, rotate_limit=15, p=0.5, border_mode=cv2.BORDER_CONSTANT, value=0),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),
        A.CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=0.3),
    ] + norm)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 4 — Dataset Class (Handles Grayscale X-Rays)                             ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
class XRayDataset(Dataset):
    def __init__(self, samples, class_names, transform=None):
        self.samples = samples; self.transform = transform
        self.class_to_idx = {n: i for i, n in enumerate(class_names)}
    def __len__(self): return len(self.samples)
    def __getitem__(self, idx):
        p, l = self.samples[idx]; l = self.class_to_idx[l]
        # Read as grayscale and convert to 3 channels for ImageNet backbone
        img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if img is None: img = np.array(Image.open(p).convert('L'))
        img = np.stack([img, img, img], axis=-1) 
        if self.transform: img = self.transform(image=img)['image']
        return img.float(), l
    def class_weights(self):
        labels = [self.class_to_idx[l] for _, l in self.samples]
        counts = torch.bincount(torch.tensor(labels), minlength=NUM_CLASSES).float()
        beta = 0.9999
        effective_num = 1.0 - torch.pow(beta, counts)
        return ((1.0 - beta) / (effective_num + 1e-8)) / NUM_CLASSES * NUM_CLASSES

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 5 — Load Dataset (Using Exact Path & Pre-defined Splits)                 ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def load_fracture_dataset():
    print("📂 Loading Fracture X-Ray dataset from exact path...")
    
    # EXACT PATH PROVIDED BY USER
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification/')
    
    if not base.exists():
        raise ValueError(f"❌ Dataset not found at: {base}")
    print(f"✅ Found dataset at: {base}")
    
    # Pre-allocate lists for the dataset's official splits
    train_s, val_s, test_s = [], [], []
    
    # Map to loop over the folders
    split_map = {'train': train_s, 'val': val_s, 'test': test_s}
    
    for split_name, split_list in split_map.items():
        split_dir = base / split_name
        if not split_dir.exists():
            print(f"   ⚠️ Warning: {split_name} folder not found. Skipping.")
            continue
            
        # Iterate through class folders inside train/val/test
        for cls_dir in split_dir.iterdir():
            if not cls_dir.is_dir(): continue
            cls_name_raw = cls_dir.name.lower().replace('-', ' ')
            
            # Smart folder name mapping
            if 'not fractured' in cls_name_raw or 'normal' in cls_name_raw or 'negative' in cls_name_raw:
                cls_label = 'Not Fractured'
            elif 'fractured' in cls_name_raw or 'positive' in cls_name_raw:
                cls_label = 'Fractured'
            else:
                print(f"   ⚠️ Skipping unknown folder: {cls_dir.name}")
                continue
                
            # Collect image paths
            count = 0
            for ext in VALID_EXTS:
                for p in cls_dir.rglob(f'*{ext}'):
                    split_list.append((str(p), cls_label))
                    count += 1
            
            if count > 0:
                print(f"   ✅ {split_name.upper():>5} | {cls_label:15s}: {count:5d} images")
                
    return train_s, val_s, test_s

# Execute loading
train_s, val_s, test_s = load_fracture_dataset()

if not train_s and not val_s and not test_s:
    raise ValueError("❌ No images loaded! Please check folder names inside train/val/test.")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 6 — Dataset Analysis (No Re-splitting Needed)                             ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

# Combine all just for overall stats printing
all_samples = train_s + val_s + test_s
df_all = pd.DataFrame(all_samples, columns=['path', 'label'])

print("\n" + "="*60)
print("📊 BONE FRACTURE DATASET ANALYSIS")
print("="*60)
print(f"\n🔹 Total Images: {len(df_all)} | Classes: {len(class_names)}")
print(f"🔹 Split (Pre-defined from folders):")
print(f"   - Train:      {len(train_s):5d} images ({len(train_s)/len(df_all)*100:.1f}%)")
print(f"   - Validation: {len(val_s):5d} images ({len(val_s)/len(df_all)*100:.1f}%)")
print(f"   - Test:       {len(test_s):5d} images ({len(test_s)/len(df_all)*100:.1f}%)")

print("\n🔹 Class Distribution (Overall):")
for cls in class_names:
    count = len(df_all[df_all['label'] == cls])
    pct = count/len(df_all)*100 if len(df_all) > 0 else 0
    print(f"   - {cls:15s}: {count:5d} images ({pct:.1f}%)")
print("="*60 + "\n")


# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 7 — Create Dataloaders                                                    ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
train_ds = XRayDataset(train_s, class_names, get_xray_transforms('train'))
val_ds = XRayDataset(val_s, class_names, get_xray_transforms('val'))
test_ds = XRayDataset(test_s, class_names, get_xray_transforms('test'))

class_weights = train_ds.class_weights()
sample_weights = [class_weights[class_names.index(l)] for _, l in train_s]

loaders = {
    'train': DataLoader(train_ds, batch_size=HPARAMS['batch_size'], sampler=WeightedRandomSampler(sample_weights, len(sample_weights), replacement=True), num_workers=2, pin_memory=True),
    'val': DataLoader(val_ds, batch_size=HPARAMS['batch_size'], shuffle=False, num_workers=2, pin_memory=True),
    'test': DataLoader(test_ds, batch_size=HPARAMS['batch_size'], shuffle=False, num_workers=2, pin_memory=True)
}

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 8 — Model Architecture (ConvNeXt-Base + Hybrid ViT)                      ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): super().__init__(); self.drop_prob = drop_prob
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training: return x
        return x * (torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * float(1.0 - self.drop_prob))

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3); self.proj = nn.Linear(d_model, d_model)
        self.attn_drop, self.proj_drop, self.attn_weights = nn.Dropout(dropout), nn.Dropout(dropout), None
    def forward(self, x, H, W):
        B, N, C = x.shape; x = x.reshape(B, H, W, C)
        pad_h, pad_w = (self.window_size - H % self.window_size) % self.window_size, (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0: x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(-1, self.window_size**2, C)
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2,0,3,1,4)
        attn = (qkv[0] @ qkv[1].transpose(-2,-1)) * self.scale; attn = attn.softmax(dim=-1); self.attn_weights = attn.detach()
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1,2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(B, Hp, Wp, C)
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model*2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop), nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop))
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights, self.window_attn_weights = None, None
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        window_out = self.window_attn(x_norm[:, 1:, :], H, W); self.window_attn_weights = self.window_attn.attn_weights
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1-gate) * window_out], dim=1)
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self): super().__init__(); self.backbone = models.convnext_base(weights=models.ConvNeXt_Base_Weights.IMAGENET1K_V1); self.feature_extractor = self.backbone.features
    def forward(self, x): return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes=NUM_CLASSES, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536):
        super().__init__(); self.d_model = d_model; self.img_size = img_size; self.backbone = ImprovedBackbone()
        with torch.no_grad(): _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        self.spatial_h, self.spatial_w = h, w
        self.proj = nn.Sequential(nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(), nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU())
        self.norm_proj = nn.LayerNorm(d_model); self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h*w + 1, d_model))
        nn.init.trunc_normal_(self.cls_token, std=0.02); nn.init.trunc_normal_(self.pos_embed, std=0.02)
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i]) for i in range(n_layers)])
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, d_model//2), nn.GELU(), nn.Dropout(0.3), nn.Linear(d_model//2, d_model//4), nn.GELU(), nn.Dropout(0.15), nn.Linear(d_model//4, num_classes))
    def forward(self, x):
        B = x.size(0); tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens); tokens = torch.cat([self.cls_token.expand(B,-1,-1), tokens], dim=1) + self.pos_embed
        for block in self.encoder: tokens = block(tokens, self.spatial_h, self.spatial_w)
        return self.head(self.norm(tokens)[:, 0])
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w
    def unfreeze_backbone_partial(self, n=3):
        for p in self.backbone.feature_extractor.parameters(): p.requires_grad_(False)
        for child in list(self.backbone.feature_extractor.children())[-n:]:
            for p in child.parameters(): p.requires_grad_(True)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 9 — Training Functions                                                    ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
class SmoothCELoss(nn.Module):
    def __init__(self, label_smoothing=0.05): super().__init__(); self.criterion = nn.CrossEntropyLoss(label_smoothing=label_smoothing)
    def forward(self, inputs, targets): return self.criterion(inputs, targets)

def mixup_data(x, y, alpha=0.2):
    lam = np.random.beta(alpha, alpha); idx = torch.randperm(x.size(0)).to(x.device)
    return lam*x + (1-lam)*x[idx], y, y[idx], lam

def cutmix_data(x, y, alpha=1.0):
    lam = np.random.beta(alpha, alpha); idx = torch.randperm(x.size(0)).to(x.device)
    W, H = x.size(2), x.size(3); r = np.sqrt(1.0 - lam)
    cw, ch = int(W*r), int(H*r); cx, cy = np.random.randint(W), np.random.randint(H)
    x1, y1 = np.clip(cx - cw//2, 0, W), np.clip(cy - ch//2, 0, H)
    x2, y2 = np.clip(cx + cw//2, 0, W), np.clip(cy + ch//2, 0, H)
    mixed = x.clone(); mixed[:, :, x1:x2, y1:y2] = x[idx, :, x1:x2, y1:y2]
    return mixed, y, y[idx], 1 - (x2-x1)*(y2-y1)/(W*H)

def train_one_epoch(model, loader, optimizer, scaler, criterion, epoch):
    model.train(); total_loss, preds, labs = 0, [], []
    for images, labels in tqdm(loader, desc=f'Epoch {epoch+1}', leave=False):
        images, labels = images.to(DEVICE), labels.to(DEVICE); optimizer.zero_grad(set_to_none=True)
        mixed = False
        if np.random.random() < HPARAMS['mixup_prob']:
            if np.random.random() < 0.5: images, la, lb, lam = mixup_data(images, labels, HPARAMS['mixup_alpha']); mixed = True
            else: images, la, lb, lam = cutmix_data(images, labels, HPARAMS['cutmix_alpha']); mixed = True
        with autocast():
            out = model(images)
            loss = lam * criterion(out, la) + (1-lam) * criterion(out, lb) if mixed else criterion(out, labels)
            if not mixed: preds.extend(out.argmax(1).cpu().numpy()); labs.extend(labels.cpu().numpy())
        scaler.scale(loss).backward(); scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0); scaler.step(optimizer); scaler.update()
        total_loss += loss.item()
    return total_loss/len(loader), accuracy_score(labs, preds) if labs else 0, f1_score(labs, preds, average='weighted', zero_division=0) if labs else 0

@torch.no_grad()
def validate(model, loader, criterion):
    model.eval(); total_loss, preds, labs, probs = 0, [], [], []
    for images, labels in tqdm(loader, desc='Validating', leave=False):
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        with autocast(): out = model(images); loss = criterion(out, labels)
        total_loss += loss.item(); p = F.softmax(out, dim=1)
        preds.extend(out.argmax(1).cpu().numpy()); labs.extend(labels.cpu().numpy()); probs.extend(p.cpu().numpy())
    return total_loss/len(loader), accuracy_score(labs, preds), f1_score(labs, preds, average='weighted', zero_division=0), preds, labs, probs

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 10 — Main Training Loop                                                    ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
print("\n🚀 Starting Training for Bone Fracture Detection...")
model = EnhancedHViT(num_classes=NUM_CLASSES).to(DEVICE)
criterion = SmoothCELoss(label_smoothing=HPARAMS['label_smoothing'])
optimizer = optim.AdamW([
    {'params': [p for n, p in model.named_parameters() if 'backbone' in n], 'lr': HPARAMS['lr_backbone']},
    {'params': [p for n, p in model.named_parameters() if 'backbone' not in n], 'lr': HPARAMS['lr_head']}
], weight_decay=HPARAMS['weight_decay'])
scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=HPARAMS['num_epochs']//3, T_mult=2, eta_min=1e-7)
scaler = GradScaler()

for p in model.backbone.feature_extractor.parameters(): p.requires_grad_(False) # Freeze initially

best_f1, patience_counter = 0, 0
history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': [], 'val_f1': [], 'lr': []}

for epoch in range(HPARAMS['num_epochs']):
    if epoch == HPARAMS['freeze_epochs']:
        print("\n🔄 Unfreezing backbone..."); model.unfreeze_backbone_partial(3)
    train_loss, train_acc, train_f1 = train_one_epoch(model, loaders['train'], optimizer, scaler, criterion, epoch)
    val_loss, val_acc, val_f1, _, _, _ = validate(model, loaders['val'], criterion)
    scheduler.step()
    history['train_loss'].append(train_loss); history['val_loss'].append(val_loss)
    history['train_acc'].append(train_acc); history['val_acc'].append(val_acc)
    history['val_f1'].append(val_f1); history['lr'].append(optimizer.param_groups[0]['lr'])
    print(f"Epoch {epoch+1}: Train Acc={train_acc:.3f} | Val Acc={val_acc:.3f}, Val F1={val_f1:.3f}")
    
    if val_f1 > best_f1:
        best_f1 = val_f1; patience_counter = 0
        torch.save({'model': model.state_dict(), 'f1': val_f1, 'class_names': class_names}, CKDIR / 'best_model.pth')
        print(f"   ✅ Best model saved!")
    else:
        patience_counter += 1
        if patience_counter >= HPARAMS['patience']: print("⏹️ Early stopping"); break
print("\n✅ Training complete!")
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  STANDALONE HEATMAP GENERATOR - FRACTURE X-RAYS                                 ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
import torch, torch.nn as nn, torch.nn.functional as F
import numpy as np, cv2, matplotlib.pyplot as plt
from pathlib import Path
from scipy.ndimage import zoom
from einops import rearrange
import torchvision.models as models
from tqdm.auto import tqdm
from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings('ignore')

class_names = ['Not Fractured', 'Fractured']
colors = ['#2ecc71', '#e74c3c']

def get_tensor(img_input):
    if len(img_input.shape) == 2: img_input = np.stack([img_input]*3, axis=-1)
    t = torch.from_numpy(cv2.resize(img_input, (224, 224))).float().permute(2,0,1)/255.0
    return ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)

class XRayHeatmapGenerator:
    def __init__(self, model):
        self.model = model; self.model.eval(); self.cnn_f, self.cnn_g = None, None
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        if last_conv:
            last_conv.register_forward_hook(lambda m, i, o: setattr(self, 'cnn_f', o.detach()))
            last_conv.register_backward_hook(lambda m, gi, go: setattr(self, 'cnn_g', go[0].detach()))

    def get_cnn(self, x, target):
        out = self.model(x); self.model.zero_grad()
        oh = torch.zeros_like(out); oh[0, target] = 1
        out.backward(gradient=oh, retain_graph=True)
        if self.cnn_f is None: return None
        weights = self.cnn_g.mean(dim=(2, 3), keepdim=True)
        cam = F.relu((weights * self.cnn_f).sum(1)).squeeze().cpu().numpy()
        return (cam - cam.min()) / (cam.max() - cam.min() + 1e-8) if cam.max() > cam.min() else cam

    def get_attn(self, x):
        with torch.no_grad(): self.model(x)
        g, _ = self.model.get_attentions()
        if not g: return None
        a = g[-1]
        a = a[0, :, 0, 1:].mean(0).cpu().numpy() if a.dim() == 4 else a[0, 0, 1:].cpu().numpy()
        a = a.reshape(self.model.spatial_h, self.model.spatial_w)
        return (a - a.min()) / (a.max() - a.min() + 1e-8) if a.max() > a.min() else a

    def draw(self, img_path, target, save_path):
        img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        if img is None: return
        rgb = np.stack([img, img, img], axis=-1)
        sz = rgb.shape[:2]; t = get_tensor(rgb)
        c, a = self.get_cnn(t.clone(), target), self.get_attn(t.clone())
        comb = 0.5 * c + 0.5 * a if c is not None and a is not None else c
        if comb is not None and comb.max() > comb.min(): comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8)

        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        axes[0].imshow(img, cmap='gray'); axes[0].set_title(f"True: {class_names[target]}", fontweight='bold'); axes[0].axis('off')
        for i, (hmap, title) in enumerate(zip([c, a, comb], ['Grad-CAM', 'Attention', 'Combined'])):
            if hmap is not None:
                hr = zoom(hmap, (sz[0]/hmap.shape[0], sz[1]/hmap.shape[1]), order=1)
                colored = (plt.cm.jet(hr)[:,:,:3] * 255).astype(np.uint8)
                axes[i+1].imshow(cv2.addWeighted(rgb, 0.6, colored, 0.4, 0))
            else: axes[i+1].imshow(img, cmap='gray')
            axes[i+1].set_title(title, fontweight='bold'); axes[i+1].axis('off')
        plt.tight_layout(); plt.savefig(save_path, dpi=150, bbox_inches='tight'); plt.show()

hm_gen = XRayHeatmapGenerator(model)
print("🔥 Generating Heatmaps for Fracture X-Rays...")
for cls_idx in range(2):
    cls_tests = [s for s in test_s if class_names.index(s[1]) == cls_idx]
    if cls_tests:
        print(f"\n📊 {class_names[cls_idx]}:")
        for p, _ in cls_tests[:2]: hm_gen.draw(p, cls_idx, PLTDIR / f'hm_{class_names[cls_idx]}_{Path(p).stem}.png')
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  ULTIMATE MEGA CELL: ALL 20 PLOTS + EXACT METRICS (Bone Fracture Binary)       ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

import os, time, random, warnings
from pathlib import Path
import numpy as np, pandas as pd, matplotlib.pyplot as plt, seaborn as sns, cv2
from tqdm.auto import tqdm
from scipy.ndimage import zoom
import torch, torch.nn as nn, torch.nn.functional as F
import torchvision.models as models
from sklearn.metrics import (accuracy_score, f1_score, roc_curve, auc, confusion_matrix,
                             classification_report, precision_recall_curve, average_precision_score,
                             roc_auc_score, matthews_corrcoef, cohen_kappa_score, log_loss, brier_score_loss)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE
from torch.cuda.amp import autocast
from einops import rearrange
import warnings
warnings.filterwarnings('ignore')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
PLTDIR = OUTDIR / 'plots/fracture_final'; PLTDIR.mkdir(exist_ok=True, parents=True)
CKDIR = OUTDIR / 'checkpoints'
MEAN, STD = (0.485, 0.456, 0.406), (0.229, 0.224, 0.225)
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff'}
class_names = ['Not Fractured', 'Fractured']
COLORS = ['#2ecc71', '#e74c3c']

# ─── 1. RE-DEFINE MODEL ARCHITECTURE (Required for standalone execution) ─────────
class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): super().__init__(); self.drop_prob = drop_prob
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training: return x
        return x * (torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * float(1.0 - self.drop_prob))

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3); self.proj = nn.Linear(d_model, d_model)
        self.attn_drop, self.proj_drop, self.attn_weights = nn.Dropout(dropout), nn.Dropout(dropout), None
    def forward(self, x, H, W):
        B, N, C = x.shape; x = x.reshape(B, H, W, C)
        pad_h, pad_w = (self.window_size - H % self.window_size) % self.window_size, (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0: x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(-1, self.window_size**2, C)
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2,0,3,1,4)
        attn = (qkv[0] @ qkv[1].transpose(-2,-1)) * self.scale; attn = attn.softmax(dim=-1); self.attn_weights = attn.detach()
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1,2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(B, Hp, Wp, C)
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model*2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop), nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop))
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights, self.window_attn_weights = None, None
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        window_out = self.window_attn(x_norm[:, 1:, :], H, W); self.window_attn_weights = self.window_attn.attn_weights
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1-gate) * window_out], dim=1)
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self): super().__init__(); self.backbone = models.convnext_base(weights=None); self.feature_extractor = self.backbone.features
    def forward(self, x): return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes=2, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536):
        super().__init__(); self.d_model = d_model; self.img_size = img_size; self.backbone = ImprovedBackbone()
        with torch.no_grad(): _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        self.spatial_h, self.spatial_w = h, w
        self.proj = nn.Sequential(nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(), nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU())
        self.norm_proj = nn.LayerNorm(d_model); self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h*w + 1, d_model))
        nn.init.trunc_normal_(self.cls_token, std=0.02); nn.init.trunc_normal_(self.pos_embed, std=0.02)
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i]) for i in range(n_layers)])
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, d_model//2), nn.GELU(), nn.Dropout(0.3), nn.Linear(d_model//2, d_model//4), nn.GELU(), nn.Dropout(0.15), nn.Linear(d_model//4, num_classes))
    def forward(self, x):
        B = x.size(0); tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens); tokens = torch.cat([self.cls_token.expand(B,-1,-1), tokens], dim=1) + self.pos_embed
        for block in self.encoder: tokens = block(tokens, self.spatial_h, self.spatial_w)
        return self.head(self.norm(tokens)[:, 0])
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w

def get_tensor(img):
    if len(img.shape) == 2: img = np.stack([img]*3, axis=-1)
    t = torch.from_numpy(cv2.resize(img, (224, 224))).float().permute(2,0,1)/255.0
    return ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)

# ─── 2. LOAD MODEL & ROBUST TEST DATA LOADING ───────────────────────────────────
print("📦 Loading Bone Fracture Model & Test Data...")
model = EnhancedHViT(num_classes=2).to(DEVICE)
model.load_state_dict(torch.load(CKDIR / 'best_model.pth', map_location=DEVICE, weights_only=False)['model'])
model.eval()

base = Path('/kaggle/input/datasets/aayush4204/daataaset/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification/Bone_Fracture_Binary_Classification')
test_s = []
test_dir = base / 'test'

print(f"Scanning {test_dir}...")
if test_dir.exists():
    for cls_dir in sorted(test_dir.iterdir()):
        if not cls_dir.is_dir(): continue  # Skip files like .DS_Store

        # FIX: Normalize the folder name once, then check on the SAME normalized string
        name_lower = cls_dir.name.lower().replace('-', ' ').replace('_', ' ')

        if 'fractured' in name_lower and 'not' not in name_lower:
            cls_name = 'Fractured'
        else:
            cls_name = 'Not Fractured'

        count = 0
        for ext in VALID_EXTS:
            for p in cls_dir.rglob(f'*{ext}'):
                test_s.append((str(p), cls_name))
                count += 1
        print(f"   ✅ Loaded {count:>4d} images for '{cls_name}' from '{cls_dir.name}'")
else:
    print(f"❌ Test directory not found!")

# Quick sanity check before inference
from collections import Counter
label_counts = Counter(l for _, l in test_s)
print(f"\n   🔍 Sanity Check: {dict(label_counts)}")
if label_counts.get('Fractured', 0) == len(test_s) or label_counts.get('Not Fractured', 0) == 0:
    print("   ⚠️  WARNING: All images loaded as same class — check folder names!")
else:
    print("   ✅ Both classes present — loading looks correct.\n")

# Inference
all_p, all_l, all_pr = [], [], []
for p, l in tqdm(test_s, desc="Predicting"):
    img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
    if img is None: continue
    with torch.no_grad(): out = model(get_tensor(img)); pr = F.softmax(out, dim=1)
    all_p.append(out.argmax(1).item()); all_l.append(class_names.index(l)); all_pr.append(pr.cpu().numpy()[0])

all_p, all_l, all_pr = np.array(all_p), np.array(all_l), np.array(all_pr)
max_probs = all_pr.max(axis=1); correct_mask = (all_p == all_l)
labels_bin = label_binarize(all_l, classes=[0, 1])
cm = confusion_matrix(all_l, all_p)
cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 1: TRAINING HISTORY
# ─────────────────────────────────────────────────────────────────────────────────
try:
    if len(history.get('train_loss', [])) > 0:
        fig, ax = plt.subplots(2, 2, figsize=(14, 10))
        fig.suptitle('Fracture Model Training Dynamics', fontsize=16, fontweight='bold')
        ax[0,0].plot(history['train_loss'], label='Train', color='red'); ax[0,0].plot(history['val_loss'], label='Val', color='blue'); ax[0,0].set_title('Loss'); ax[0,0].legend(); ax[0,0].grid(True, alpha=0.3)
        ax[0,1].plot(history['train_acc'], label='Train', color='red'); ax[0,1].plot(history['val_acc'], label='Val', color='blue'); ax[0,1].set_title('Accuracy'); ax[0,1].legend(); ax[0,1].grid(True, alpha=0.3)
        ax[1,0].plot(history['val_f1'], label='Val F1', color='green'); ax[1,0].set_title('F1 Score'); ax[1,0].legend(); ax[1,0].grid(True, alpha=0.3)
        ax[1,1].plot(history['lr'], color='purple'); ax[1,1].set_title('Learning Rate'); ax[1,1].set_yscale('log'); ax[1,1].grid(True, alpha=0.3)
        plt.tight_layout(); plt.savefig(PLTDIR / '1_training_history.png', dpi=300, bbox_inches='tight'); plt.show()
except Exception as e: pass

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 2: CONFUSION MATRICES
# ─────────────────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Bone Fracture Confusion Matrix', fontsize=16, fontweight='bold')
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[0])
axes[0].set_title('Absolute Counts'); axes[0].set_ylabel('True'); axes[0].set_xlabel('Predicted')
sns.heatmap(cm_norm, annot=True, fmt='.2%', cmap='Greens', xticklabels=class_names, yticklabels=class_names, ax=axes[1])
axes[1].set_title('Normalized (Recall)'); axes[1].set_ylabel('True'); axes[1].set_xlabel('Predicted')
plt.tight_layout(); plt.savefig(PLTDIR / '2_confusion_matrices.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 3: ROC CURVE
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 8))
fpr, tpr, _ = roc_curve(all_l, all_pr[:, 1]); roc_auc = auc(fpr, tpr)
ax.plot(fpr, tpr, color=COLORS[1], lw=2, label=f'Fractured (AUC = {roc_auc:.3f})')
ax.plot([0, 1], [0, 1], 'k--', lw=1); ax.set_title('ROC Curve - Fracture Detection', fontweight='bold')
ax.set_xlabel('False Positive Rate'); ax.set_ylabel('True Positive Rate'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '3_roc_curve.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 4: PRECISION-RECALL CURVE
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(8, 8))
prec, rec, _ = precision_recall_curve(all_l, all_pr[:, 1]); ap = average_precision_score(all_l, all_pr[:, 1])
ax.plot(rec, prec, color=COLORS[1], lw=2, label=f'Fractured (AP = {ap:.3f})')
ax.set_title('Precision-Recall Curve', fontweight='bold'); ax.set_xlabel('Recall'); ax.set_ylabel('Precision')
ax.legend(); ax.grid(True, alpha=0.3); ax.set_xlim([0,1]); ax.set_ylim([0,1.05])
plt.tight_layout(); plt.savefig(PLTDIR / '4_precision_recall.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 5: PER-CLASS METRICS BARS
# ─────────────────────────────────────────────────────────────────────────────────
report = classification_report(all_l, all_p, target_names=class_names, output_dict=True)
m_df = pd.DataFrame(report).T.iloc[:-3, :3]
fig, ax = plt.subplots(figsize=(8, 6))
x = np.arange(2); width = 0.25
ax.bar(x - width, m_df['precision'], width, label='Precision', color='#3498db', edgecolor='black')
ax.bar(x, m_df['recall'], width, label='Recall', color='#2ecc71', edgecolor='black')
ax.bar(x + width, m_df['f1-score'], width, label='F1-Score', color='#e74c3c', edgecolor='black')
ax.set_title('Per-Class Performance Metrics', fontweight='bold'); ax.set_xticks(x); ax.set_xticklabels(class_names)
ax.legend(); ax.set_ylim(0,1.15); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '5_per_class_metrics.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 6: CONFIDENCE DISTRIBUTION
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(max_probs[correct_mask], bins=20, alpha=0.7, label='Correct', color='#2ecc71', edgecolor='black')
if (~correct_mask).sum() > 0:
    ax.hist(max_probs[~correct_mask], bins=20, alpha=0.7, label='Incorrect', color='#e74c3c', edgecolor='black')
ax.set_title('Model Confidence Distribution', fontweight='bold'); ax.set_xlabel('Maximum Probability'); ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '6_confidence_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 7: CALIBRATION CURVE
# ─────────────────────────────────────────────────────────────────────────────────
def calibration_curve(y_true, y_prob, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1); accs, confs = [], []
    for i in range(n_bins):
        mask = (y_prob > bins[i]) & (y_prob <= bins[i+1])
        if i == n_bins - 1: mask = (y_prob >= bins[i]) & (y_prob <= bins[i+1])
        if mask.sum() > 0: accs.append(y_true[mask].mean()); confs.append(y_prob[mask].mean())
    return np.array(confs), np.array(accs)

conf, acc = calibration_curve(all_l, all_pr[:, 1])
fig, ax = plt.subplots(figsize=(8, 8))
if len(conf) > 0:
    ax.plot(conf, acc, marker='o', color='blue', lw=2, label='Fractured')
ax.plot([0, 1], [0, 1], 'k--', lw=1, label='Perfectly Calibrated')
ax.set_title('Reliability Diagram', fontweight='bold'); ax.set_xlabel('Mean Predicted Probability'); ax.set_ylabel('Fraction of Positives')
ax.legend(); ax.grid(True, alpha=0.3); ax.set_xlim([0,1]); ax.set_ylim([0,1])
plt.tight_layout(); plt.savefig(PLTDIR / '7_calibration_curve.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 8: SAMPLE PREDICTIONS
# ─────────────────────────────────────────────────────────────────────────────────
correct_idxs, incorrect_idxs = np.where(correct_mask)[0], np.where(~correct_mask)[0]
np.random.seed(42)
samp_corr = np.random.choice(correct_idxs, min(6, len(correct_idxs)), replace=False)
samp_incorr = np.random.choice(incorrect_idxs, min(6, len(incorrect_idxs)), replace=False) if len(incorrect_idxs) > 0 else []
n_cols = max(6, len(samp_corr))
fig, axes = plt.subplots(2, n_cols, figsize=(18, 6))
if len(samp_incorr) == 0:
    for c in range(n_cols): axes[1, c].axis('off')
for i, idx in enumerate(samp_corr):
    img = cv2.imread(test_s[idx][0], cv2.IMREAD_GRAYSCALE)
    axes[0, i].imshow(img, cmap='gray'); axes[0, i].set_title(f"T:{class_names[all_l[idx]]}\nP:{class_names[all_p[idx]]}", fontsize=9, color='green'); axes[0, i].axis('off')
for i, idx in enumerate(samp_incorr):
    img = cv2.imread(test_s[idx][0], cv2.IMREAD_GRAYSCALE)
    axes[1, i].imshow(img, cmap='gray'); axes[1, i].set_title(f"T:{class_names[all_l[idx]]}\nP:{class_names[all_p[idx]]}", fontsize=9, color='red'); axes[1, i].axis('off')
axes[0, 0].set_ylabel('CORRECT', fontsize=12, fontweight='bold', color='green', rotation=0, labelpad=60, va='center')
axes[1, 0].set_ylabel('INCORRECT', fontsize=12, fontweight='bold', color='red', rotation=0, labelpad=60, va='center')
fig.suptitle('Visual Inspection (T=True, P=Predicted)', fontsize=14, fontweight='bold', y=1.05)
plt.tight_layout(); plt.savefig(PLTDIR / '8_sample_predictions.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 9: VIOLIN PLOT PROBABILITIES
# ─────────────────────────────────────────────────────────────────────────────────
plot_data = [{'True Class': class_names[i], 'Assigned Probability': prob} for i in range(2) for prob in all_pr[all_l == i, i]]
df_probs = pd.DataFrame(plot_data)
fig, ax = plt.subplots(figsize=(10, 6))
sns.violinplot(data=df_probs, x='True Class', y='Assigned Probability', palette=COLORS, inner='quartile', ax=ax)
ax.set_title('Probability Density for True Class', fontweight='bold'); ax.set_ylim(0,1.05); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '9_probability_violin.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 10: t-SNE FEATURE VISUALIZATION
# ─────────────────────────────────────────────────────────────────────────────────
print("⏳ Extracting t-SNE features...")
feature_tensor = None
def hook_fn(module, inp, out): global feature_tensor; feature_tensor = out
handle = model.norm.register_forward_hook(hook_fn)
tsne_indices = np.random.choice(len(test_s), min(500, len(test_s)), replace=False)
features_list, labels_tsne = [], []
for idx in tsne_indices:
    img = cv2.imread(test_s[idx][0], cv2.IMREAD_GRAYSCALE)
    with torch.no_grad(): _ = model(get_tensor(img))
    features_list.append(feature_tensor[:, 0, :].cpu().numpy()[0]); labels_tsne.append(all_l[idx])
handle.remove()
tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000).fit_transform(np.array(features_list))
fig, ax = plt.subplots(figsize=(10, 8))
for i, cls in enumerate(class_names):
    mask = np.array(labels_tsne) == i
    ax.scatter(tsne[mask, 0], tsne[mask, 1], c=COLORS[i], label=cls, alpha=0.6, s=20, edgecolors='white', linewidth=0.5)
ax.set_title('t-SNE Visualization of Deep Features', fontweight='bold'); ax.legend(markerscale=2); ax.grid(True, alpha=0.2)
ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.savefig(PLTDIR / '10_tsne_features.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 11: ERROR ANALYSIS
# ─────────────────────────────────────────────────────────────────────────────────
errors = [{'True': class_names[all_l[i]], 'Predicted': class_names[all_p[i]]} for i in range(len(all_l)) if all_l[i] != all_p[i]]
if errors:
    error_df = pd.DataFrame(errors)
    error_counts = error_df.groupby(['True', 'Predicted']).size().reset_index(name='Count').sort_values(by='Count', ascending=False)
    fig, ax = plt.subplots(figsize=(8, 5))
    error_counts['Pair'] = error_counts['True'] + ' → ' + error_counts['Predicted']
    sns.barplot(data=error_counts, x='Count', y='Pair', palette='Reds_r', ax=ax)
    ax.set_title('Misclassification Pairs', fontweight='bold')
    for i, v in enumerate(error_counts['Count']): ax.text(v + 0.5, i, str(v), color='black', va='center', fontweight='bold')
    plt.tight_layout(); plt.savefig(PLTDIR / '11_error_analysis.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 12: CLASS DISTRIBUTION (FIXED)
# ─────────────────────────────────────────────────────────────────────────────────
all_samples = test_s
df_all = pd.DataFrame(all_samples, columns=['path', 'label'])
counts = df_all['label'].value_counts().reindex(class_names).fillna(0).astype(int)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
fig.suptitle('Test Set Class Distribution', fontsize=16, fontweight='bold')
sns.barplot(x=counts.index, y=counts.values, ax=axes[0], palette=COLORS, edgecolor='black')
axes[0].set_title('Image Count per Class')
for i, v in enumerate(counts.values): axes[0].text(i, v + 1, str(v), ha='center', fontweight='bold')

if counts.sum() > 0:
    axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%', colors=COLORS, startangle=90, wedgeprops={'edgecolor': 'black', 'linewidth': 1.5})
else:
    axes[1].text(0.5, 0.5, "No Data Found", ha='center', va='center', fontsize=14)
axes[1].set_title('Percentage Distribution')
plt.tight_layout(); plt.savefig(PLTDIR / '12_class_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 13: SPLIT DISTRIBUTION
# ─────────────────────────────────────────────────────────────────────────────────
split_data = []
for split_name in ['train', 'val', 'test']:
    split_dir = base / split_name
    if not split_dir.exists(): continue
    for cls_dir in split_dir.iterdir():
        if not cls_dir.is_dir(): continue
        name_lower = cls_dir.name.lower().replace('-', ' ').replace('_', ' ')
        cls_name = 'Fractured' if ('fractured' in name_lower and 'not' not in name_lower) else 'Not Fractured'
        count = sum(1 for ext in VALID_EXTS for p in cls_dir.rglob(f'*{ext}'))
        split_data.append({'split': split_name.capitalize(), 'label': cls_name, 'count': count})
df_splits = pd.DataFrame(split_data)
split_pivot = df_splits.pivot(index='split', columns='label', values='count').reindex(['Train', 'Val', 'Test'])[class_names].fillna(0).astype(int)

fig, ax = plt.subplots(figsize=(10, 6))
split_pivot.plot(kind='barh', stacked=True, ax=ax, color=COLORS, edgecolor='black')
ax.set_title('Class Distribution Across Data Splits', fontweight='bold'); ax.set_xlabel('Number of Images')
for i, split in enumerate(split_pivot.index): ax.text(split_pivot.loc[split].sum() + 5, i, f'Total: {split_pivot.loc[split].sum()}', va='center', fontweight='bold')
plt.tight_layout(); plt.savefig(PLTDIR / '13_split_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 14: AVERAGE IMAGES PER CLASS
# ─────────────────────────────────────────────────────────────────────────────────
print("⏳ Computing average images...")
fig, axes = plt.subplots(1, 2, figsize=(10, 5))
fig.suptitle('Average X-Ray Image per Class', fontweight='bold')
for i, cls in enumerate(class_names):
    cls_paths = df_all[df_all['label'] == cls]['path'].tolist()
    if not cls_paths:
        axes[i].text(0.5, 0.5, 'No Images', ha='center', va='center', transform=axes[i].transAxes)
        axes[i].set_title(cls); axes[i].axis('off')
        continue
    cls_paths = cls_paths[:200] # Limit for speed
    avg_img = np.zeros((224, 224), dtype=np.float32)
    valid_count = 0
    for p in cls_paths:
        img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if img is not None: avg_img += cv2.resize(img, (224, 224)); valid_count += 1
    if valid_count > 0:
        axes[i].imshow((avg_img / valid_count).astype(np.uint8), cmap='gray')
    axes[i].set_title(f'{cls}'); axes[i].axis('off')
plt.tight_layout(); plt.savefig(PLTDIR / '14_average_images.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 15: PIXEL INTENSITY DISTRIBUTION
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 6))
for i, cls in enumerate(class_names):
    cls_paths = df_all[df_all['label'] == cls]['path'].sample(n=min(50, len(df_all[df_all['label']==cls])), random_state=42).tolist()
    if not cls_paths: continue
    pixels = []
    for p in cls_paths:
        img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if img is not None: pixels.extend(np.random.choice(img.flatten(), size=1000, replace=False))
    if pixels: sns.kdeplot(pixels, fill=True, color=COLORS[i], label=cls, ax=ax, alpha=0.3, linewidth=2)
ax.set_title('Pixel Intensity Distribution by Class', fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '15_pixel_intensity.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 16: IMAGE DIMENSIONS
# ─────────────────────────────────────────────────────────────────────────────────
dim_data = []
for p in df_all['path'].sample(n=min(500, len(df_all)), random_state=42):
    img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
    if img is not None:
        h, w = img.shape[:2]; lbl = df_all.loc[df_all['path']==p, 'label'].values[0]
        dim_data.append({'Height': h, 'Width': w, 'Label': lbl})
if dim_data:
    dim_df = pd.DataFrame(dim_data)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    fig.suptitle('Image Dimension Analysis', fontweight='bold')
    sns.boxplot(data=dim_df, x='Label', y='Height', palette=COLORS, ax=axes[0]); axes[0].set_title('Image Height')
    sns.boxplot(data=dim_df, x='Label', y='Width', palette=COLORS, ax=axes[1]); axes[1].set_title('Image Width')
    plt.tight_layout(); plt.savefig(PLTDIR / '16_image_dimensions.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#               PLOT 17: XAI GRAD-CAM + ATTENTION (5 SAMPLES)
# ─────────────────────────────────────────────────────────────────────────────────
print("\n🔥 Generating XAI Grad-CAM + Attention Overlays...")
class AdvancedXAI:
    def __init__(self, model):
        self.model = model; self.model.eval(); self.grads, self.activations = None, None
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        if last_conv:
            last_conv.register_forward_hook(lambda m, i, o: setattr(self, 'activations', o.detach()))
            last_conv.register_backward_hook(lambda m, gi, go: setattr(self, 'grads', go[0].detach()))

    def generate(self, img_path, target_class):
        img_gray = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        rgb = np.stack([img_gray, img_gray, img_gray], axis=-1)
        orig_h, orig_w = rgb.shape[:2]
        x = get_tensor(img_gray).float().requires_grad_(True)

        out = self.model(x); self.model.zero_grad(); oh = torch.zeros_like(out); oh[0, target_class] = 1
        out.backward(gradient=oh, retain_graph=True)

        if self.grads is not None and self.activations is not None:
            weights = self.grads.mean(dim=(2, 3), keepdim=True)
            cam = F.relu((weights * self.activations).sum(1)).squeeze().cpu().numpy()
            cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8) if cam.max() > cam.min() else np.zeros_like(cam)
        else: cam = np.zeros((self.model.spatial_h, self.model.spatial_w))

        with torch.no_grad(): _ = self.model(x)
        g_attn, _ = self.model.get_attentions()
        attn = np.zeros((self.model.spatial_h, self.model.spatial_w))
        if g_attn:
            a = g_attn[-1]; attn = a[0, :, 0, 1:].mean(0).cpu().numpy() if a.dim() == 4 else a[0, 0, 1:].cpu().numpy()
            attn = attn.reshape(self.model.spatial_h, self.model.spatial_w)
            attn = (attn - attn.min()) / (attn.max() - attn.min() + 1e-8) if attn.max() > attn.min() else np.zeros_like(attn)

        def overlay(hmap):
            if hmap.max() == 0: return rgb.copy()
            hr = cv2.resize(hmap, (orig_w, orig_h)); colored = cv2.applyColorMap((hr * 255).astype(np.uint8), cv2.COLORMAP_JET)
            return cv2.addWeighted(rgb, 0.6, colored, 0.4, 0)

        comb = 0.5 * cam + 0.5 * attn
        comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8) if comb.max() > comb.min() else comb
        return img_gray, overlay(cam), overlay(attn), overlay(comb)

xai = AdvancedXAI(model)
sample_indices = []
for cls_idx in range(2):
    mask = (all_p == cls_idx) & (all_l == cls_idx)
    if mask.any(): sample_indices.append(np.where(mask)[0][0])
sample_indices = sample_indices[:5]

fig, axes = plt.subplots(len(sample_indices), 4, figsize=(16, 4 * len(sample_indices)))
fig.suptitle('Model Explainability: Grad-CAM vs Attention vs Combined', fontsize=18, fontweight='bold', y=1.01)
if len(sample_indices) == 1: axes = axes.reshape(1, -1)
for row_idx, idx in enumerate(sample_indices):
    gray, cam_rgb, attn_rgb, comb_rgb = xai.generate(test_s[idx][0], all_l[idx])
    imgs = [gray, cam_rgb, attn_rgb, comb_rgb]; cmaps = ['gray', None, None, None]
    for col_idx, (img, cm_) in enumerate(zip(imgs, cmaps)):
        axes[row_idx, col_idx].imshow(img, cmap=cm_)
        if col_idx == 0: axes[row_idx, col_idx].set_title(f"True: {class_names[all_l[idx]]}\nPred: {class_names[all_p[idx]]} ({max_probs[idx]:.1%})", fontweight='bold')
        else: axes[row_idx, col_idx].set_title(['Original', 'Grad-CAM', 'Attention', 'Fused'][col_idx], fontweight='bold')
        axes[row_idx, col_idx].axis('off')
plt.tight_layout(); plt.savefig(PLTDIR / '17_xai_5_samples.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#               PLOT 18: DEPLOYMENT THRESHOLDS
# ─────────────────────────────────────────────────────────────────────────────────
thresholds = np.arange(0.5, 1.0, 0.05); metrics_data = []
for thresh in thresholds:
    mask = max_probs >= thresh
    if mask.sum() == 0: continue
    p_thresh, l_thresh = all_p[mask], all_l[mask]
    metrics_data.append({'Threshold': thresh, 'Accuracy': accuracy_score(l_thresh, p_thresh), 'F1-Score': f1_score(l_thresh, p_thresh, average='weighted', zero_division=0), 'Coverage': mask.sum() / len(all_p) * 100})
df_thresh = pd.DataFrame(metrics_data)
fig, ax1 = plt.subplots(figsize=(10, 6))
ax1.set_xlabel('Minimum Confidence Threshold'); ax1.set_ylabel('Metric Score', color='#3498db')
ax1.plot(df_thresh['Threshold'], df_thresh['Accuracy'], color='#3498db', marker='o', lw=2, label='Accuracy')
ax1.plot(df_thresh['Threshold'], df_thresh['F1-Score'], color='#2ecc71', marker='s', lw=2, label='F1-Score')
ax1.tick_params(axis='y', labelcolor='#3498db'); ax1.grid(True, alpha=0.3)
ax2 = ax1.twinx(); ax2.set_ylabel('Dataset Coverage (%)', color='#e74c3c')
ax2.plot(df_thresh['Threshold'], df_thresh['Coverage'], color='#e74c3c', marker='^', lw=2, linestyle='--', label='Coverage')
ax2.tick_params(axis='y', labelcolor='#e74c3c'); fig.legend(loc='lower left', bbox_to_anchor=(0.1, -0.1), ncol=3)
plt.tight_layout(); plt.savefig(PLTDIR / '18_deployment_thresholds.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#               PLOT 19: CLASS-SPECIFIC THRESHOLDS
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 6))
cls_threshs = np.arange(0.4, 1.0, 0.05)
for cls_idx, cls in enumerate(class_names):
    cls_f1s = []
    for t in cls_threshs:
        mask = (all_p == cls_idx) & (max_probs >= t)
        cls_f1s.append(f1_score(all_l[mask] == cls_idx, all_p[mask] == cls_idx, zero_division=0) if mask.sum() > 0 else 0)
    opt_idx = np.argmax(cls_f1s)
    ax.plot(cls_threshs, cls_f1s, color=COLORS[cls_idx], marker='o', lw=2, label=f'{cls} (Opt @ {cls_threshs[opt_idx]:.2f})')
ax.set_title('Class-Specific F1 vs. Confidence Threshold', fontweight='bold'); ax.set_xlabel('Confidence Threshold'); ax.set_ylabel('F1-Score')
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '19_class_specific_thresholds.png', dpi=300, bbox_inches='tight'); plt.show()

# ─────────────────────────────────────────────────────────────────────────────────
#               PLOT 20: DEEP FEATURE MAPS HIERARCHY
# ─────────────────────────────────────────────────────────────────────────────────
print("\n🧠 Extracting Hierarchical Feature Maps...")
mask_idx = np.where((all_l == 1) & (all_p == 1))[0]
if len(mask_idx) == 0: mask_idx = np.where((all_l == 0) & (all_p == 0))[0]
if len(mask_idx) == 0: mask_idx = [0]
img = cv2.imread(test_s[mask_idx[0]][0], cv2.IMREAD_GRAYSCALE)
x_feat = get_tensor(img)
feature_maps = {}
def get_hook(name):
    def hook(module, input, output): feature_maps[name] = output.detach().cpu()
    return hook
model.backbone.feature_extractor[1].register_forward_hook(get_hook('Early_Layer'))
model.backbone.feature_extractor[4].register_forward_hook(get_hook('Mid_Layer'))
model.backbone.feature_extractor[7].register_forward_hook(get_hook('Late_Layer'))
with torch.no_grad(): _ = model(x_feat)

fig, axes = plt.subplots(3, 8, figsize=(20, 7))
fig.suptitle('Hierarchical Feature Extraction', fontsize=16, fontweight='bold')
for row_idx, (layer, title) in enumerate(zip(['Early_Layer', 'Mid_Layer', 'Late_Layer'], ['Early Features (Edges)', 'Mid Features (Anatomy)', 'Late Features (Pathology)'])):
    if layer in feature_maps:
        fm = feature_maps[layer].squeeze(); n_channels = fm.shape[0]
        channel_idxs = np.linspace(0, n_channels - 1, 8, dtype=int)
        for col_idx, c_idx in enumerate(channel_idxs):
            if c_idx < n_channels:
                ch_img = fm[c_idx].numpy(); ch_img = (ch_img - ch_img.min()) / (ch_img.max() - ch_img.min() + 1e-8)
                axes[row_idx, col_idx].imshow(ch_img, cmap='magma'); axes[row_idx, col_idx].set_title(f'Ch {c_idx}', fontsize=9); axes[row_idx, col_idx].axis('off')
    axes[row_idx, -1].axis('off')
    axes[row_idx, -1].text(1.1, 0.5, title, transform=axes[row_idx, -1].transAxes, fontsize=12, fontweight='bold', va='center', rotation=270)
plt.tight_layout(); plt.savefig(PLTDIR / '20_feature_maps_hierarchy.png', dpi=300, bbox_inches='tight'); plt.show()


# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  EXACT FINAL EVALUATION METRICS BLOCK (BINARY FORMAT)                          ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

test_acc = accuracy_score(all_l, all_p)
test_f1 = f1_score(all_l, all_p, average='weighted')
test_auc = roc_auc_score(all_l, all_pr[:, 1])
test_ap = average_precision_score(all_l, all_pr[:, 1])

mcc = matthews_corrcoef(all_l, all_p)
kappa = cohen_kappa_score(all_l, all_p)
ll = log_loss(all_l, all_pr)
brier = brier_score_loss(all_l, all_pr[:, 1])

# Binary Metrics Calculation (Class 1 = Fractured is Positive)
tp = cm[1, 1]; fp = cm[0, 1]; fn = cm[1, 0]; tn = cm[0, 0]
sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0.0
specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0
npv = tn / (tn + fn) if (tn + fn) > 0 else 0.0
fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0
fnr = fn / (fn + tp) if (fn + tp) > 0 else 0.0

print("\n============================================================")
print("📊 COMPREHENSIVE EVALUATION METRICS")
print("============================================================")

print(f"\n🎯 PRIMARY METRICS:")
print(f"   Accuracy:  {test_acc:.4f}")
print(f"   F1-Score:  {test_f1:.4f}")
print(f"   AUC-ROC:   {test_auc:.4f}")
print(f"   AP Score:  {test_ap:.4f}")

print(f"\n📈 ADDITIONAL METRICS:")
print(f"   MCC:       {mcc:.4f}")
print(f"   Cohen's Kappa: {kappa:.4f}")
print(f"   Log Loss:  {ll:.4f}")
print(f"   Brier Score: {brier:.4f}")

print(f"\n🔬 PER-CLASS METRICS (Not Fractured vs Fractured):")
print(f"   Sensitivity (Recall): {sensitivity:.4f}")
print(f"   Specificity:          {specificity:.4f}")
print(f"   Precision:            {precision:.4f}")
print(f"   NPV:                  {npv:.4f}")
print(f"   FPR:                  {fpr:.4f}")
print(f"   FNR:                  {fnr:.4f}")

print(f"\n📊 CONFUSION MATRIX:")
print(f"{'':18s}{'Not Fractured':>15s}{'Fractured':>12s}")
print(f"{'Not Fractured':18s}{tn:>15d}{fp:>12d}")
print(f"{'Fractured':18s}{fn:>15d}{tp:>12d}")

print(f"\n📋 CLASSIFICATION REPORT:")
print(classification_report(all_l, all_p, target_names=class_names))

print("="*60)
print(f"✅ ALL 20 PLOTS GENERATED SUCCESSFULLY FOR FRACTURE DATASET!")
print(f"📁 Location: {PLTDIR}")
print("="*60)
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 15 — Download Entire Kaggle Working Space (FIXED)                        ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

import shutil
from IPython.display import FileLink

def download_workspace(output_name='bone_fracture.zip'):
    """Zip the entire /kaggle/working directory and provide download link."""
    
    zip_path = OUTDIR / output_name
    
    # Remove old zip if exists
    if zip_path.exists():
        zip_path.unlink()
    
    print("📦 Scanning /kaggle/working for files to archive...")
    
    total_size = 0
    for item in OUTDIR.rglob('*'):
        if item.is_file() and item != zip_path:
            size_mb = item.stat().st_size / (1024 * 1024)
            total_size += size_mb
            if size_mb > 0.1:
                rel_path = item.relative_to(OUTDIR)
                print(f"      📄 {rel_path}: {size_mb:.2f} MB")
    
    print(f"\n   Total uncompressed size: {total_size:.2f} MB")
    
    # ✅ FIXED: Removed 'with' statement
    print("\n🔒 Compressing... (this may take a minute for large models)")
    archive_name = str(zip_path.with_suffix(''))
    shutil.make_archive(archive_name, 'zip', OUTDIR)
    
    # Re-find the zip in case of naming quirks
    final_zip = Path(archive_name + '.zip')
    final_size = final_zip.stat().st_size / (1024 * 1024)
    
    print(f"✅ Archive created successfully!")
    print(f"   File: {final_zip.name}")
    print(f"   Size: {final_size:.2f} MB")
    
    # Provide download link
    print("\n⬇️ Click the link below to download:")
    display(FileLink(str(final_zip), result_html_prefix="🔗 Download Link: "))
    
    return str(final_zip)

# Run it
zip_file = download_workspace()
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  PART 1: SETUP, HYPERPARAMETERS & TRANSFORMS                                   ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

!pip install -q timm einops albumentations opencv-python-headless

import os, time, random, copy, warnings
from pathlib import Path
from collections import Counter
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageFile
import cv2
from tqdm.auto import tqdm
from scipy.ndimage import zoom
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler
from torch.cuda.amp import GradScaler, autocast
import torchvision.models as models
import albumentations as A
from albumentations.pytorch import ToTensorV2
from einops import rearrange
from sklearn.metrics import (accuracy_score, f1_score, roc_curve, auc, confusion_matrix,
                             classification_report, precision_recall_curve, average_precision_score,
                             roc_auc_score, matthews_corrcoef, cohen_kappa_score, log_loss, brier_score_loss)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE
import ipywidgets as widgets
from IPython.display import display, clear_output
import io

ImageFile.LOAD_TRUNCATED_IMAGES = True 
warnings.filterwarnings('ignore')

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True; torch.backends.cudnn.benchmark = False

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
CKDIR = OUTDIR / 'checkpoints'
PLTDIR = OUTDIR / 'plots/fracatlas_final'
for d in [CKDIR, PLTDIR]: d.mkdir(exist_ok=True, parents=True)

MEAN, STD = (0.485, 0.456, 0.406), (0.229, 0.224, 0.225)
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.JPG', '.JPEG', '.PNG'}
class_names = ['Not Fractured', 'Fractured']
COLORS = ['#2ecc71', '#e74c3c']

HPARAMS = {
    'img_size': 224, 'batch_size': 32, 'num_workers': 4, 'num_epochs': 40,
    'lr_backbone': 5e-6, 'lr_head': 1e-4, 'weight_decay': 1e-4,
    'freeze_epochs': 5, 'patience': 10, 'aug_level': 'light',
    'd_model': 384, 'nhead': 6, 'n_layers': 4, 'dim_ffn': 1536,
    'label_smoothing': 0.1, 'mixup_alpha': 0.2, 'cutmix_alpha': 1.0, 'mixup_prob': 0.3,
    'backbone': 'convnext_base', 'num_classes': 2, 'class_names': class_names
}
print(f"✅ Setup complete on {DEVICE}\n✅ Hyperparameters ready (FracAtlas)")

def get_xray_transforms(split, img_size=224, level='light'):
    norm = [A.Normalize(mean=MEAN, std=STD), ToTensorV2()]
    if split != 'train': return A.Compose([A.Resize(height=img_size, width=img_size)] + norm)
    base = [A.Resize(height=img_size, width=img_size), A.HorizontalFlip(p=0.5), A.Rotate(limit=15, p=0.5, border_mode=cv2.BORDER_CONSTANT, value=0)]
    light = [A.RandomBrightnessContrast(brightness_limit=0.15, contrast_limit=0.15, p=0.4), A.GaussNoise(var_limit=(10.0, 30.0), p=0.15)]
    medium = [A.ShiftScaleRotate(shift_limit=0.1, scale_limit=0.1, rotate_limit=10, p=0.4, border_mode=cv2.BORDER_CONSTANT, value=0)]
    heavy = [A.ElasticTransform(alpha=30, sigma=3, p=0.1), A.CoarseDropout(max_holes=6, max_height=int(img_size*0.08), max_width=int(img_size*0.08), p=0.15, fill_value=0)]
    aug = base + light if level == 'light' else base + light + medium if level == 'medium' else base + light + medium + heavy
    return A.Compose(aug + norm)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  PART 2: MODEL ARCHITECTURE                                                    ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class SmoothCELoss(nn.Module):
    def __init__(self, label_smoothing=0.1): super().__init__(); self.criterion = nn.CrossEntropyLoss(label_smoothing=label_smoothing)
    def forward(self, inputs, targets): return self.criterion(inputs, targets)

class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): super().__init__(); self.drop_prob = drop_prob
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training: return x
        return x * (torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * float(1.0 - self.drop_prob))

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3); self.proj = nn.Linear(d_model, d_model)
        self.attn_drop, self.proj_drop, self.attn_weights = nn.Dropout(dropout), nn.Dropout(dropout), None
    def forward(self, x, H, W):
        B, N, C = x.shape; x = x.reshape(B, H, W, C)
        pad_h, pad_w = (self.window_size - H % self.window_size) % self.window_size, (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0: x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(-1, self.window_size**2, C)
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2,0,3,1,4)
        attn = (qkv[0] @ qkv[1].transpose(-2,-1)) * self.scale; attn = attn.softmax(dim=-1); self.attn_weights = attn.detach()
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1,2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(B, Hp, Wp, C)
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model*2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop), nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop))
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights, self.window_attn_weights = None, None
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        window_out = self.window_attn(x_norm[:, 1:, :], H, W); self.window_attn_weights = self.window_attn.attn_weights
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1-gate) * window_out], dim=1)
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self, backbone_name='convnext_base'):
        super().__init__()
        self.backbone = models.convnext_base(weights=None) if backbone_name == 'convnext_base' else models.convnext_large(weights=None)
        self.feature_extractor = self.backbone.features
    def forward(self, x): return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes=2, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536, backbone_name='convnext_base'):
        super().__init__(); self.d_model = d_model; self.img_size = img_size; self.backbone = ImprovedBackbone(backbone_name)
        with torch.no_grad(): _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        self.spatial_h, self.spatial_w = h, w
        self.proj = nn.Sequential(nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(), nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU())
        self.norm_proj = nn.LayerNorm(d_model); self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h*w + 1, d_model))
        nn.init.trunc_normal_(self.cls_token, std=0.02); nn.init.trunc_normal_(self.pos_embed, std=0.02)
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i]) for i in range(n_layers)])
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(nn.LayerNorm(d_model), nn.Linear(d_model, d_model//2), nn.GELU(), nn.Dropout(0.3), nn.Linear(d_model//2, d_model//4), nn.GELU(), nn.Dropout(0.15), nn.Linear(d_model//4, num_classes))
    def forward(self, x):
        B = x.size(0); tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens); tokens = torch.cat([self.cls_token.expand(B,-1,-1), tokens], dim=1) + self.pos_embed
        for block in self.encoder: tokens = block(tokens, self.spatial_h, self.spatial_w)
        return self.head(self.norm(tokens)[:, 0])
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w
    def unfreeze_backbone_partial(self, n=3):
        for p in self.backbone.feature_extractor.parameters(): p.requires_grad_(False)
        for child in list(self.backbone.feature_extractor.children())[-n:]:
            for p in child.parameters(): p.requires_grad_(True)

def get_tensor(img):
    if len(img.shape) == 2: img = np.stack([img]*3, axis=-1)
    t = torch.from_numpy(cv2.resize(img, (224, 224))).float().permute(2,0,1)/255.0
    return ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)

print("✅ Model Architecture ready")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  PART 3: DATA LOADING, MIXUP, TRAINER                                         ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class XRayDataset(Dataset):
    def __init__(self, samples, class_names, transform=None):
        self.samples, self.transform, self.class_names = samples, transform, class_names
        self.class_to_idx = {n: i for i, n in enumerate(class_names)}
    def __len__(self): return len(self.samples)
    def __getitem__(self, idx):
        p, l = self.samples[idx]; l = self.class_to_idx.get(l, l)
        img = cv2.imread(p)
        if img is None: img = np.array(Image.open(p).convert('RGB'))
        else: img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        if self.transform: img = self.transform(image=img)['image']
        elif not isinstance(img, torch.Tensor):
            if img.max() > 1: img = img / 255.0
            img = torch.from_numpy(img).float().permute(2, 0, 1)
        return img.float(), l
    def class_weights(self):
        labels = [self.class_to_idx.get(l, l) for _, l in self.samples]
        counts = torch.bincount(torch.tensor(labels), minlength=len(self.class_names)).float()
        beta = 0.9999; effective_num = 1.0 - torch.pow(beta, counts)
        return ((1.0 - beta) / (effective_num + 1e-8)) / len(self.class_names) * len(self.class_names)

# def load_fracatlas():
#     samples = []
#     base = Path('/kaggle/input/datasets/aayush4204/daataaset/Fracture Dataset/Fracture Dataset/FracAtlas/images')
#     print(f"📂 Loading FracAtlas dataset...\n✅ Found dataset at: {base}")
#     for cls_dir in sorted(base.iterdir()):
#         if not cls_dir.is_dir(): continue
#         name_lower = cls_dir.name.lower().replace('-', ' ').replace('_', ' ')
#         cname = 'Fractured' if ('fractured' in name_lower) else 'Not Fractured'
#         count = 0
#         for ext in VALID_EXTS:
#             for p in cls_dir.rglob(f'*{ext}'):
#                 samples.append((str(p), cname)); count += 1
#         print(f"   ✅ {cname.ljust(15)}: {count:>4d} images")
#     return samples

def load_fracatlas():
    """Load FracAtlas dataset with exact path"""
    samples = []
    class_names = ['Fractured', 'Not Fractured']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Fracture Dataset/Fracture Dataset/FracAtlas/images')
    
    if not base.exists():
        print(f"❌ FracAtlas dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading FracAtlas dataset...\n✅ Found dataset at: {base}")
    
    for cls_dir in sorted(base.iterdir()):
        if not cls_dir.is_dir(): 
            continue
        name_lower = cls_dir.name.lower().replace('-', ' ').replace('_', ' ')
        cname = 'Fractured' if ('fractured' in name_lower) else 'Not Fractured'
        count = 0
        for ext in VALID_EXTS:
            for p in cls_dir.rglob(f'*{ext}'):
                samples.append((str(p), cname))
                count += 1
        print(f"   ✅ {cname.ljust(15)}: {count:>4d} images")
    
    return samples

def mixup_data(x, y, alpha=0.2):
    lam = np.random.beta(alpha, alpha); idx = torch.randperm(x.size(0)).to(x.device)
    return lam * x + (1 - lam) * x[idx], y, y[idx], lam

def cutmix_data(x, y, alpha=1.0):
    lam = np.random.beta(alpha, alpha); idx = torch.randperm(x.size(0)).to(x.device)
    W, H = x.size(2), x.size(3); r = np.sqrt(1.0 - lam)
    cw, ch = int(W * r), int(H * r); cx, cy = np.random.randint(W), np.random.randint(H)
    x1, y1 = np.clip(cx - cw // 2, 0, W), np.clip(cy - ch // 2, 0, H)
    x2, y2 = np.clip(cx + cw // 2, 0, W), np.clip(cy + ch // 2, 0, H)
    mixed = x.clone(); mixed[:, :, x1:x2, y1:y2] = x[idx, :, x1:x2, y1:y2]
    return mixed, y, y[idx], 1 - (x2 - x1) * (y2 - y1) / (W * H)

class Trainer:
    def __init__(self, model, loaders, hparams):
        self.model, self.loaders, self.hparams = model, loaders, hparams
        self.criterion = SmoothCELoss(hparams['label_smoothing'])
        bp = [p for n, p in model.named_parameters() if p.requires_grad and 'backbone' in n]
        hp = [p for n, p in model.named_parameters() if p.requires_grad and 'backbone' not in n]
        self.optimizer = optim.AdamW([{'params': bp, 'lr': hparams['lr_backbone']}, {'params': hp, 'lr': hparams['lr_head']}], weight_decay=hparams['weight_decay'])
        self.scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(self.optimizer, T_0=hparams['num_epochs'] // 3, T_mult=2, eta_min=1e-7)
        self.scaler = GradScaler()
        self.history = {k: [] for k in ['train_loss', 'val_loss', 'train_acc', 'val_acc', 'val_f1', 'lr']}
        self.best_val_f1, self.best_state, self.patience_counter, self.unfrozen = 0, None, 0, False

    def train_one_epoch(self, epoch):
        self.model.train(); total_loss, preds, labs = 0, [], []
        for images, labels in tqdm(self.loaders['train'], desc=f'Ep {epoch+1}', leave=False):
            images, labels = images.to(DEVICE, non_blocking=True), labels.to(DEVICE, non_blocking=True)
            self.optimizer.zero_grad(set_to_none=True); mixed = False
            if np.random.random() < self.hparams['mixup_prob']:
                if np.random.random() < 0.5: images, la, lb, lam = mixup_data(images, labels, self.hparams['mixup_alpha'])
                else: images, la, lb, lam = cutmix_data(images, labels, self.hparams['cutmix_alpha'])
                mixed = True
            with autocast():
                out = self.model(images)
                loss = lam * self.criterion(out, la) + (1 - lam) * self.criterion(out, lb) if mixed else self.criterion(out, labels)
            self.scaler.scale(loss).backward(); self.scaler.unscale_(self.optimizer)
            torch.nn.utils.clip_grad_norm_(self.model.parameters(), max_norm=1.0); self.scaler.step(self.optimizer); self.scaler.update()
            total_loss += loss.item()
            if not mixed: preds.extend(out.argmax(1).cpu().numpy()); labs.extend(labels.cpu().numpy())
        return total_loss / len(self.loaders['train']), accuracy_score(labs, preds) if labs else 0, f1_score(labs, preds, average='weighted', zero_division=0) if labs else 0

    @torch.no_grad()
    def validate(self, loader):
        self.model.eval(); total_loss, preds, labs, probs = 0, [], [], []
        for images, labels in tqdm(loader, desc='Val', leave=False):
            images, labels = images.to(DEVICE, non_blocking=True), labels.to(DEVICE, non_blocking=True)
            with autocast(): out = self.model(images); loss = self.criterion(out, labels)
            total_loss += loss.item(); p = F.softmax(out, dim=1)
            preds.extend(out.argmax(1).cpu().numpy()); labs.extend(labels.cpu().numpy()); probs.extend(p.cpu().numpy())
        return total_loss / len(loader), accuracy_score(labs, preds), f1_score(labs, preds, average='weighted', zero_division=0), 0, preds, labs, probs

    def train(self):
        print(f"\n{'='*60}\nStarting training for {self.hparams['num_epochs']} epochs\n{'='*60}\n")
        for epoch in range(self.hparams['num_epochs']):
            if epoch == self.hparams['freeze_epochs'] and not self.unfrozen:
                print("\n🔄 Unfreezing backbone..."); self.model.unfreeze_backbone_partial(3); self.unfrozen = True
            tl, ta, tf = self.train_one_epoch(epoch); vl, va, vf, _, _, _, _ = self.validate(self.loaders['val']); self.scheduler.step()
            self.history['train_loss'].append(tl); self.history['val_loss'].append(vl); self.history['train_acc'].append(ta)
            self.history['val_acc'].append(va); self.history['val_f1'].append(vf); self.history['lr'].append(self.optimizer.param_groups[0]['lr'])
            print(f"Ep {epoch+1:2d}: T_Loss={tl:.3f} T_Acc={ta:.3f} | V_Loss={vl:.3f} V_Acc={va:.3f} V_F1={vf:.3f}")
            if vf > self.best_val_f1:
                self.best_val_f1 = vf; self.best_state = copy.deepcopy(self.model.state_dict()); self.patience_counter = 0
                torch.save({'model': self.model.state_dict(), 'f1': vf, 'acc': va}, CKDIR / 'best_model.pth'); print(f"   ✅ New best F1: {vf:.4f}")
            else:
                self.patience_counter += 1
                if self.patience_counter >= self.hparams['patience']: print(f"\n⏹️ Early stopping at epoch {epoch+1}"); break
        return self.history

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  PART 4: MAIN EXECUTION (TRAINING)                                             ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def main():
    IMG_SIZE, BATCH_SIZE, NUM_WORKERS = HPARAMS['img_size'], HPARAMS['batch_size'], HPARAMS['num_workers']
    all_samples = load_fracatlas()
    labels = [class_names.index(l) for _, l in all_samples]
    
    train_s, temp_s = train_test_split(all_samples, test_size=0.2, stratify=labels, random_state=42)
    val_s, test_s = train_test_split(temp_s, test_size=0.5, stratify=[class_names.index(l) for _, l in temp_s], random_state=42)
    print(f"\n📊 Created splits:\n   - Train:       {len(train_s)} images (70.0%)\n   - Validation:   {len(val_s)} images (10.0%)\n   - Test:         {len(test_s)} images (20.0%)\n")
    print("📊 Creating datasets...")

    train_ds = XRayDataset(train_s, class_names, get_xray_transforms('train', IMG_SIZE, HPARAMS['aug_level']))
    val_ds = XRayDataset(val_s, class_names, get_xray_transforms('val', IMG_SIZE))
    test_ds = XRayDataset(test_s, class_names, get_xray_transforms('test', IMG_SIZE))
    
    sw = [train_ds.class_weights()[class_names.index(l)] for _, l in train_s]
    loaders = {
        'train': DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=WeightedRandomSampler(sw, len(sw), replacement=True), num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=True),
        'val': DataLoader(val_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=True),
        'test': DataLoader(test_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS, pin_memory=True, persistent_workers=True)
    }
    
    print("🏗️ Building Model...")
    model = EnhancedHViT(num_classes=HPARAMS['num_classes'], img_size=IMG_SIZE, d_model=HPARAMS['d_model'], nhead=HPARAMS['nhead'], n_layers=HPARAMS['n_layers'], dim_ffn=HPARAMS['dim_ffn'], backbone_name=HPARAMS['backbone']).to(DEVICE)
    trainer = Trainer(model, loaders, HPARAMS)
    history = trainer.train()
    
    print("\n📊 Final Evaluation...")
    model.load_state_dict(torch.load(CKDIR / 'best_model.pth', map_location=DEVICE, weights_only=False)['model']); model.eval()
    all_p, all_l, all_pr = [], [], []
    with torch.no_grad():
        for im, l in tqdm(loaders['test'], desc='Final Test'):
            with autocast(): out = model(im.to(DEVICE))
            p = F.softmax(out, dim=1); all_p.extend(out.argmax(1).cpu().numpy()); all_l.extend(l.numpy()); all_pr.extend(p.cpu().numpy())
    
    test_acc = accuracy_score(all_l, all_p); test_f1 = f1_score(all_l, all_p, average='weighted')
    print(f"\n{'='*50}\n🎯 FINAL TEST RESULTS\n   Accuracy: {test_acc:.4f}\n   F1 Score: {test_f1:.4f}\n{'='*50}")
    print(classification_report(all_l, all_p, target_names=class_names))
    return history, test_s

print("\n" + "🚀 STARTING FULL PIPELINE ".center(60, "=") + "\n")
history, test_s = main()

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  PART 5: ULTIMATE MEGA CELL - ALL 20 PLOTS + EXACT METRICS                     ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

print("\n" + "📈 GENERATING 20 PLOTS ".center(60, "=") + "\n")
model = EnhancedHViT(num_classes=2).to(DEVICE)
model.load_state_dict(torch.load(CKDIR / 'best_model.pth', map_location=DEVICE, weights_only=False)['model'])
model.eval()

# Re-create splits exactly as in training for Plot 13 consistency
all_samples = load_fracatlas()
labels = [class_names.index(l) for _, l in all_samples]
train_s, temp_s = train_test_split(all_samples, test_size=0.2, stratify=labels, random_state=42)
val_s, test_s = train_test_split(temp_s, test_size=0.5, stratify=[class_names.index(l) for _, l in temp_s], random_state=42)

all_p, all_l, all_pr = [], [], []
for p, l in tqdm(test_s, desc="Predicting"):
    img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
    if img is None: continue
    with torch.no_grad(): out = model(get_tensor(img)); pr = F.softmax(out, dim=1)
    all_p.append(out.argmax(1).item()); all_l.append(class_names.index(l)); all_pr.append(pr.cpu().numpy()[0])

all_p, all_l, all_pr = np.array(all_p), np.array(all_l), np.array(all_pr)
max_probs = all_pr.max(axis=1); correct_mask = (all_p == all_l)
labels_bin = label_binarize(all_l, classes=[0, 1])
cm = confusion_matrix(all_l, all_p); cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

# --- PLOTS 1 to 12 ---
try:
    if len(history.get('train_loss', [])) > 0:
        fig, ax = plt.subplots(2, 2, figsize=(14, 10)); fig.suptitle('FracAtlas Training Dynamics', fontsize=16, fontweight='bold')
        ax[0,0].plot(history['train_loss'], label='Train', color='red'); ax[0,0].plot(history['val_loss'], label='Val', color='blue'); ax[0,0].set_title('Loss'); ax[0,0].legend(); ax[0,0].grid(True, alpha=0.3)
        ax[0,1].plot(history['train_acc'], label='Train', color='red'); ax[0,1].plot(history['val_acc'], label='Val', color='blue'); ax[0,1].set_title('Accuracy'); ax[0,1].legend(); ax[0,1].grid(True, alpha=0.3)
        ax[1,0].plot(history['val_f1'], label='Val F1', color='green'); ax[1,0].set_title('F1 Score'); ax[1,0].legend(); ax[1,0].grid(True, alpha=0.3)
        ax[1,1].plot(history['lr'], color='purple'); ax[1,1].set_title('Learning Rate'); ax[1,1].set_yscale('log'); ax[1,1].grid(True, alpha=0.3)
        plt.tight_layout(); plt.savefig(PLTDIR / '1_training_history.png', dpi=300, bbox_inches='tight'); plt.show()
except: pass

fig, axes = plt.subplots(1, 2, figsize=(14, 6)); fig.suptitle('FracAtlas Confusion Matrix', fontsize=16, fontweight='bold')
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names, ax=axes[0]); axes[0].set_title('Absolute Counts'); axes[0].set_ylabel('True'); axes[0].set_xlabel('Predicted')
sns.heatmap(cm_norm, annot=True, fmt='.2%', cmap='Greens', xticklabels=class_names, yticklabels=class_names, ax=axes[1]); axes[1].set_title('Normalized'); axes[1].set_ylabel('True'); axes[1].set_xlabel('Predicted')
plt.tight_layout(); plt.savefig(PLTDIR / '2_confusion_matrices.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(8, 8))
fpr, tpr, _ = roc_curve(all_l, all_pr[:, 1]); roc_auc = auc(fpr, tpr)
ax.plot(fpr, tpr, color=COLORS[1], lw=2, label=f'Fractured (AUC = {roc_auc:.3f})'); ax.plot([0, 1], [0, 1], 'k--', lw=1)
ax.set_title('ROC Curve', fontweight='bold'); ax.set_xlabel('FPR'); ax.set_ylabel('TPR'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '3_roc_curve.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(8, 8))
prec, rec, _ = precision_recall_curve(all_l, all_pr[:, 1]); ap = average_precision_score(all_l, all_pr[:, 1])
ax.plot(rec, prec, color=COLORS[1], lw=2, label=f'Fractured (AP = {ap:.3f})'); ax.set_title('Precision-Recall', fontweight='bold')
ax.set_xlabel('Recall'); ax.set_ylabel('Precision'); ax.legend(); ax.grid(True, alpha=0.3); ax.set_xlim([0,1]); ax.set_ylim([0,1.05])
plt.tight_layout(); plt.savefig(PLTDIR / '4_precision_recall.png', dpi=300, bbox_inches='tight'); plt.show()

report = classification_report(all_l, all_p, target_names=class_names, output_dict=True); m_df = pd.DataFrame(report).T.iloc[:-3, :3]
fig, ax = plt.subplots(figsize=(8, 6)); x = np.arange(2); width = 0.25
ax.bar(x - width, m_df['precision'], width, label='Precision', color='#3498db', edgecolor='black')
ax.bar(x, m_df['recall'], width, label='Recall', color='#2ecc71', edgecolor='black')
ax.bar(x + width, m_df['f1-score'], width, label='F1-Score', color='#e74c3c', edgecolor='black')
ax.set_title('Per-Class Metrics', fontweight='bold'); ax.set_xticks(x); ax.set_xticklabels(class_names); ax.legend(); ax.set_ylim(0,1.15); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '5_per_class_metrics.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
ax.hist(max_probs[correct_mask], bins=20, alpha=0.7, label='Correct', color='#2ecc71', edgecolor='black')
if (~correct_mask).sum() > 0: ax.hist(max_probs[~correct_mask], bins=20, alpha=0.7, label='Incorrect', color='#e74c3c', edgecolor='black')
ax.set_title('Confidence Distribution', fontweight='bold'); ax.set_xlabel('Max Probability'); ax.legend(); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '6_confidence_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

def calibration_curve(y_true, y_prob, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1); accs, confs = [], []
    for i in range(n_bins):
        mask = (y_prob > bins[i]) & (y_prob <= bins[i+1]) if i < n_bins - 1 else (y_prob >= bins[i]) & (y_prob <= bins[i+1])
        if mask.sum() > 0: accs.append(y_true[mask].mean()); confs.append(y_prob[mask].mean())
    return np.array(confs), np.array(accs)

conf, acc = calibration_curve(all_l, all_pr[:, 1])
fig, ax = plt.subplots(figsize=(8, 8))
if len(conf) > 0: ax.plot(conf, acc, marker='o', color='blue', lw=2, label='Fractured')
ax.plot([0, 1], [0, 1], 'k--', lw=1, label='Perfectly Calibrated'); ax.set_title('Reliability Diagram', fontweight='bold')
ax.set_xlabel('Mean Predicted Probability'); ax.set_ylabel('Fraction of Positives'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '7_calibration_curve.png', dpi=300, bbox_inches='tight'); plt.show()

correct_idxs, incorrect_idxs = np.where(correct_mask)[0], np.where(~correct_mask)[0]
np.random.seed(42)
samp_corr = np.random.choice(correct_idxs, min(6, len(correct_idxs)), replace=False)
samp_incorr = np.random.choice(incorrect_idxs, min(6, len(incorrect_idxs)), replace=False) if len(incorrect_idxs) > 0 else []
n_cols = max(6, len(samp_corr))
fig, axes = plt.subplots(2, n_cols, figsize=(18, 6))
if len(samp_incorr) == 0:
    for c in range(n_cols): axes[1, c].axis('off')
for i, idx in enumerate(samp_corr):
    img = cv2.imread(test_s[idx][0], cv2.IMREAD_GRAYSCALE)
    axes[0, i].imshow(img, cmap='gray'); axes[0, i].set_title(f"T:{class_names[all_l[idx]]}\nP:{class_names[all_p[idx]]}", fontsize=9, color='green'); axes[0, i].axis('off')
for i, idx in enumerate(samp_incorr):
    img = cv2.imread(test_s[idx][0], cv2.IMREAD_GRAYSCALE)
    axes[1, i].imshow(img, cmap='gray'); axes[1, i].set_title(f"T:{class_names[all_l[idx]]}\nP:{class_names[all_p[idx]]}", fontsize=9, color='red'); axes[1, i].axis('off')
axes[0, 0].set_ylabel('CORRECT', fontsize=12, fontweight='bold', color='green', rotation=0, labelpad=60, va='center')
axes[1, 0].set_ylabel('INCORRECT', fontsize=12, fontweight='bold', color='red', rotation=0, labelpad=60, va='center')
plt.tight_layout(); plt.savefig(PLTDIR / '8_sample_predictions.png', dpi=300, bbox_inches='tight'); plt.show()

plot_data = [{'True Class': class_names[i], 'Assigned Probability': prob} for i in range(2) for prob in all_pr[all_l == i, i]]
df_probs = pd.DataFrame(plot_data)
fig, ax = plt.subplots(figsize=(10, 6))
sns.violinplot(data=df_probs, x='True Class', y='Assigned Probability', palette=COLORS, inner='quartile', ax=ax)
ax.set_title('Probability Density for True Class', fontweight='bold'); ax.set_ylim(0,1.05); ax.grid(axis='y', alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '9_probability_violin.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 10: t-SNE ---
print("⏳ Extracting t-SNE features...")
feature_tensor = None
def hook_fn(module, inp, out): global feature_tensor; feature_tensor = out
handle = model.norm.register_forward_hook(hook_fn)
tsne_indices = np.random.choice(len(test_s), min(500, len(test_s)), replace=False)
features_list, labels_tsne = [], []
for idx in tsne_indices:
    img = cv2.imread(test_s[idx][0], cv2.IMREAD_GRAYSCALE)
    with torch.no_grad(): _ = model(get_tensor(img))
    features_list.append(feature_tensor[:, 0, :].cpu().numpy()[0]); labels_tsne.append(all_l[idx])
handle.remove()
tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000).fit_transform(np.array(features_list))
fig, ax = plt.subplots(figsize=(10, 8))
for i, cls in enumerate(class_names):
    mask = np.array(labels_tsne) == i
    ax.scatter(tsne[mask, 0], tsne[mask, 1], c=COLORS[i], label=cls, alpha=0.6, s=20, edgecolors='white', linewidth=0.5)
ax.set_title('t-SNE Visualization', fontweight='bold'); ax.legend(markerscale=2); ax.grid(True, alpha=0.2); ax.set_xticks([]); ax.set_yticks([])
plt.tight_layout(); plt.savefig(PLTDIR / '10_tsne_features.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 11: Error Analysis ---
errors = [{'True': class_names[all_l[i]], 'Predicted': class_names[all_p[i]]} for i in range(len(all_l)) if all_l[i] != all_p[i]]
if errors:
    error_df = pd.DataFrame(errors); error_counts = error_df.groupby(['True', 'Predicted']).size().reset_index(name='Count').sort_values(by='Count', ascending=False)
    fig, ax = plt.subplots(figsize=(8, 5)); error_counts['Pair'] = error_counts['True'] + ' → ' + error_counts['Predicted']
    sns.barplot(data=error_counts, x='Count', y='Pair', palette='Reds_r', ax=ax); ax.set_title('Misclassification Pairs', fontweight='bold')
    for i, v in enumerate(error_counts['Count']): ax.text(v + 0.5, i, str(v), color='black', va='center', fontweight='bold')
    plt.tight_layout(); plt.savefig(PLTDIR / '11_error_analysis.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 12: Class Distribution ---
df_all = pd.DataFrame(test_s, columns=['path', 'label'])
counts = df_all['label'].value_counts().reindex(class_names).fillna(0).astype(int)
fig, axes = plt.subplots(1, 2, figsize=(12, 5)); fig.suptitle('Test Set Class Distribution', fontsize=16, fontweight='bold')
sns.barplot(x=counts.index, y=counts.values, ax=axes[0], palette=COLORS, edgecolor='black'); axes[0].set_title('Count')
for i, v in enumerate(counts.values): axes[0].text(i, v + 1, str(v), ha='center', fontweight='bold')
if counts.sum() > 0: axes[1].pie(counts.values, labels=counts.index, autopct='%1.1f%%', colors=COLORS, startangle=90, wedgeprops={'edgecolor': 'black', 'linewidth': 1.5})
plt.tight_layout(); plt.savefig(PLTDIR / '12_class_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 13: Split Distribution (Uses exact train/val/test logic) ---
split_data = []
for split_name, split_s in [('Train', train_s), ('Val', val_s), ('Test', test_s)]:
    c = Counter(l for _, l in split_s)
    for cls in class_names: split_data.append({'split': split_name, 'label': cls, 'count': c.get(cls, 0)})
df_splits = pd.DataFrame(split_data)
split_pivot = df_splits.pivot(index='split', columns='label', values='count').reindex(['Train', 'Val', 'Test'])[class_names].fillna(0).astype(int)
fig, ax = plt.subplots(figsize=(10, 6))
split_pivot.plot(kind='barh', stacked=True, ax=ax, color=COLORS, edgecolor='black'); ax.set_title('Class Distribution Across Splits', fontweight='bold'); ax.set_xlabel('Images')
for i, split in enumerate(split_pivot.index): ax.text(split_pivot.loc[split].sum() + 5, i, f'Total: {split_pivot.loc[split].sum()}', va='center', fontweight='bold')
plt.tight_layout(); plt.savefig(PLTDIR / '13_split_distribution.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 14: Average Images ---
print("⏳ Computing average images...")
fig, axes = plt.subplots(1, 2, figsize=(10, 5)); fig.suptitle('Average X-Ray per Class', fontweight='bold')
for i, cls in enumerate(class_names):
    cls_paths = df_all[df_all['label'] == cls]['path'].tolist()[:200]
    if not cls_paths: continue
    avg_img = np.zeros((224, 224), dtype=np.float32); valid = 0
    for p in cls_paths:
        img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if img is not None: avg_img += cv2.resize(img, (224, 224)); valid += 1
    if valid > 0: axes[i].imshow((avg_img / valid).astype(np.uint8), cmap='gray')
    axes[i].set_title(cls); axes[i].axis('off')
plt.tight_layout(); plt.savefig(PLTDIR / '14_average_images.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 15: Pixel Intensity ---
fig, ax = plt.subplots(figsize=(10, 6))
for i, cls in enumerate(class_names):
    cls_paths = df_all[df_all['label'] == cls]['path'].sample(n=min(50, len(df_all[df_all['label']==cls])), random_state=42).tolist()
    if not cls_paths: continue
    pixels = []
    for p in cls_paths:
        img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
        if img is not None: pixels.extend(np.random.choice(img.flatten(), size=1000, replace=False))
    if pixels: sns.kdeplot(pixels, fill=True, color=COLORS[i], label=cls, ax=ax, alpha=0.3, linewidth=2)
ax.set_title('Pixel Intensity Distribution', fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '15_pixel_intensity.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 16: Image Dimensions ---
dim_data = []
for p in df_all['path'].sample(n=min(500, len(df_all)), random_state=42):
    img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
    if img is not None:
        h, w = img.shape[:2]; lbl = df_all.loc[df_all['path']==p, 'label'].values[0]
        dim_data.append({'Height': h, 'Width': w, 'Label': lbl})
if dim_data:
    dim_df = pd.DataFrame(dim_data); fig, axes = plt.subplots(1, 2, figsize=(14, 6)); fig.suptitle('Image Dimensions', fontweight='bold')
    sns.boxplot(data=dim_df, x='Label', y='Height', palette=COLORS, ax=axes[0]); axes[0].set_title('Height')
    sns.boxplot(data=dim_df, x='Label', y='Width', palette=COLORS, ax=axes[1]); axes[1].set_title('Width')
    plt.tight_layout(); plt.savefig(PLTDIR / '16_image_dimensions.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 17: XAI GRAD-CAM ---
print("\n🔥 Generating XAI Grad-CAM + Attention Overlays...")
class AdvancedXAI:
    def __init__(self, model):
        self.model = model; self.model.eval(); self.grads, self.activations = None, None
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        if last_conv:
            last_conv.register_forward_hook(lambda m, i, o: setattr(self, 'activations', o.detach()))
            last_conv.register_backward_hook(lambda m, gi, go: setattr(self, 'grads', go[0].detach()))
    def generate(self, img_path, target_class):
        img_gray = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        rgb = np.stack([img_gray, img_gray, img_gray], axis=-1)
        orig_h, orig_w = rgb.shape[:2]
        x = get_tensor(img_gray).float().requires_grad_(True)
        out = self.model(x); self.model.zero_grad(); oh = torch.zeros_like(out); oh[0, target_class] = 1
        out.backward(gradient=oh, retain_graph=True)
        if self.grads is not None and self.activations is not None:
            weights = self.grads.mean(dim=(2, 3), keepdim=True)
            cam = F.relu((weights * self.activations).sum(1)).squeeze().cpu().numpy()
            cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8) if cam.max() > cam.min() else np.zeros_like(cam)
        else: cam = np.zeros((self.model.spatial_h, self.model.spatial_w))
        with torch.no_grad(): _ = self.model(x)
        g_attn, _ = self.model.get_attentions()
        attn = np.zeros((self.model.spatial_h, self.model.spatial_w))
        if g_attn:
            a = g_attn[-1]; attn = a[0, :, 0, 1:].mean(0).cpu().numpy() if a.dim() == 4 else a[0, 0, 1:].cpu().numpy()
            attn = attn.reshape(self.model.spatial_h, self.model.spatial_w)
            attn = (attn - attn.min()) / (attn.max() - attn.min() + 1e-8) if attn.max() > attn.min() else np.zeros_like(attn)
        def overlay(hmap):
            if hmap.max() == 0: return rgb.copy()
            hr = cv2.resize(hmap, (orig_w, orig_h)); colored = cv2.applyColorMap((hr * 255).astype(np.uint8), cv2.COLORMAP_JET)
            return cv2.addWeighted(rgb, 0.6, colored, 0.4, 0)
        comb = 0.5 * cam + 0.5 * attn
        comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8) if comb.max() > comb.min() else comb
        return img_gray, overlay(cam), overlay(attn), overlay(comb)

xai = AdvancedXAI(model)
sample_indices = []
for cls_idx in range(2):
    mask = (all_p == cls_idx) & (all_l == cls_idx)
    if mask.any(): sample_indices.append(np.where(mask)[0][0])
sample_indices = sample_indices[:5]
fig, axes = plt.subplots(len(sample_indices), 4, figsize=(16, 4 * len(sample_indices))); fig.suptitle('Model Explainability', fontsize=18, fontweight='bold', y=1.01)
if len(sample_indices) == 1: axes = axes.reshape(1, -1)
for row_idx, idx in enumerate(sample_indices):
    gray, cam_rgb, attn_rgb, comb_rgb = xai.generate(test_s[idx][0], all_l[idx])
    for col_idx, (img, cm_) in enumerate(zip([gray, cam_rgb, attn_rgb, comb_rgb], ['gray', None, None, None])):
        axes[row_idx, col_idx].imshow(img, cmap=cm_)
        if col_idx == 0: axes[row_idx, col_idx].set_title(f"True: {class_names[all_l[idx]]}\nPred: {class_names[all_p[idx]]} ({max_probs[idx]:.1%})", fontweight='bold')
        else: axes[row_idx, col_idx].set_title(['Original', 'Grad-CAM', 'Attention', 'Fused'][col_idx], fontweight='bold')
        axes[row_idx, col_idx].axis('off')
plt.tight_layout(); plt.savefig(PLTDIR / '17_xai_5_samples.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 18 & 19: Thresholds ---
thresholds = np.arange(0.5, 1.0, 0.05); metrics_data = []
for thresh in thresholds:
    mask = max_probs >= thresh
    if mask.sum() == 0: continue
    p_thresh, l_thresh = all_p[mask], all_l[mask]
    metrics_data.append({'Threshold': thresh, 'Accuracy': accuracy_score(l_thresh, p_thresh), 'F1-Score': f1_score(l_thresh, p_thresh, average='weighted', zero_division=0), 'Coverage': mask.sum() / len(all_p) * 100})
df_thresh = pd.DataFrame(metrics_data)
fig, ax1 = plt.subplots(figsize=(10, 6))
ax1.plot(df_thresh['Threshold'], df_thresh['Accuracy'], color='#3498db', marker='o', lw=2, label='Accuracy')
ax1.plot(df_thresh['Threshold'], df_thresh['F1-Score'], color='#2ecc71', marker='s', lw=2, label='F1-Score')
ax1.set_ylabel('Metric Score'); ax1.grid(True, alpha=0.3)
ax2 = ax1.twinx(); ax2.plot(df_thresh['Threshold'], df_thresh['Coverage'], color='#e74c3c', marker='^', lw=2, linestyle='--', label='Coverage'); ax2.set_ylabel('Coverage (%)')
fig.legend(loc='lower left', bbox_to_anchor=(0.1, -0.1), ncol=3); plt.tight_layout(); plt.savefig(PLTDIR / '18_deployment_thresholds.png', dpi=300, bbox_inches='tight'); plt.show()

fig, ax = plt.subplots(figsize=(10, 6))
cls_threshs = np.arange(0.4, 1.0, 0.05)
for cls_idx, cls in enumerate(class_names):
    cls_f1s = []
    for t in cls_threshs:
        mask = (all_p == cls_idx) & (max_probs >= t)
        cls_f1s.append(f1_score(all_l[mask] == cls_idx, all_p[mask] == cls_idx, zero_division=0) if mask.sum() > 0 else 0)
    opt_idx = np.argmax(cls_f1s)
    ax.plot(cls_threshs, cls_f1s, color=COLORS[cls_idx], marker='o', lw=2, label=f'{cls} (Opt @ {cls_threshs[opt_idx]:.2f})')
ax.set_title('Class-Specific F1 vs. Threshold', fontweight='bold'); ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(PLTDIR / '19_class_specific_thresholds.png', dpi=300, bbox_inches='tight'); plt.show()

# --- PLOT 20: Feature Maps ---
print("\n🧠 Extracting Hierarchical Feature Maps...")
mask_idx = np.where((all_l == 1) & (all_p == 1))[0]
if len(mask_idx) == 0: mask_idx = np.where((all_l == 0) & (all_p == 0))[0]
if len(mask_idx) == 0: mask_idx = [0]
img = cv2.imread(test_s[mask_idx[0]][0], cv2.IMREAD_GRAYSCALE)
x_feat = get_tensor(img); feature_maps = {}
def get_hook(name):
    def hook(module, input, output): feature_maps[name] = output.detach().cpu()
    return hook
model.backbone.feature_extractor[1].register_forward_hook(get_hook('Early_Layer'))
model.backbone.feature_extractor[4].register_forward_hook(get_hook('Mid_Layer'))
model.backbone.feature_extractor[7].register_forward_hook(get_hook('Late_Layer'))
with torch.no_grad(): _ = model(x_feat)
fig, axes = plt.subplots(3, 8, figsize=(20, 7)); fig.suptitle('Hierarchical Feature Extraction', fontsize=16, fontweight='bold')
for row_idx, (layer, title) in enumerate(zip(['Early_Layer', 'Mid_Layer', 'Late_Layer'], ['Early', 'Mid', 'Late'])):
    if layer in feature_maps:
        fm = feature_maps[layer].squeeze(); n_channels = fm.shape[0]; channel_idxs = np.linspace(0, n_channels - 1, 8, dtype=int)
        for col_idx, c_idx in enumerate(channel_idxs):
            if c_idx < n_channels:
                ch_img = fm[c_idx].numpy(); ch_img = (ch_img - ch_img.min()) / (ch_img.max() - ch_img.min() + 1e-8)
                axes[row_idx, col_idx].imshow(ch_img, cmap='magma'); axes[row_idx, col_idx].set_title(f'Ch {c_idx}', fontsize=9); axes[row_idx, col_idx].axis('off')
    axes[row_idx, -1].axis('off')
plt.tight_layout(); plt.savefig(PLTDIR / '20_feature_maps_hierarchy.png', dpi=300, bbox_inches='tight'); plt.show()

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  FINAL EVALUATION METRICS                                                       ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
test_acc = accuracy_score(all_l, all_p); test_f1 = f1_score(all_l, all_p, average='weighted')
test_auc = roc_auc_score(all_l, all_pr[:, 1]); test_ap = average_precision_score(all_l, all_pr[:, 1])
mcc = matthews_corrcoef(all_l, all_p); kappa = cohen_kappa_score(all_l, all_p)
ll = log_loss(all_l, all_pr); brier = brier_score_loss(all_l, all_pr[:, 1])
tp, fp, fn, tn = cm[1, 1], cm[0, 1], cm[1, 0], cm[0, 0]
sensitivity = tp / (tp + fn) if (tp + fn) > 0 else 0.0; specificity = tn / (tn + fp) if (tn + fp) > 0 else 0.0
precision = tp / (tp + fp) if (tp + fp) > 0 else 0.0; npv = tn / (tn + fn) if (tn + fn) > 0 else 0.0
fpr = fp / (fp + tn) if (fp + tn) > 0 else 0.0; fnr = fn / (fn + tp) if (fn + tp) > 0 else 0.0

print("\n============================================================")
print("📊 COMPREHENSIVE EVALUATION METRICS")
print("============================================================")
print(f"\n🎯 PRIMARY METRICS:\n   Accuracy:  {test_acc:.4f}\n   F1-Score:  {test_f1:.4f}\n   AUC-ROC:   {test_auc:.4f}\n   AP Score:  {test_ap:.4f}")
print(f"\n📈 ADDITIONAL METRICS:\n   MCC:       {mcc:.4f}\n   Kappa:     {kappa:.4f}\n   Log Loss:  {ll:.4f}\n   Brier:     {brier:.4f}")
print(f"\n🔬 CLINICAL METRICS:\n   Sensitivity: {sensitivity:.4f}\n   Specificity: {specificity:.4f}\n   Precision:   {precision:.4f}\n   NPV:         {npv:.4f}\n   FPR:         {fpr:.4f}\n   FNR:         {fnr:.4f}")
print(f"\n📋 CLASSIFICATION REPORT:\n{classification_report(all_l, all_p, target_names=class_names)}")
print("="*60); print(f"✅ ALL 20 PLOTS GENERATED SUCCESSFULLY!\n📁 Location: {PLTDIR}"); print("="*60)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  PART 6: INTERACTIVE WIDGETS & STANDALONE HEATMAPS                             ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

SEVERITY_MAP = {'Not Fractured': '✅ NORMAL - No Fracture Detected', 'Fractured': '🚨 FRACTURE - Immediate Clinical Attention Required'}

def predict_single_image(image_array):
    if len(image_array.shape) == 2: image_array = np.stack([image_array]*3, axis=-1)
    img_resized = cv2.resize(image_array, (224, 224))
    t = torch.from_numpy(img_resized).float().permute(2, 0, 1) / 255.0
    t = ((t - torch.tensor(MEAN).view(3,1,1)) / torch.tensor(STD).view(3,1,1)).unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        with autocast(): logits = model(t); probs = F.softmax(logits, dim=1)
        conf, pred_idx = probs.max(dim=1)
    return pred_idx.item(), conf.item(), probs.cpu().numpy()[0]

upload_widget = widgets.FileUpload(accept='.jpg,.jpeg,.png', multiple=False, description='📤 Upload X-Ray', button_style='info')
path_input = widgets.Text(value='', placeholder='/kaggle/input/.../xray.jpg', description='📁 Or path:', layout={'width': '600px'})
predict_button = widgets.Button(description='🔍 Predict', button_style='success', layout={'width': '200px'})
output_area = widgets.Output()

def on_predict_clicked(b):
    with output_area:
        clear_output(wait=True); image_rgb = None
        if upload_widget.value:
            content = list(upload_widget.value.values())[0]['content']
            image_rgb = np.array(Image.open(io.BytesIO(content)).convert('RGB'))
        elif path_input.value.strip():
            p = Path(path_input.value.strip())
            if p.exists(): image_rgb = cv2.cvtColor(cv2.imread(str(p)), cv2.COLOR_BGR2RGB)
            else: print("❌ File not found"); return
        else: print("⚠️ Provide an image"); return
            
        pred_idx, confidence, probs = predict_single_image(image_rgb)
        print(f"\n{'='*50}\n   PREDICTION: {class_names[pred_idx].upper()}\n   {SEVERITY_MAP[class_names[pred_idx]]}\n   CONFIDENCE: {confidence:.2%}\n{'='*50}")
        
        fig, axes = plt.subplots(1, 2, figsize=(12, 5))
        axes[0].imshow(image_rgb, cmap='gray')
        color = 'red' if pred_idx == 1 else 'green'
        axes[0].set_title(f"{SEVERITY_MAP[class_names[pred_idx]]}\n{confidence:.2%}", fontsize=13, fontweight='bold', color=color); axes[0].axis('off')
        
        colors_bar = ['#e74c3c' if i == pred_idx else '#95a5a6' for i in range(2)]
        bars = axes[1].barh(class_names, probs * 100, color=colors_bar, edgecolor='black')
        axes[1].set_xlim(0, 105); axes[1].set_title('Probabilities', fontweight='bold')
        for bar, prob in zip(bars, probs): axes[1].text(bar.get_width() + 1, bar.get_y() + bar.get_height()/2, f'{prob:.1f}%', va='center', fontweight='bold')
        axes[1].grid(axis='x', alpha=0.3); plt.tight_layout(); plt.show()

predict_button.on_click(on_predict_clicked)
print("\n╔══════════════════════════════════════════════════════════════╗")
print("║           🦴 FRACATLAS X-RAY CLASSIFICATION                 ║")
print("╚══════════════════════════════════════════════════════════════╝\n")
display(widgets.VBox([upload_widget, widgets.Label("── OR ──"), path_input, predict_button, output_area]))

# --- Heatmap Generator ---
class XRayHeatmapGenerator:
    def __init__(self, model):
        self.model = model; self.model.eval(); self.cnn_f, self.cnn_g = None, None
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d): last_conv = m
        if last_conv:
            last_conv.register_forward_hook(lambda m, i, o: setattr(self, 'cnn_f', o.detach()))
            last_conv.register_backward_hook(lambda m, gi, go: setattr(self, 'cnn_g', go[0].detach()))
    def get_cnn(self, x, target):
        out = self.model(x); self.model.zero_grad(); oh = torch.zeros_like(out); oh[0, target] = 1
        out.backward(gradient=oh, retain_graph=True)
        if self.cnn_f is None: return None
        weights = self.cnn_g.mean(dim=(2, 3), keepdim=True)
        cam = F.relu((weights * self.cnn_f).sum(1)).squeeze().cpu().numpy()
        return (cam - cam.min()) / (cam.max() - cam.min() + 1e-8) if cam.max() > cam.min() else cam
    def get_attn(self, x):
        with torch.no_grad(): self.model(x)
        g, _ = self.model.get_attentions()
        if not g: return None
        a = g[-1]; a = a[0, :, 0, 1:].mean(0).cpu().numpy() if a.dim() == 4 else a[0, 0, 1:].cpu().numpy()
        a = a.reshape(self.model.spatial_h, self.model.spatial_w)
        return (a - a.min()) / (a.max() - a.min() + 1e-8) if a.max() > a.min() else a
    def draw(self, img_path, target, save_path):
        img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        if img is None: return
        rgb = np.stack([img, img, img], axis=-1); sz = rgb.shape[:2]; t = get_tensor(rgb)
        c, a = self.get_cnn(t.clone(), target), self.get_attn(t.clone())
        comb = 0.5 * c + 0.5 * a if c is not None and a is not None else c
        if comb is not None and comb.max() > comb.min(): comb = (comb - comb.min()) / (comb.max() - comb.min() + 1e-8)
        fig, axes = plt.subplots(1, 4, figsize=(16, 4))
        axes[0].imshow(img, cmap='gray'); axes[0].set_title(f"True: {class_names[target]}", fontweight='bold'); axes[0].axis('off')
        for i, (hmap, title) in enumerate(zip([c, a, comb], ['Grad-CAM', 'Attention', 'Combined'])):
            if hmap is not None:
                hr = zoom(hmap, (sz[0]/hmap.shape[0], sz[1]/hmap.shape[1]), order=1)
                colored = (plt.cm.jet(hr)[:,:,:3] * 255).astype(np.uint8)
                axes[i+1].imshow(cv2.addWeighted(rgb, 0.6, colored, 0.4, 0))
            else: axes[i+1].imshow(img, cmap='gray')
            axes[i+1].set_title(title, fontweight='bold'); axes[i+1].axis('off')
        plt.tight_layout(); plt.savefig(save_path, dpi=150, bbox_inches='tight'); plt.show()

hm_gen = XRayHeatmapGenerator(model)
print("\n🔥 Generating Heatmaps...")
for cls_idx in range(2):
    cls_tests = [s for s in test_s if class_names.index(s[1]) == cls_idx]
    if cls_tests:
        print(f"\n📊 {class_names[cls_idx]}:")
        for p, _ in cls_tests[:2]: hm_gen.draw(p, cls_idx, PLTDIR / f'hm_{class_names[cls_idx]}_{Path(p).stem}.png')
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 1 — Setup, Imports, and Global Configuration                             ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝
!pip install -q timm einops albumentations opencv-python-headless

import os, time, random, copy, warnings
from pathlib import Path
from collections import Counter, defaultdict
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageFile
import cv2
from tqdm.auto import tqdm
from scipy.ndimage import zoom
import json
import re

import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, WeightedRandomSampler, ConcatDataset
from torch.cuda.amp import GradScaler, autocast
import torchvision.models as models

import albumentations as A
from albumentations.pytorch import ToTensorV2
from einops import rearrange

from sklearn.metrics import (accuracy_score, f1_score, roc_auc_score, roc_curve, 
                             confusion_matrix, classification_report, precision_recall_curve,
                             average_precision_score, matthews_corrcoef, cohen_kappa_score,
                             log_loss, brier_score_loss)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE

ImageFile.LOAD_TRUNCATED_IMAGES = True
warnings.filterwarnings('ignore')

SEED = 42
random.seed(SEED); np.random.seed(SEED); torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
OUTDIR = Path('/kaggle/working')
CKDIR = OUTDIR / 'checkpoints'
PLTDIR = OUTDIR / 'plots'
for d in [CKDIR, PLTDIR]: d.mkdir(exist_ok=True)

MEAN = (0.485, 0.456, 0.406)
STD = (0.229, 0.224, 0.225)
VALID_EXTS = {'.jpg', '.jpeg', '.png', '.bmp', '.tiff', '.npy'}

# Hyperparameters
HPARAMS = {
    'img_size': 224,
    'batch_size': 32,
    'num_epochs': 50,
    'lr_backbone': 1e-5,
    'lr_head': 1e-4,
    'weight_decay': 1e-4,
    'freeze_epochs': 10,
    'patience': 15,
    'd_model': 384,
    'nhead': 6,
    'n_layers': 4,
    'dim_ffn': 1536,
    'label_smoothing': 0.05,
    'mixup_alpha': 0.2,
    'cutmix_alpha': 1.0,
    'mixup_prob': 0.3,
}

print(f"✅ Setup complete on {DEVICE}")

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 2 — Dataset Loaders                                                      ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def load_fracatlas():
    """Load FracAtlas dataset - properly handles both classes"""
    samples = []
    class_names = ['Fractured', 'Not Fractured']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Fracture Dataset/Fracture Dataset/FracAtlas/images')
    
    if not base.exists():
        print(f"❌ FracAtlas dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading FracAtlas dataset...\n✅ Found dataset at: {base}")
    
    class_counts = {'Fractured': 0, 'Not Fractured': 0}
    
    for cls_dir in sorted(base.iterdir()):
        if not cls_dir.is_dir(): 
            continue
        
        folder_name = cls_dir.name
        name_lower = folder_name.lower().replace('-', ' ').replace('_', ' ')
        
        cname = None
        
        # Check for class indicators - ORDER MATTERS!
        # Check 'non'/'not' first to catch 'Non_fractured'
        if 'non' in name_lower or 'not' in name_lower:
            cname = 'Not Fractured'
        elif 'fractured' in name_lower or 'fract' in name_lower:
            cname = 'Fractured'
        elif 'normal' in name_lower or 'healthy' in name_lower:
            cname = 'Not Fractured'
        elif 'positive' in name_lower or 'abnormal' in name_lower:
            cname = 'Fractured'
        else:
            # Check subfolders
            for sub in cls_dir.iterdir():
                if sub.is_dir():
                    sub_lower = sub.name.lower()
                    if 'non' in sub_lower or 'not' in sub_lower:
                        cname = 'Not Fractured'
                        break
                    elif 'fractured' in sub_lower or 'fract' in sub_lower:
                        cname = 'Fractured'
                        break
        
        if cname is None:
            print(f"⚠️ Skipping unknown folder: {folder_name}")
            continue
        
        count = 0
        for ext in VALID_EXTS:
            for p in cls_dir.rglob(f'*{ext}'):
                samples.append((str(p), cname))
                count += 1
        
        class_counts[cname] += count
        print(f"   {cname}: {count} images from '{folder_name}'")
    
    print(f"\n📊 FracAtlas Summary:")
    print(f"   Fractured: {class_counts['Fractured']} images")
    print(f"   Not Fractured: {class_counts['Not Fractured']} images")
    print(f"   Total: {len(samples)} images")
    
    return samples, class_names

def load_tb_xray():
    """Load Tuberculosis Chest X-Ray dataset"""
    samples = []
    class_names = ['Normal', 'Tuberculosis']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Tuberculosis (TB) Chest X-ray Database/Tuberculosis (TB) Chest X-ray Database/TB_Chest_Radiography_Database')
    
    if not base.exists():
        print(f"❌ TB dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading Tuberculosis Chest X-Ray...\n✅ Found dataset at: {base}")
    
    class_counts = {'Normal': 0, 'Tuberculosis': 0}
    
    for folder in base.iterdir():
        if not folder.is_dir(): 
            continue
        
        fn = folder.name.lower()
        
        # Check for class indicators
        if 'normal' in fn or 'healthy' in fn:
            cname = 'Normal'
        elif 'tb' in fn or 'tuberculosis' in fn:
            cname = 'Tuberculosis'
        else:
            continue
        
        count = 0
        for ext in VALID_EXTS:
            for p in folder.rglob(f'*{ext}'):
                samples.append((str(p), cname))
                count += 1
        
        if count > 0:
            class_counts[cname] += count
            print(f"   {cname}: {count} images")
    
    print(f"\n📊 TB Chest X-Ray Summary:")
    print(f"   Normal: {class_counts['Normal']} images")
    print(f"   Tuberculosis: {class_counts['Tuberculosis']} images")
    print(f"   Total: {len(samples)} images")
    
    return samples, class_names

def load_pneumonia_xray():
    """Load Chest X-Ray Pneumonia dataset"""
    samples = []
    class_names = ['Normal', 'Pneumonia']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Chest X-Ray Images (Pneumonia)/Chest X-Ray Images (Pneumonia)/chest_xray')
    
    if not base.exists():
        print(f"❌ Pneumonia dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading Chest X-Ray Pneumonia...\n✅ Found dataset at: {base}")
    
    class_counts = {'Normal': 0, 'Pneumonia': 0}
    
    for split in ['train', 'test', 'val']:
        split_path = base / split
        if not split_path.exists(): 
            continue
        
        for folder in split_path.iterdir():
            if not folder.is_dir(): 
                continue
            
            fn = folder.name.lower()
            
            # Check for class indicators
            if 'normal' in fn:
                cname = 'Normal'
            elif 'pneumonia' in fn:
                cname = 'Pneumonia'
            else:
                continue
            
            count = 0
            for ext in VALID_EXTS:
                for p in folder.rglob(f'*{ext}'):
                    samples.append((str(p), cname))
                    count += 1
            
            if count > 0:
                class_counts[cname] += count
                print(f"   {split}/{cname}: {count} images")
    
    print(f"\n📊 Pneumonia Chest X-Ray Summary:")
    print(f"   Normal: {class_counts['Normal']} images")
    print(f"   Pneumonia: {class_counts['Pneumonia']} images")
    print(f"   Total: {len(samples)} images")
    
    return samples, class_names

def load_retina_dataset():
    """Load Diabetic Retinopathy dataset"""
    print("📂 Loading Diabetic Retinopathy dataset...")
    samples = []
    class_names = ['No_DR', 'Mild', 'Moderate', 'Severe', 'Proliferate_DR']
    
    # Try multiple possible paths
    possible_paths = [
        Path('/kaggle/input/datasets/aayush4204/daataaset/Diabetic Retinopathy 224x224 (2019 Data)/Diabetic Retinopathy 224x224 (2019 Data)/colored_images'),
        Path('/kaggle/input/datasets/aayush4204/daataaset/Diabetic Retinopathy 224x224 (2019 Data)/colored_images'),
        Path('/kaggle/input/datasets/sovitrath/diabetic-retinopathy-224x224-2019-data/colored_images'),
        Path('/kaggle/input/diabetic-retinopathy-224x224-2019-data/colored_images'),
    ]
    
    base = None
    for path in possible_paths:
        if path.exists():
            base = path
            print(f"✅ Found dataset at: {base}")
            break
    
    if base is None:
        print("❌ Dataset not found!")
        return [], class_names
    
    # Map folder names to class names
    folder_to_class = {
        '0': 'No_DR',
        '1': 'Mild',
        '2': 'Moderate',
        '3': 'Severe',
        '4': 'Proliferate_DR',
        '0 - No_DR': 'No_DR',
        '1 - Mild': 'Mild',
        '2 - Moderate': 'Moderate',
        '3 - Severe': 'Severe',
        '4 - Proliferate_DR': 'Proliferate_DR',
        'No_DR': 'No_DR',
        'Mild': 'Mild',
        'Moderate': 'Moderate',
        'Severe': 'Severe',
        'Proliferate_DR': 'Proliferate_DR'
    }
    
    class_counts = {cls: 0 for cls in class_names}
    
    # Collect images
    for folder in base.iterdir():
        if not folder.is_dir():
            continue
        folder_name = folder.name.strip()
        class_name = folder_to_class.get(folder_name, None)
        
        if class_name is None:
            # Try to find class from folder name containing number
            for key in folder_to_class:
                if key in folder_name:
                    class_name = folder_to_class[key]
                    break
        
        if class_name is None:
            print(f"⚠️ Skipping unknown folder: {folder_name}")
            continue
        
        count = 0
        for ext in VALID_EXTS:
            for p in folder.rglob(f'*{ext}'):
                samples.append((str(p), class_name))
                count += 1
        if count > 0:
            class_counts[class_name] += count
            print(f"   {class_name}: {count} images")
    
    print(f"✅ Total images: {len(samples)}")
    return samples, class_names

def load_brain_tumor():
    """Load Brain Tumor MRI dataset"""
    samples = []
    class_names = ['glioma', 'meningioma', 'notumor', 'pituitary']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Brain Tumor MRI Dataset')
    
    if not base.exists():
        print(f"❌ Brain Tumor dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading Brain Tumor MRI...\n✅ Found dataset at: {base}")
    
    class_counts = {cls: 0 for cls in class_names}
    
    for ext in VALID_EXTS:
        for img_path in base.rglob(f'*{ext}'):
            parent = img_path.parent.name.lower()
            fname = img_path.stem.lower()
            
            if 'glioma' in parent or 'glioma' in fname:
                cls = 'glioma'
            elif 'meningioma' in parent or 'meningioma' in fname:
                cls = 'meningioma'
            elif 'pituitary' in parent or 'pituitary' in fname:
                cls = 'pituitary'
            elif 'notumor' in parent or 'notumor' in fname or 'no tumor' in parent:
                cls = 'notumor'
            else:
                continue
            
            samples.append((str(img_path), cls))
            class_counts[cls] += 1
    
    for cls in class_names:
        print(f"   {cls}: {class_counts[cls]} images")
    
    return samples, class_names

def load_bone_fracture():
    """Load Bone Fracture Binary Classification dataset"""
    samples = []
    class_names = ['Fractured', 'Not Fractured']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/Bone_Fracture_Binary_Classification')
    
    if not base.exists():
        print(f"❌ Bone Fracture dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading Bone Fracture Binary Classification...\n✅ Found dataset at: {base}")
    
    class_counts = {'Fractured': 0, 'Not Fractured': 0}
    
    for ext in VALID_EXTS:
        for img_path in base.rglob(f'*{ext}'):
            parent = img_path.parent.name.lower()
            fname = img_path.stem.lower()
            
            if 'not' in parent or 'normal' in parent or 'negative' in parent:
                cls = 'Not Fractured'
            elif 'fractured' in parent or 'positive' in parent or 'fract' in parent:
                cls = 'Fractured'
            else:
                # Check filename
                if 'not' in fname or 'normal' in fname:
                    cls = 'Not Fractured'
                elif 'fract' in fname:
                    cls = 'Fractured'
                else:
                    continue
            
            samples.append((str(img_path), cls))
            class_counts[cls] += 1
    
    print(f"   Fractured: {class_counts['Fractured']} images")
    print(f"   Not Fractured: {class_counts['Not Fractured']} images")
    
    return samples, class_names

def load_ham10000():
    """Load HAM10000 Skin Cancer dataset using metadata CSV"""
    samples = []
    class_names = ['akiec', 'bcc', 'bkl', 'df', 'mel', 'nv', 'vasc']
    base = Path('/kaggle/input/datasets/aayush4204/daataaset/skin cancer')
    
    if not base.exists():
        print(f"❌ HAM10000 dataset not found at: {base}")
        return [], class_names
    
    print(f"📂 Loading HAM10000 Skin Cancer...\n✅ Found dataset at: {base}")
    
    # Find the metadata CSV
    metadata_file = None
    for csv_file in base.rglob('*.csv'):
        if 'metadata' in csv_file.name.lower():
            metadata_file = csv_file
            break
    
    if metadata_file is None:
        print("❌ No metadata CSV file found!")
        return [], class_names
    
    print(f"📄 Found metadata CSV: {metadata_file.name}")
    
    try:
        df = pd.read_csv(metadata_file)
        print(f"   Columns: {list(df.columns)}")
        
        # Get image paths
        img_paths = {}
        for ext in ['.jpg', '.png', '.jpeg']:
            for p in base.rglob(f'*{ext}'):
                img_paths[p.stem] = str(p)
        
        print(f"   Found {len(img_paths)} image files")
        
        class_counts = {cls: 0 for cls in class_names}
        
        for _, row in df.iterrows():
            img_id = row['image_id']
            label = row['dx'].lower()
            
            if label in class_names:
                cls = label
            else:
                # Map full names to short codes
                label_map = {
                    'actinic keratosis': 'akiec',
                    'basal cell carcinoma': 'bcc',
                    'benign keratosis': 'bkl',
                    'dermatofibroma': 'df',
                    'melanoma': 'mel',
                    'melanocytic nevus': 'nv',
                    'vascular lesion': 'vasc'
                }
                cls = label_map.get(label, label)
            
            if cls in class_names and img_id in img_paths:
                samples.append((img_paths[img_id], cls))
                class_counts[cls] += 1
        
        print(f"\n📊 HAM10000 Class Distribution:")
        total = 0
        for cls in class_names:
            count = class_counts[cls]
            total += count
            print(f"   {cls}: {count} images")
        print(f"   Total: {total} images")
        
        if total > 0:
            return samples, class_names
        
    except Exception as e:
        print(f"❌ Error: {e}")
    
    print("❌ Failed to load HAM10000")
    return [], class_names

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 3 — Helper Functions for Data Processing                                 ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def train_test_split_samples(samples, class_names, test_size=0.3, val_size=0.15):
    """Split samples into train/val/test with stratification"""
    if not samples:
        return [], [], [], class_names
    
    # Get labels
    labels = [class_names.index(l) for _, l in samples]
    
    # Check if we have enough samples per class for stratification
    class_counts = Counter(labels)
    can_stratify = all(count >= 2 for count in class_counts.values()) and len(set(labels)) > 1
    
    if can_stratify:
        try:
            # First split: train + (val + test)
            train_s, temp_s = train_test_split(samples, test_size=test_size + val_size, 
                                              stratify=labels, random_state=SEED)
            
            # Second split: val and test
            temp_labels = [class_names.index(l) for _, l in temp_s]
            val_s, test_s = train_test_split(temp_s, test_size=test_size/(test_size + val_size),
                                            stratify=temp_labels, random_state=SEED)
        except ValueError:
            # If stratification fails, use non-stratified
            print("   ⚠️ Stratification failed, using non-stratified split")
            train_s, temp_s = train_test_split(samples, test_size=test_size + val_size, random_state=SEED)
            val_s, test_s = train_test_split(temp_s, test_size=test_size/(test_size + val_size), random_state=SEED)
    else:
        # Use non-stratified split
        print("   ℹ️ Using non-stratified split (not enough samples per class)")
        train_s, temp_s = train_test_split(samples, test_size=test_size + val_size, random_state=SEED)
        val_s, test_s = train_test_split(temp_s, test_size=test_size/(test_size + val_size), random_state=SEED)
    
    return train_s, val_s, test_s, class_names

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 4 — Data Augmentations (Medical Image Safe)                              ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

def get_medical_transforms(split, img_size=224, is_grayscale=False):
    """Get transforms for medical images - NO vertical flip for X-rays"""
    norm = [A.Normalize(mean=MEAN, std=STD), ToTensorV2()]
    
    if split != 'train':
        return A.Compose([
            A.Resize(height=img_size, width=img_size),
            A.ToGray(p=1.0 if is_grayscale else 0.0),
        ] + norm)
    
    transforms = [
        A.Resize(height=img_size, width=img_size),
        A.HorizontalFlip(p=0.5),
        A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.1, rotate_limit=15, 
                          p=0.5, border_mode=cv2.BORDER_CONSTANT, value=0),
        A.RandomBrightnessContrast(brightness_limit=0.2, contrast_limit=0.2, p=0.5),
        A.GaussNoise(var_limit=(10.0, 50.0), p=0.3),
    ]
    
    if is_grayscale:
        transforms.append(A.ToGray(p=1.0))
    else:
        transforms.append(A.CLAHE(clip_limit=2.0, tile_grid_size=(8, 8), p=0.3))
    
    return A.Compose(transforms + norm)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 5 — Unified Dataset Class                                                ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class UnifiedMedicalDataset(Dataset):
    def __init__(self, samples, class_names, transform=None, is_grayscale=False, dataset_name=''):
        self.samples = samples
        self.class_names = class_names
        self.class_to_idx = {n: i for i, n in enumerate(class_names)}
        self.transform = transform
        self.is_grayscale = is_grayscale
        self.dataset_name = dataset_name
        self.num_classes = len(class_names)
    
    def __len__(self):
        return len(self.samples)
    
    def __getitem__(self, idx):
        p, l = self.samples[idx]
        l = self.class_to_idx[l]
        
        try:
            if p.endswith('.npy'):
                img = np.load(p)
                if len(img.shape) == 2:
                    img = np.stack([img] * 3, axis=-1)
                elif len(img.shape) == 3 and img.shape[-1] == 1:
                    img = np.concatenate([img] * 3, axis=-1)
            else:
                if self.is_grayscale:
                    img = cv2.imread(p, cv2.IMREAD_GRAYSCALE)
                    if img is None:
                        img = np.array(Image.open(p).convert('L'))
                    img = np.stack([img, img, img], axis=-1)
                else:
                    img = cv2.imread(p)
                    if img is None:
                        img = np.array(Image.open(p).convert('RGB'))
                    else:
                        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            
            if self.transform:
                img = self.transform(image=img)['image']
            
            return img.float(), l
            
        except Exception as e:
            dummy = torch.zeros((3, 224, 224))
            return dummy.float(), l

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 6 — Model Architecture (EnhancedHViT)                                   ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class DropPath(nn.Module):
    def __init__(self, drop_prob=0.0): 
        super().__init__(); self.drop_prob = drop_prob
    def forward(self, x):
        if self.drop_prob == 0.0 or not self.training: return x
        return x * (torch.rand((x.shape[0], 1, 1), dtype=torch.float32, device=x.device) * float(1.0 - self.drop_prob))

class WindowAttention(nn.Module):
    def __init__(self, d_model, nhead, window_size=7, dropout=0.1):
        super().__init__()
        self.nhead, self.window_size, self.head_dim = nhead, window_size, d_model // nhead
        self.scale = self.head_dim ** -0.5
        self.qkv = nn.Linear(d_model, d_model * 3)
        self.proj = nn.Linear(d_model, d_model)
        self.attn_drop, self.proj_drop, self.attn_weights = nn.Dropout(dropout), nn.Dropout(dropout), None
        
    def forward(self, x, H, W):
        B, N, C = x.shape
        x = x.reshape(B, H, W, C)
        pad_h, pad_w = (self.window_size - H % self.window_size) % self.window_size, (self.window_size - W % self.window_size) % self.window_size
        if pad_h > 0 or pad_w > 0:
            x = F.pad(x, (0, 0, 0, pad_w, 0, pad_h))
        Hp, Wp = x.shape[1], x.shape[2]
        x = x.reshape(B, Hp//self.window_size, self.window_size, Wp//self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(-1, self.window_size**2, C)
        qkv = self.qkv(x).reshape(-1, self.window_size**2, 3, self.nhead, self.head_dim).permute(2,0,3,1,4)
        attn = (qkv[0] @ qkv[1].transpose(-2,-1)) * self.scale
        attn = attn.softmax(dim=-1)
        self.attn_weights = attn.detach()
        x = self.proj_drop(self.proj((self.attn_drop(attn) @ qkv[2]).transpose(1,2).reshape(-1, self.window_size**2, C)))
        x = x.reshape(B, Hp//self.window_size, Wp//self.window_size, self.window_size, self.window_size, C).permute(0,1,3,2,4,5).reshape(B, Hp, Wp, C)
        return x[:, :H, :W, :].reshape(B, -1, C) if pad_h > 0 or pad_w > 0 else x.reshape(B, -1, C)

class TransformerBlockWithWindow(nn.Module):
    def __init__(self, d_model, nhead, ffn_dim, window_size=7, attn_drop=0.1, ffn_drop=0.1, drop_path=0.0):
        super().__init__()
        self.norm1, self.norm2 = nn.LayerNorm(d_model), nn.LayerNorm(d_model)
        self.global_attn = nn.MultiheadAttention(d_model, nhead, dropout=attn_drop, batch_first=True)
        self.window_attn = WindowAttention(d_model, nhead, window_size, attn_drop)
        self.gate = nn.Sequential(nn.Linear(d_model*2, d_model), nn.Sigmoid())
        self.ffn = nn.Sequential(
            nn.Linear(d_model, ffn_dim), nn.GELU(), nn.Dropout(ffn_drop),
            nn.Linear(ffn_dim, d_model), nn.Dropout(ffn_drop)
        )
        self.drop_path = DropPath(drop_path) if drop_path > 0. else nn.Identity()
        self.attn_weights, self.window_attn_weights = None, None
        
    def forward(self, x, H, W):
        x_norm = self.norm1(x)
        global_out, global_attn = self.global_attn(x_norm, x_norm, x_norm, need_weights=True)
        self.attn_weights = global_attn.detach()
        window_out = self.window_attn(x_norm[:, 1:, :], H, W)
        self.window_attn_weights = self.window_attn.attn_weights
        gate = self.gate(torch.cat([global_out[:, 1:, :], window_out], dim=-1))
        combined = torch.cat([global_out[:, :1, :], gate * global_out[:, 1:, :] + (1-gate) * window_out], dim=1)
        x = x + self.drop_path(combined)
        return x + self.drop_path(self.ffn(self.norm2(x)))

class ImprovedBackbone(nn.Module):
    def __init__(self): 
        super().__init__(); 
        self.backbone = models.convnext_base(weights=models.ConvNeXt_Base_Weights.IMAGENET1K_V1)
        self.feature_extractor = self.backbone.features
    def forward(self, x): return self.feature_extractor(x)

class EnhancedHViT(nn.Module):
    def __init__(self, num_classes, img_size=224, d_model=384, nhead=6, n_layers=4, dim_ffn=1536):
        super().__init__()
        self.d_model = d_model
        self.img_size = img_size
        self.backbone = ImprovedBackbone()
        with torch.no_grad():
            _, c, h, w = self.backbone(torch.zeros(1, 3, img_size, img_size)).shape
        self.spatial_h, self.spatial_w = h, w
        self.proj = nn.Sequential(
            nn.Conv2d(c, d_model, 1, bias=False), nn.BatchNorm2d(d_model), nn.GELU(),
            nn.Conv2d(d_model, d_model, 3, padding=1, bias=False), nn.BatchNorm2d(d_model), nn.GELU()
        )
        self.norm_proj = nn.LayerNorm(d_model)
        self.cls_token = nn.Parameter(torch.zeros(1, 1, d_model))
        self.pos_embed = nn.Parameter(torch.zeros(1, h*w + 1, d_model))
        nn.init.trunc_normal_(self.cls_token, std=0.02)
        nn.init.trunc_normal_(self.pos_embed, std=0.02)
        dpr = [x.item() for x in torch.linspace(0, 0.15, n_layers)]
        self.encoder = nn.ModuleList([
            TransformerBlockWithWindow(d_model, nhead, dim_ffn, min(7, h, w), drop_path=dpr[i])
            for i in range(n_layers)
        ])
        self.norm = nn.LayerNorm(d_model)
        self.head = nn.Sequential(
            nn.LayerNorm(d_model), nn.Linear(d_model, d_model//2), nn.GELU(), nn.Dropout(0.3),
            nn.Linear(d_model//2, d_model//4), nn.GELU(), nn.Dropout(0.15),
            nn.Linear(d_model//4, num_classes)
        )
        
    def forward(self, x):
        B = x.size(0)
        tokens = rearrange(self.proj(self.backbone(x)), 'b d h w -> b (h w) d')
        tokens = self.norm_proj(tokens)
        tokens = torch.cat([self.cls_token.expand(B,-1,-1), tokens], dim=1) + self.pos_embed
        for block in self.encoder:
            tokens = block(tokens, self.spatial_h, self.spatial_w)
        return self.head(self.norm(tokens)[:, 0])
    
    def get_attentions(self):
        g, w = [], []
        for b in self.encoder:
            if b.attn_weights is not None: g.append(b.attn_weights)
            if b.window_attn_weights is not None: w.append(b.window_attn_weights)
        return g, w
    
    def unfreeze_backbone_partial(self, n=3):
        for p in self.backbone.feature_extractor.parameters(): 
            p.requires_grad_(False)
        for child in list(self.backbone.feature_extractor.children())[-n:]:
            for p in child.parameters(): 
                p.requires_grad_(True)

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 7 — Training Functions                                                    ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

class SmoothCELoss(nn.Module):
    def __init__(self, label_smoothing=0.05): 
        super().__init__(); 
        self.criterion = nn.CrossEntropyLoss(label_smoothing=label_smoothing)
    def forward(self, inputs, targets): 
        return self.criterion(inputs, targets)

def mixup_data(x, y, alpha=0.2):
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(x.size(0)).to(x.device)
    return lam*x + (1-lam)*x[idx], y, y[idx], lam

def cutmix_data(x, y, alpha=1.0):
    lam = np.random.beta(alpha, alpha)
    idx = torch.randperm(x.size(0)).to(x.device)
    W, H = x.size(2), x.size(3)
    r = np.sqrt(1.0 - lam)
    cw, ch = int(W*r), int(H*r)
    cx, cy = np.random.randint(W), np.random.randint(H)
    x1, y1 = np.clip(cx - cw//2, 0, W), np.clip(cy - ch//2, 0, H)
    x2, y2 = np.clip(cx + cw//2, 0, W), np.clip(cy + ch//2, 0, H)
    mixed = x.clone()
    mixed[:, :, x1:x2, y1:y2] = x[idx, :, x1:x2, y1:y2]
    return mixed, y, y[idx], 1 - (x2-x1)*(y2-y1)/(W*H)

def train_one_epoch(model, loader, optimizer, scaler, criterion, epoch, mixup_prob=0.3):
    model.train()
    total_loss, preds, labs = 0, [], []
    for images, labels in tqdm(loader, desc=f'Epoch {epoch+1}', leave=False):
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        optimizer.zero_grad(set_to_none=True)
        mixed = False
        if np.random.random() < mixup_prob:
            if np.random.random() < 0.5:
                images, la, lb, lam = mixup_data(images, labels, 0.2)
                mixed = True
            else:
                images, la, lb, lam = cutmix_data(images, labels, 1.0)
                mixed = True
        with autocast():
            out = model(images)
            loss = lam * criterion(out, la) + (1-lam) * criterion(out, lb) if mixed else criterion(out, labels)
            if not mixed:
                preds.extend(out.argmax(1).cpu().numpy())
                labs.extend(labels.cpu().numpy())
        scaler.scale(loss).backward()
        scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        scaler.step(optimizer)
        scaler.update()
        total_loss += loss.item()
    return total_loss/len(loader), accuracy_score(labs, preds) if labs else 0

@torch.no_grad()
def validate(model, loader, criterion):
    model.eval()
    total_loss, preds, labs, probs = 0, [], [], []
    for images, labels in tqdm(loader, desc='Validating', leave=False):
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        with autocast():
            out = model(images)
            loss = criterion(out, labels)
        total_loss += loss.item()
        p = F.softmax(out, dim=1)
        preds.extend(out.argmax(1).cpu().numpy())
        labs.extend(labels.cpu().numpy())
        probs.extend(p.cpu().numpy())
    return total_loss/len(loader), accuracy_score(labs, preds), f1_score(labs, preds, average='weighted', zero_division=0), preds, labs, probs

def create_combined_loaders(all_datasets, batch_size=32):
    """Create combined dataloaders from all datasets"""
    train_ds_list, val_ds_list, test_ds_list = [], [], []
    
    for name, data in all_datasets.items():
        config = data['config']
        is_grayscale = config.get('is_grayscale', False)
        class_names = config['class_names']
        
        if data['train']:
            train_ds = UnifiedMedicalDataset(
                data['train'], class_names,
                transform=get_medical_transforms('train', is_grayscale=is_grayscale),
                is_grayscale=is_grayscale,
                dataset_name=name
            )
            train_ds_list.append(train_ds)
        
        if data['val']:
            val_ds = UnifiedMedicalDataset(
                data['val'], class_names,
                transform=get_medical_transforms('val', is_grayscale=is_grayscale),
                is_grayscale=is_grayscale,
                dataset_name=name
            )
            val_ds_list.append(val_ds)
        
        if data['test']:
            test_ds = UnifiedMedicalDataset(
                data['test'], class_names,
                transform=get_medical_transforms('test', is_grayscale=is_grayscale),
                is_grayscale=is_grayscale,
                dataset_name=name
            )
            test_ds_list.append(test_ds)
    
    combined_train = ConcatDataset(train_ds_list) if train_ds_list else None
    combined_val = ConcatDataset(val_ds_list) if val_ds_list else None
    combined_test = ConcatDataset(test_ds_list) if test_ds_list else None
    
    loaders = {}
    if combined_train:
        loaders['train'] = DataLoader(combined_train, batch_size=batch_size, shuffle=True, num_workers=2, pin_memory=True)
    if combined_val:
        loaders['val'] = DataLoader(combined_val, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
    if combined_test:
        loaders['test'] = DataLoader(combined_test, batch_size=batch_size, shuffle=False, num_workers=2, pin_memory=True)
    
    return loaders, combined_train, combined_val, combined_test

# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 8 — Main Execution                                                       ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

print("\n" + "="*60)
print("🚀 LOADING ALL MEDICAL IMAGING DATASETS")
print("="*60)

# Dictionary of dataset loaders
dataset_loaders = {
    'FracAtlas': load_fracatlas,
    'TB Chest X-Ray': load_tb_xray,
    'Pneumonia Chest X-Ray': load_pneumonia_xray,
    'Diabetic Retinopathy': load_retina_dataset,
    'Brain Tumor': load_brain_tumor,
    'Bone Fracture': load_bone_fracture,
    'HAM10000': load_ham10000,
}

# Load all datasets
all_datasets = {}
all_class_names = []
skipped_datasets = []

for name, loader in dataset_loaders.items():
    print(f"\n{'='*60}")
    print(f"📂 Loading: {name}")
    print(f"{'='*60}")
    
    samples, class_names = loader()
    
    if samples:
        # Split into train/val/test
        train_s, val_s, test_s, class_names = train_test_split_samples(samples, class_names)
        
        if train_s or val_s or test_s:
            # Determine if grayscale
            is_grayscale = any(x in name.lower() for x in ['xray', 'chest', 'tb', 'fracture', 'bone'])
            
            all_datasets[name] = {
                'train': train_s,
                'val': val_s,
                'test': test_s,
                'config': {
                    'name': name,
                    'class_names': class_names,
                    'is_grayscale': is_grayscale
                }
            }
            
            total_samples = len(train_s) + len(val_s) + len(test_s)
            print(f"\n✅ Successfully loaded {name}:")
            print(f"   Train: {len(train_s)} | Val: {len(val_s)} | Test: {len(test_s)} | Total: {total_samples}")
            print(f"   Classes: {class_names}")
            all_class_names.extend(class_names)
        else:
            print(f"❌ Failed to split {name}")
            skipped_datasets.append(name)
    else:
        print(f"❌ No samples found for {name}")
        skipped_datasets.append(name)

# Display summary
print("\n" + "="*60)
print("📊 LOADING SUMMARY")
print("="*60)

total_images = 0
for name, data in all_datasets.items():
    total = len(data['train']) + len(data['val']) + len(data['test'])
    total_images += total
    print(f"   {name}: {total} images")

print(f"\n✅ Total images loaded: {total_images}")
print(f"✅ Datasets loaded: {len(all_datasets)}")

if skipped_datasets:
    print(f"⚠️ Datasets skipped: {skipped_datasets}")

if not all_datasets:
    raise ValueError("❌ No datasets loaded! Please check your dataset paths.")

# Get unique classes
unique_classes = sorted(list(set(all_class_names)))
NUM_CLASSES = len(unique_classes)
print(f"\n📊 Total Unique Classes: {NUM_CLASSES}")
print(f"   Classes: {unique_classes}")

# Create combined loaders
print("\n" + "="*60)
print("📊 CREATING COMBINED DATALOADERS")
print("="*60)

loaders, train_ds, val_ds, test_ds = create_combined_loaders(all_datasets, batch_size=HPARAMS['batch_size'])

if not loaders or not loaders.get('train'):
    raise ValueError("❌ Failed to create dataloaders!")

print(f"\n✅ Dataloaders created:")
print(f"   Train batches: {len(loaders['train'])}")
if 'val' in loaders:
    print(f"   Val batches: {len(loaders['val'])}")
if 'test' in loaders:
    print(f"   Test batches: {len(loaders['test'])}")

# Continue with training
print("\n" + "="*60)
print("🚀 STARTING COMBINED TRAINING - 50 EPOCHS")
print("="*60)

# Initialize model
model = EnhancedHViT(num_classes=NUM_CLASSES).to(DEVICE)
criterion = SmoothCELoss(label_smoothing=HPARAMS['label_smoothing'])

# Freeze backbone initially
for p in model.backbone.feature_extractor.parameters():
    p.requires_grad_(False)

# Optimizer
optimizer = optim.AdamW([
    {'params': [p for n, p in model.named_parameters() if 'backbone' in n], 'lr': HPARAMS['lr_backbone']},
    {'params': [p for n, p in model.named_parameters() if 'backbone' not in n], 'lr': HPARAMS['lr_head']}
], weight_decay=HPARAMS['weight_decay'])

scheduler = optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=15, T_mult=2, eta_min=1e-7)
scaler = GradScaler()

# Training loop - 50 epochs
best_f1, patience_counter = 0, 0
history = {'train_loss': [], 'val_loss': [], 'train_acc': [], 'val_acc': [], 'val_f1': [], 'lr': []}

for epoch in range(HPARAMS['num_epochs']):
    if epoch == HPARAMS['freeze_epochs']:
        print("\n🔄 Unfreezing backbone...")
        model.unfreeze_backbone_partial(3)
    
    # Training
    train_loss, train_acc = train_one_epoch(model, loaders['train'], optimizer, scaler, criterion, epoch)
    
    # Validation
    if 'val' in loaders:
        val_loss, val_acc, val_f1, _, _, _ = validate(model, loaders['val'], criterion)
        scheduler.step()
        
        history['train_loss'].append(train_loss)
        history['val_loss'].append(val_loss)
        history['train_acc'].append(train_acc)
        history['val_acc'].append(val_acc)
        history['val_f1'].append(val_f1)
        history['lr'].append(optimizer.param_groups[0]['lr'])
        
        print(f"Epoch {epoch+1}/{HPARAMS['num_epochs']}: Train Acc={train_acc:.3f} | Val Acc={val_acc:.3f}, Val F1={val_f1:.3f}")
        
        if val_f1 > best_f1:
            best_f1 = val_f1
            patience_counter = 0
            torch.save({
                'model': model.state_dict(),
                'f1': val_f1,
                'num_classes': NUM_CLASSES,
                'class_names': unique_classes,
                'epoch': epoch
            }, CKDIR / 'best_model.pth')
            print(f"   ✅ Best model saved! (F1={val_f1:.4f})")
        else:
            patience_counter += 1
            if patience_counter >= HPARAMS['patience']:
                print(f"⏹️ Early stopping after {epoch+1} epochs")
                break
    else:
        history['train_loss'].append(train_loss)
        history['train_acc'].append(train_acc)
        print(f"Epoch {epoch+1}/{HPARAMS['num_epochs']}: Train Acc={train_acc:.3f}")

print("\n✅ Combined Training Complete!")

# Test evaluation
if 'test' in loaders:
    print("\n" + "="*60)
    print("📊 EVALUATING ON TEST SET")
    print("="*60)
    
    # Load best model
    if (CKDIR / 'best_model.pth').exists():
        checkpoint = torch.load(CKDIR / 'best_model.pth', map_location=DEVICE, weights_only=False)
        model.load_state_dict(checkpoint['model'])
        print(f"✅ Loaded best model from epoch {checkpoint.get('epoch', 'unknown')} with F1={checkpoint.get('f1', 0):.4f}")
    
    model.eval()
    all_p, all_l, all_pr = [], [], []
    
    for images, labels in tqdm(loaders['test'], desc="Testing"):
        images, labels = images.to(DEVICE), labels.to(DEVICE)
        with torch.no_grad():
            with autocast():
                out = model(images)
                pr = F.softmax(out, dim=1)
        all_p.extend(out.argmax(1).cpu().numpy())
        all_l.extend(labels.cpu().numpy())
        all_pr.extend(pr.cpu().numpy())
    
    all_p, all_l, all_pr = np.array(all_p), np.array(all_l), np.array(all_pr)
    
    # Metrics
    test_acc = accuracy_score(all_l, all_p)
    test_f1 = f1_score(all_l, all_p, average='weighted')
    
    if NUM_CLASSES > 2:
        test_auc = roc_auc_score(all_l, all_pr, multi_class='ovr', average='weighted')
    else:
        test_auc = roc_auc_score(all_l, all_pr[:, 1])
    
    test_mcc = matthews_corrcoef(all_l, all_p)
    test_kappa = cohen_kappa_score(all_l, all_p)
    test_logloss = log_loss(all_l, all_pr)
    
    print(f"\n✅ Test Results:")
    print(f"   Accuracy:  {test_acc:.4f}")
    print(f"   F1-Score:  {test_f1:.4f}")
    print(f"   AUC-ROC:   {test_auc:.4f}")
    print(f"   MCC:       {test_mcc:.4f}")
    print(f"   Kappa:     {test_kappa:.4f}")
    print(f"   Log Loss:  {test_logloss:.4f}")
    print(f"   Total Samples: {len(all_p)}")
    
    # Confusion Matrix
    cm = confusion_matrix(all_l, all_p)
    print(f"\n📊 Confusion Matrix:")
    print(cm)
    
    # Save results
    results = {
        'accuracy': test_acc,
        'f1_score': test_f1,
        'auc_roc': test_auc,
        'mcc': test_mcc,
        'kappa': test_kappa,
        'log_loss': test_logloss,
        'num_samples': len(all_p),
        'num_classes': NUM_CLASSES,
        'class_names': unique_classes,
        'confusion_matrix': cm.tolist()
    }
    
    with open(OUTDIR / 'test_results.json', 'w') as f:
        json.dump(results, f, indent=2)
    
    print(f"\n✅ Results saved to {OUTDIR / 'test_results.json'}")

print("\n" + "="*60)
print("🎉 ALL COMPLETE!")
print(f"📁 Checkpoints: {CKDIR}")
print(f"📁 Plots: {PLTDIR}")
print("="*60)
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════════╗
# ║  CELL 10 — ALL 20 PLOTS FOR COMBINED MEDICAL DATASET                          ║
# ╚══════════════════════════════════════════════════════════════════════════════════╝

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import cv2
from sklearn.metrics import (accuracy_score, f1_score, roc_curve, auc, confusion_matrix,
                             classification_report, precision_recall_curve, average_precision_score,
                             roc_auc_score, matthews_corrcoef, cohen_kappa_score, log_loss, brier_score_loss)
from sklearn.preprocessing import label_binarize
from sklearn.manifold import TSNE
from scipy.ndimage import zoom
import warnings
warnings.filterwarnings('ignore')

# Re-run inference to get predictions if not already available
print("📊 Generating all 20 plots for combined medical dataset...")

# Get unique class names from all datasets
all_class_names = []
dataset_class_maps = {}
for name, data in all_datasets.items():
    class_names = data['config']['class_names']
    dataset_class_maps[name] = class_names
    all_class_names.extend(class_names)

unique_classes = sorted(list(set(all_class_names)))
NUM_CLASSES = len(unique_classes)

# Create class name mapping
class_to_idx = {cls: i for i, cls in enumerate(unique_classes)}

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 1: TRAINING HISTORY
# ─────────────────────────────────────────────────────────────────────────────────
try:
    if len(history.get('train_loss', [])) > 0:
        fig, ax = plt.subplots(2, 2, figsize=(14, 10))
        fig.suptitle('Combined Medical Dataset Training Dynamics (7 Datasets)', 
                    fontsize=16, fontweight='bold', y=1.02)
        
        ax[0,0].plot(history['train_loss'], label='Train', color='#e74c3c', lw=2)
        ax[0,0].plot(history['val_loss'], label='Validation', color='#3498db', lw=2)
        ax[0,0].set_title('Loss Curves', fontsize=12, fontweight='bold')
        ax[0,0].legend(); ax[0,0].grid(True, alpha=0.3)
        
        ax[0,1].plot(history['train_acc'], label='Train', color='#e74c3c', lw=2)
        ax[0,1].plot(history['val_acc'], label='Validation', color='#3498db', lw=2)
        ax[0,1].set_title('Accuracy Curves', fontsize=12, fontweight='bold')
        ax[0,1].set_ylim(0,1.05); ax[0,1].legend(); ax[0,1].grid(True, alpha=0.3)
        
        ax[1,0].plot(history['val_f1'], label='Validation F1', color='#2ecc71', lw=2)
        ax[1,0].set_title('F1 Score', fontsize=12, fontweight='bold')
        ax[1,0].set_ylim(0,1.05); ax[1,0].legend(); ax[1,0].grid(True, alpha=0.3)
        
        ax[1,1].plot(history['lr'], color='#9b59b6', lw=2)
        ax[1,1].set_title('Learning Rate Schedule', fontsize=12, fontweight='bold')
        ax[1,1].set_yscale('log'); ax[1,1].grid(True, alpha=0.3)
        
        plt.tight_layout()
        plt.savefig(PLTDIR / '1_training_history.png', dpi=300, bbox_inches='tight')
        plt.show()
        print("✅ Plot 1: Training History")
except Exception as e:
    print(f"Plot 1 error: {e}")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 2: CONFUSION MATRICES
# ─────────────────────────────────────────────────────────────────────────────────
cm = confusion_matrix(all_l, all_p)
cm_norm = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]

fig, axes = plt.subplots(1, 2, figsize=(20, 8))
fig.suptitle('Combined Medical Dataset Confusion Matrix', fontsize=16, fontweight='bold')

# Show top 20 classes if too many
display_classes = unique_classes[:20] if len(unique_classes) > 20 else unique_classes
display_cm = cm[:len(display_classes), :len(display_classes)]
display_cm_norm = cm_norm[:len(display_classes), :len(display_classes)]

sns.heatmap(display_cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=display_classes, yticklabels=display_classes, 
            ax=axes[0], cbar=False)
axes[0].set_title('Absolute Counts', fontsize=12, fontweight='bold')
axes[0].set_ylabel('True Label'); axes[0].set_xlabel('Predicted Label')

sns.heatmap(display_cm_norm, annot=True, fmt='.2%', cmap='Greens',
            xticklabels=display_classes, yticklabels=display_classes,
            ax=axes[1], cbar=False)
axes[1].set_title('Normalized (Recall per Class)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('True Label'); axes[1].set_xlabel('Predicted Label')

plt.tight_layout()
plt.savefig(PLTDIR / '2_confusion_matrices.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 2: Confusion Matrices")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 3: ROC CURVES
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(12, 8))

# Limit to first 10 classes for readability
n_classes_plot = min(NUM_CLASSES, 10)
colors = plt.cm.tab10(np.linspace(0, 1, n_classes_plot))

for i in range(n_classes_plot):
    fpr, tpr, _ = roc_curve(labels_bin[:, i], all_pr[:, i])
    roc_auc_val = auc(fpr, tpr)
    label = unique_classes[i] if i < len(unique_classes) else f'Class {i}'
    ax.plot(fpr, tpr, color=colors[i], lw=2, label=f'{label} (AUC = {roc_auc_val:.3f})')

ax.plot([0, 1], [0, 1], 'k--', lw=1, alpha=0.5)
ax.set_title('ROC Curves - Combined Medical Dataset', fontsize=14, fontweight='bold')
ax.set_xlabel('False Positive Rate', fontsize=12)
ax.set_ylabel('True Positive Rate', fontsize=12)
ax.legend(loc='lower right', fontsize=8)
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig(PLTDIR / '3_roc_curves.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 3: ROC Curves")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 4: PRECISION-RECALL CURVES
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(12, 8))

for i in range(n_classes_plot):
    prec, rec, _ = precision_recall_curve(labels_bin[:, i], all_pr[:, i])
    ap = average_precision_score(labels_bin[:, i], all_pr[:, i])
    label = unique_classes[i] if i < len(unique_classes) else f'Class {i}'
    ax.plot(rec, prec, color=colors[i], lw=2, label=f'{label} (AP = {ap:.3f})')

ax.set_title('Precision-Recall Curves - Combined Dataset', fontsize=14, fontweight='bold')
ax.set_xlabel('Recall', fontsize=12)
ax.set_ylabel('Precision', fontsize=12)
ax.legend(loc='lower left', fontsize=8)
ax.grid(True, alpha=0.3)
ax.set_xlim([0,1]); ax.set_ylim([0,1.05])

plt.tight_layout()
plt.savefig(PLTDIR / '4_precision_recall.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 4: Precision-Recall Curves")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 5: PER-CLASS METRICS BARS
# ─────────────────────────────────────────────────────────────────────────────────
report = classification_report(all_l, all_p, target_names=unique_classes[:len(set(all_l))], 
                              output_dict=True, zero_division=0)
metrics_df = pd.DataFrame(report).T

# Filter only class metrics (not averages)
class_metrics = metrics_df[~metrics_df.index.str.contains('weighted|macro|accuracy')]
class_metrics = class_metrics[['precision', 'recall', 'f1-score']].head(20)  # Limit to 20

fig, ax = plt.subplots(figsize=(14, 8))
x = np.arange(len(class_metrics))
width = 0.25

bars1 = ax.bar(x - width, class_metrics['precision'], width, label='Precision', 
               color='#3498db', edgecolor='black')
bars2 = ax.bar(x, class_metrics['recall'], width, label='Recall', 
               color='#2ecc71', edgecolor='black')
bars3 = ax.bar(x + width, class_metrics['f1-score'], width, label='F1-Score', 
               color='#e74c3c', edgecolor='black')

ax.set_title('Per-Class Performance Metrics (Top 20 Classes)', fontsize=14, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(class_metrics.index, rotation=45, ha='right')
ax.legend()
ax.set_ylim(0,1.15)
ax.grid(axis='y', alpha=0.3)

# Add value labels on bars
for bars in [bars1, bars2, bars3]:
    for bar in bars:
        height = bar.get_height()
        if height > 0:
            ax.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                    f'{height:.2f}', ha='center', va='bottom', fontsize=8)

plt.tight_layout()
plt.savefig(PLTDIR / '5_per_class_metrics.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 5: Per-Class Metrics")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 6: CONFIDENCE DISTRIBUTION
# ─────────────────────────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(max_probs[correct_mask], bins=20, alpha=0.7, 
        label=f'Correct Predictions (n={correct_mask.sum()})', 
        color='#2ecc71', edgecolor='black')
ax.hist(max_probs[~correct_mask], bins=20, alpha=0.7, 
        label=f'Incorrect Predictions (n={(~correct_mask).sum()})', 
        color='#e74c3c', edgecolor='black')

ax.set_title('Model Confidence Distribution', fontsize=14, fontweight='bold')
ax.set_xlabel('Maximum Class Probability', fontsize=12)
ax.set_ylabel('Count', fontsize=12)
ax.legend()
ax.grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.savefig(PLTDIR / '6_confidence_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 6: Confidence Distribution")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 7: CALIBRATION CURVE
# ─────────────────────────────────────────────────────────────────────────────────
def calibration_curve(y_true, y_prob, n_bins=10):
    bins = np.linspace(0, 1, n_bins + 1)
    accs, confs = [], []
    for i in range(n_bins):
        mask = (y_prob > bins[i]) & (y_prob <= bins[i+1])
        if i == n_bins - 1:
            mask = (y_prob >= bins[i]) & (y_prob <= bins[i+1])
        if mask.sum() > 0:
            accs.append(y_true[mask].mean())
            confs.append(y_prob[mask].mean())
    return np.array(confs), np.array(accs)

fig, ax = plt.subplots(figsize=(8, 8))

# Plot calibration for first few classes
for i in range(min(NUM_CLASSES, 5)):
    conf, acc = calibration_curve((all_l == i).astype(int), all_pr[:, i])
    if len(conf) > 0:
        label = unique_classes[i] if i < len(unique_classes) else f'Class {i}'
        ax.plot(conf, acc, marker='o', lw=2, label=label, markersize=6)

ax.plot([0, 1], [0, 1], 'k--', lw=1, label='Perfectly Calibrated')
ax.set_title('Reliability Diagram - Calibration', fontsize=14, fontweight='bold')
ax.set_xlabel('Mean Predicted Probability', fontsize=12)
ax.set_ylabel('Fraction of Positives', fontsize=12)
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_xlim([0,1]); ax.set_ylim([0,1])

plt.tight_layout()
plt.savefig(PLTDIR / '7_calibration_curve.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 7: Calibration Curve")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 8: SAMPLE PREDICTIONS
# ─────────────────────────────────────────────────────────────────────────────────
correct_idxs, incorrect_idxs = np.where(correct_mask)[0], np.where(~correct_mask)[0]
np.random.seed(42)

n_samples = min(8, len(correct_idxs), len(incorrect_idxs))
samp_corr = np.random.choice(correct_idxs, n_samples, replace=False) if len(correct_idxs) > 0 else []
samp_incorr = np.random.choice(incorrect_idxs, n_samples, replace=False) if len(incorrect_idxs) > 0 else []

fig, axes = plt.subplots(2, n_samples, figsize=(4*n_samples, 6))

if n_samples > 0:
    for i in range(n_samples):
        # Correct predictions
        if i < len(samp_corr):
            idx = samp_corr[i]
            img_path = None
            # Find image path from datasets
            for name, data in all_datasets.items():
                if idx < len(data['test']):
                    img_path = data['test'][idx][0]
                    break
            if img_path:
                img = cv2.imread(img_path)
                if img is not None:
                    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    axes[0, i].imshow(img)
                    axes[0, i].set_title(f"T:{unique_classes[all_l[idx]]}\nP:{unique_classes[all_p[idx]]}\n{max_probs[idx]:.1%}",
                                        fontsize=8, color='green')
                axes[0, i].axis('off')
        
        # Incorrect predictions
        if i < len(samp_incorr):
            idx = samp_incorr[i]
            img_path = None
            for name, data in all_datasets.items():
                if idx < len(data['test']):
                    img_path = data['test'][idx][0]
                    break
            if img_path:
                img = cv2.imread(img_path)
                if img is not None:
                    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
                    axes[1, i].imshow(img)
                    axes[1, i].set_title(f"T:{unique_classes[all_l[idx]]}\nP:{unique_classes[all_p[idx]]}\n{max_probs[idx]:.1%}",
                                        fontsize=8, color='red')
                axes[1, i].axis('off')

if n_samples > 0:
    axes[0, 0].set_ylabel('CORRECT', fontsize=12, fontweight='bold', color='green', 
                         rotation=0, labelpad=60, va='center')
    axes[1, 0].set_ylabel('INCORRECT', fontsize=12, fontweight='bold', color='red', 
                         rotation=0, labelpad=60, va='center')

fig.suptitle('Sample Predictions (T=True, P=Predicted)', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig(PLTDIR / '8_sample_predictions.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 8: Sample Predictions")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 9: VIOLIN PLOT PROBABILITIES
# ─────────────────────────────────────────────────────────────────────────────────
plot_data = []
n_classes_violin = min(NUM_CLASSES, 10)
for i in range(n_classes_violin):
    class_probs = all_pr[all_l == i, i]
    for prob in class_probs[:1000]:  # Limit for memory
        label = unique_classes[i] if i < len(unique_classes) else f'Class {i}'
        plot_data.append({'True Class': label, 'Assigned Probability': prob})

df_probs = pd.DataFrame(plot_data)

fig, ax = plt.subplots(figsize=(14, 6))
sns.violinplot(data=df_probs, x='True Class', y='Assigned Probability', 
               palette='husl', inner='quartile', ax=ax)
ax.set_title('Probability Density for True Classes', fontsize=14, fontweight='bold')
ax.set_ylim(0,1.05)
ax.grid(axis='y', alpha=0.3)
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.savefig(PLTDIR / '9_probability_violin.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 9: Probability Violin Plot")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 10: t-SNE FEATURE VISUALIZATION
# ─────────────────────────────────────────────────────────────────────────────────
print("⏳ Extracting t-SNE features (may take a minute)...")

feature_tensor = None
def hook_fn(module, inp, out):
    global feature_tensor
    feature_tensor = out

handle = model.norm.register_forward_hook(hook_fn)

# Sample indices for t-SNE
tsne_indices = np.random.choice(len(test_ds), min(800, len(test_ds)), replace=False)
features_list, labels_tsne = [], []

for idx in tsne_indices:
    # Get image and label from combined dataset
    img, label = test_ds[idx]
    img = img.unsqueeze(0).to(DEVICE)
    with torch.no_grad():
        _ = model(img)
    features_list.append(feature_tensor[:, 0, :].cpu().numpy()[0])
    labels_tsne.append(label)

handle.remove()

# t-SNE
tsne = TSNE(n_components=2, perplexity=30, random_state=42, n_iter=1000)
tsne_results = tsne.fit_transform(np.array(features_list))

fig, ax = plt.subplots(figsize=(12, 10))

# Color by class
unique_labels = np.unique(labels_tsne)
n_classes_tsne = min(len(unique_labels), 10)
colors_tsne = plt.cm.tab10(np.linspace(0, 1, n_classes_tsne))

for i, cls in enumerate(unique_labels[:n_classes_tsne]):
    mask = np.array(labels_tsne) == cls
    label = unique_classes[cls] if cls < len(unique_classes) else f'Class {cls}'
    ax.scatter(tsne_results[mask, 0], tsne_results[mask, 1],
              c=[colors_tsne[i]], label=label, alpha=0.6, s=15, 
              edgecolors='white', linewidth=0.5)

ax.set_title('t-SNE Visualization of Deep Features', fontsize=14, fontweight='bold')
ax.legend(markerscale=2, loc='best', fontsize=8)
ax.grid(True, alpha=0.2)
ax.set_xticks([]); ax.set_yticks([])

plt.tight_layout()
plt.savefig(PLTDIR / '10_tsne_features.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 10: t-SNE Visualization")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 11: ERROR ANALYSIS
# ─────────────────────────────────────────────────────────────────────────────────
errors = [{'True': unique_classes[all_l[i]], 'Predicted': unique_classes[all_p[i]]} 
          for i in range(len(all_l)) if all_l[i] != all_p[i]]

if errors:
    error_df = pd.DataFrame(errors)
    error_counts = error_df.groupby(['True', 'Predicted']).size().reset_index(name='Count')
    error_counts = error_counts.sort_values(by='Count', ascending=False).head(15)
    
    fig, ax = plt.subplots(figsize=(12, 8))
    error_counts['Pair'] = error_counts['True'] + ' → ' + error_counts['Predicted']
    sns.barplot(data=error_counts, x='Count', y='Pair', palette='Reds_r', ax=ax)
    ax.set_title('Top Misclassification Pairs', fontsize=14, fontweight='bold')
    ax.set_xlabel('Number of Errors', fontsize=12)
    
    for i, v in enumerate(error_counts['Count']):
        ax.text(v + 0.5, i, str(v), color='black', va='center', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig(PLTDIR / '11_error_analysis.png', dpi=300, bbox_inches='tight')
    plt.show()
    print("✅ Plot 11: Error Analysis")
else:
    print("✅ Plot 11: No errors found!")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 12: DATASET CONTRIBUTION
# ─────────────────────────────────────────────────────────────────────────────────
dataset_counts = {}
for name, data in all_datasets.items():
    dataset_counts[name] = len(data['train']) + len(data['val']) + len(data['test'])

fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Dataset Contribution Analysis', fontsize=16, fontweight='bold')

# Bar plot
names = list(dataset_counts.keys())
counts = list(dataset_counts.values())
colors_ds = plt.cm.Set3(np.linspace(0, 1, len(names)))

axes[0].barh(names, counts, color=colors_ds, edgecolor='black')
axes[0].set_title('Images per Dataset', fontsize=13)
axes[0].set_xlabel('Number of Images')
for i, v in enumerate(counts):
    axes[0].text(v + 50, i, str(v), va='center', fontweight='bold')

# Pie chart
axes[1].pie(counts, labels=names, autopct='%1.1f%%', colors=colors_ds,
           startangle=90, wedgeprops={'edgecolor': 'black', 'linewidth': 1.5})
axes[1].set_title('Percentage Contribution', fontsize=13)

plt.tight_layout()
plt.savefig(PLTDIR / '12_dataset_contribution.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 12: Dataset Contribution")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 13: PER-DATASET PERFORMANCE
# ─────────────────────────────────────────────────────────────────────────────────
per_dataset_acc = {}
for name, data in all_datasets.items():
    # Get predictions for this dataset
    ds_preds = []
    ds_labels = []
    for i, (img, label) in enumerate(data['test']):
        if i < len(all_p):
            ds_preds.append(all_p[i])
            ds_labels.append(all_l[i])
    if ds_preds:
        per_dataset_acc[name] = accuracy_score(ds_labels, ds_preds)

fig, ax = plt.subplots(figsize=(12, 6))
datasets = list(per_dataset_acc.keys())
accuracies = list(per_dataset_acc.values())
colors_ds = plt.cm.Set3(np.linspace(0, 1, len(datasets)))

bars = ax.bar(datasets, accuracies, color=colors_ds, edgecolor='black')
ax.set_title('Per-Dataset Test Accuracy', fontsize=14, fontweight='bold')
ax.set_ylabel('Accuracy', fontsize=12)
ax.set_ylim(0, 1.05)
ax.axhline(y=test_acc, color='red', linestyle='--', label=f'Overall Accuracy: {test_acc:.3f}')
ax.legend()
ax.grid(axis='y', alpha=0.3)

for bar, acc in zip(bars, accuracies):
    ax.text(bar.get_x() + bar.get_width()/2., bar.get_height() + 0.01,
            f'{acc:.3f}', ha='center', va='bottom', fontweight='bold')

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig(PLTDIR / '13_per_dataset_performance.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 13: Per-Dataset Performance")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 14: CLASS IMBALANCE ANALYSIS
# ─────────────────────────────────────────────────────────────────────────────────
class_counts = Counter(all_l)
class_names_short = [unique_classes[i] if i < len(unique_classes) else f'Class {i}' 
                     for i in range(len(class_counts))]

fig, ax = plt.subplots(figsize=(14, 6))
classes = list(class_counts.keys())[:20]  # Show top 20
counts = [class_counts[c] for c in classes]
labels = [unique_classes[c] if c < len(unique_classes) else f'Class {c}' for c in classes]

sns.barplot(x=labels, y=counts, palette='viridis', ax=ax)
ax.set_title('Class Distribution in Combined Dataset', fontsize=14, fontweight='bold')
ax.set_xlabel('Class', fontsize=12)
ax.set_ylabel('Number of Samples', fontsize=12)
ax.set_yscale('log')  # Log scale for better visualization
ax.grid(axis='y', alpha=0.3)

# Add count labels
for i, v in enumerate(counts):
    ax.text(i, v + 5, str(v), ha='center', va='bottom', fontsize=8, rotation=45)

plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig(PLTDIR / '14_class_imbalance.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 14: Class Imbalance Analysis")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 15: FEATURE MAPS HIERARCHY
# ─────────────────────────────────────────────────────────────────────────────────
print("\n🧠 Extracting Hierarchical Feature Maps...")

# Get a sample image
sample_idx = 0
img, _ = test_ds[sample_idx]
img = img.unsqueeze(0).to(DEVICE)

feature_maps = {}
def get_hook(name):
    def hook(module, input, output):
        feature_maps[name] = output.detach().cpu()
    return hook

# Register hooks
model.backbone.feature_extractor[1].register_forward_hook(get_hook('Early_Layer'))
model.backbone.feature_extractor[4].register_forward_hook(get_hook('Mid_Layer'))
model.backbone.feature_extractor[7].register_forward_hook(get_hook('Late_Layer'))

with torch.no_grad():
    _ = model(img)

fig, axes = plt.subplots(3, 8, figsize=(20, 7))
fig.suptitle('Hierarchical Feature Extraction (Sample from Combined Dataset)', 
            fontsize=16, fontweight='bold')

for row_idx, (layer, title) in enumerate(zip(['Early_Layer', 'Mid_Layer', 'Late_Layer'],
                                             ['Early Features (Edges/Texture)', 
                                              'Mid Features (Anatomy/Structure)', 
                                              'Late Features (High-level Patterns)'])):
    if layer in feature_maps:
        fm = feature_maps[layer].squeeze()
        n_channels = fm.shape[0]
        channel_idxs = np.linspace(0, n_channels - 1, 8, dtype=int)
        for col_idx, c_idx in enumerate(channel_idxs):
            if c_idx < n_channels:
                ch_img = fm[c_idx].numpy()
                ch_img = (ch_img - ch_img.min()) / (ch_img.max() - ch_img.min() + 1e-8)
                axes[row_idx, col_idx].imshow(ch_img, cmap='magma')
                axes[row_idx, col_idx].set_title(f'Ch {c_idx}', fontsize=9)
                axes[row_idx, col_idx].axis('off')
    
    axes[row_idx, -1].axis('off')
    axes[row_idx, -1].text(1.1, 0.5, title, transform=axes[row_idx, -1].transAxes,
                          fontsize=12, fontweight='bold', va='center', rotation=270)

plt.tight_layout()
plt.savefig(PLTDIR / '15_feature_maps_hierarchy.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 15: Feature Maps Hierarchy")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 16: CONFIDENCE VS ACCURACY
# ─────────────────────────────────────────────────────────────────────────────────
confidence_bins = np.linspace(0, 1, 11)
bin_accuracies = []
bin_counts = []

for i in range(len(confidence_bins) - 1):
    mask = (max_probs >= confidence_bins[i]) & (max_probs < confidence_bins[i+1])
    if mask.sum() > 0:
        bin_accuracies.append(accuracy_score(all_l[mask], all_p[mask]))
        bin_counts.append(mask.sum())

fig, ax1 = plt.subplots(figsize=(12, 6))

# Accuracy per confidence bin
bin_centers = (confidence_bins[:-1] + confidence_bins[1:]) / 2
ax1.bar(bin_centers, bin_accuracies, width=0.08, color='#3498db', edgecolor='black',
        alpha=0.7, label='Accuracy')
ax1.set_xlabel('Confidence Level', fontsize=12)
ax1.set_ylabel('Accuracy', fontsize=12, color='#3498db')
ax1.tick_params(axis='y', labelcolor='#3498db')
ax1.set_ylim(0, 1.05)
ax1.grid(True, alpha=0.3)

# Sample count per bin
ax2 = ax1.twinx()
ax2.bar(bin_centers, bin_counts, width=0.08, color='#e74c3c', edgecolor='black',
        alpha=0.3, label='Sample Count')
ax2.set_ylabel('Number of Samples', fontsize=12, color='#e74c3c')
ax2.tick_params(axis='y', labelcolor='#e74c3c')

ax1.set_title('Confidence vs Accuracy Analysis', fontsize=14, fontweight='bold')
fig.legend(loc='upper left', bbox_to_anchor=(0.1, 0.95))

plt.tight_layout()
plt.savefig(PLTDIR / '16_confidence_vs_accuracy.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 16: Confidence vs Accuracy")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 17: GRAD-CAM VISUALIZATION
# ─────────────────────────────────────────────────────────────────────────────────
print("\n🔥 Generating Grad-CAM Visualizations...")

class GradCAMGenerator:
    def __init__(self, model):
        self.model = model
        self.model.eval()
        self.grads, self.activations = None, None
        
        # Find last conv layer
        last_conv = None
        for m in self.model.backbone.feature_extractor.modules():
            if isinstance(m, nn.Conv2d):
                last_conv = m
        
        if last_conv:
            last_conv.register_forward_hook(lambda m, i, o: setattr(self, 'activations', o.detach()))
            last_conv.register_backward_hook(lambda m, gi, go: setattr(self, 'grads', go[0].detach()))
    
    def generate(self, img_tensor, target_class):
        img_tensor = img_tensor.unsqueeze(0).to(DEVICE).requires_grad_(True)
        
        out = self.model(img_tensor)
        self.model.zero_grad()
        oh = torch.zeros_like(out)
        oh[0, target_class] = 1
        out.backward(gradient=oh, retain_graph=True)
        
        if self.grads is not None and self.activations is not None:
            weights = self.grads.mean(dim=(2, 3), keepdim=True)
            cam = F.relu((weights * self.activations).sum(1)).squeeze().cpu().numpy()
            cam = (cam - cam.min()) / (cam.max() - cam.min() + 1e-8)
        else:
            cam = np.zeros((7, 7))
        
        return cam

# Generate Grad-CAM for sample images
n_samples_gradcam = min(4, len(test_ds))
fig, axes = plt.subplots(n_samples_gradcam, 3, figsize=(12, 4*n_samples_gradcam))
fig.suptitle('Grad-CAM Visualizations', fontsize=16, fontweight='bold')

grad_cam_gen = GradCAMGenerator(model)

for i in range(n_samples_gradcam):
    img, label = test_ds[i]
    img_np = img.permute(1, 2, 0).cpu().numpy()
    img_np = (img_np - img_np.min()) / (img_np.max() - img_np.min() + 1e-8)
    
    # Get Grad-CAM
    cam = grad_cam_gen.generate(img, label)
    
    # Resize CAM to image size
    cam_resized = cv2.resize(cam, (224, 224))
    cam_colored = (plt.cm.jet(cam_resized)[:,:,:3] * 255).astype(np.uint8)
    
    # Overlay
    overlay = cv2.addWeighted((img_np * 255).astype(np.uint8), 0.6, cam_colored, 0.4, 0)
    
    axes[i, 0].imshow(img_np)
    axes[i, 0].set_title(f'Original - Class: {unique_classes[label]}', fontsize=10)
    axes[i, 0].axis('off')
    
    axes[i, 1].imshow(cam_resized, cmap='jet')
    axes[i, 1].set_title('Grad-CAM Heatmap', fontsize=10)
    axes[i, 1].axis('off')
    
    axes[i, 2].imshow(overlay)
    axes[i, 2].set_title('Overlay', fontsize=10)
    axes[i, 2].axis('off')

plt.tight_layout()
plt.savefig(PLTDIR / '17_gradcam_visualization.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 17: Grad-CAM Visualization")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 18: CONFUSION MATRIX PER DATASET
# ─────────────────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 4, figsize=(20, 12))
fig.suptitle('Confusion Matrices per Dataset', fontsize=16, fontweight='bold')

for idx, (name, data) in enumerate(all_datasets.items()):
    if idx >= 8:
        break
    
    row = idx // 4
    col = idx % 4
    
    # Get predictions for this dataset
    ds_preds, ds_labels = [], []
    start_idx = sum(len(d['test']) for n, d in list(all_datasets.items())[:idx])
    end_idx = start_idx + len(data['test'])
    
    for i in range(start_idx, min(end_idx, len(all_p))):
        ds_preds.append(all_p[i])
        ds_labels.append(all_l[i])
    
    if ds_preds:
        cm_ds = confusion_matrix(ds_labels, ds_preds)
        sns.heatmap(cm_ds, annot=True, fmt='d', cmap='Blues', ax=axes[row, col],
                   xticklabels=data['config']['class_names'],
                   yticklabels=data['config']['class_names'], cbar=False)
        axes[row, col].set_title(f'{data["config"]["name"]}\n(n={len(ds_preds)})', fontsize=10)
        axes[row, col].set_xlabel('Predicted')
        axes[row, col].set_ylabel('True')

# Hide unused subplots
for idx in range(len(all_datasets), 8):
    row = idx // 4
    col = idx % 4
    axes[row, col].axis('off')

plt.tight_layout()
plt.savefig(PLTDIR / '18_confusion_per_dataset.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 18: Confusion Matrix per Dataset")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 19: DEPLOYMENT THRESHOLDS
# ─────────────────────────────────────────────────────────────────────────────────
thresholds = np.arange(0.5, 1.0, 0.05)
metrics_data = []

for thresh in thresholds:
    mask = max_probs >= thresh
    if mask.sum() == 0:
        continue
    p_thresh, l_thresh = all_p[mask], all_l[mask]
    metrics_data.append({
        'Threshold': thresh,
        'Accuracy': accuracy_score(l_thresh, p_thresh),
        'F1-Score': f1_score(l_thresh, p_thresh, average='weighted', zero_division=0),
        'Coverage': mask.sum() / len(all_p) * 100
    })

df_thresh = pd.DataFrame(metrics_data)

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.set_xlabel('Minimum Confidence Threshold', fontsize=12)
ax1.set_ylabel('Metric Score', color='#3498db', fontsize=12)
ax1.plot(df_thresh['Threshold'], df_thresh['Accuracy'], color='#3498db', 
         marker='o', lw=2, label='Accuracy')
ax1.plot(df_thresh['Threshold'], df_thresh['F1-Score'], color='#2ecc71', 
         marker='s', lw=2, label='F1-Score')
ax1.tick_params(axis='y', labelcolor='#3498db')
ax1.grid(True, alpha=0.3)

ax2 = ax1.twinx()
ax2.set_ylabel('Dataset Coverage (%)', color='#e74c3c', fontsize=12)
ax2.plot(df_thresh['Threshold'], df_thresh['Coverage'], color='#e74c3c', 
         marker='^', lw=2, linestyle='--', label='Coverage')
ax2.tick_params(axis='y', labelcolor='#e74c3c')

fig.legend(loc='lower left', bbox_to_anchor=(0.1, -0.1), ncol=3)
plt.tight_layout()
plt.savefig(PLTDIR / '19_deployment_thresholds.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 19: Deployment Thresholds")

# ─────────────────────────────────────────────────────────────────────────────────
#                          PLOT 20: FINAL SUMMARY METRICS
# ─────────────────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
fig.suptitle('Combined Model Summary Metrics', fontsize=16, fontweight='bold')

# Metric 1: Overall metrics
metrics = {
    'Accuracy': test_acc,
    'F1-Score': test_f1,
    'AUC-ROC': test_auc,
    'MCC': matthews_corrcoef(all_l, all_p),
    'Kappa': cohen_kappa_score(all_l, all_p),
    'Log Loss': log_loss(all_l, all_pr)
}

axes[0, 0].barh(list(metrics.keys()), list(metrics.values()), 
                color=plt.cm.Set2(np.linspace(0, 1, len(metrics))))
axes[0, 0].set_title('Overall Metrics', fontsize=12, fontweight='bold')
axes[0, 0].set_xlim(0, 1.05)
for i, (k, v) in enumerate(metrics.items()):
    axes[0, 0].text(v + 0.01, i, f'{v:.3f}', va='center', fontweight='bold')

# Metric 2: Per-dataset accuracy
datasets = list(per_dataset_acc.keys())
accuracies = list(per_dataset_acc.values())
axes[0, 1].bar(datasets, accuracies, color=plt.cm.Set3(np.linspace(0, 1, len(datasets))))
axes[0, 1].set_title('Per-Dataset Accuracy', fontsize=12, fontweight='bold')
axes[0, 1].set_ylim(0, 1.05)
axes[0, 1].tick_params(axis='x', rotation=45)
for i, v in enumerate(accuracies): 
    axes[0, 1].text(i, v + 0.01, f'{v:.3f}', ha='center', fontweight='bold')

# Metric 3: Dataset contribution pie
axes[0, 2].pie(counts, labels=names, autopct='%1.1f%%', colors=colors_ds,
               startangle=90, wedgeprops={'edgecolor': 'black', 'linewidth': 1})
axes[0, 2].set_title('Dataset Contribution', fontsize=12, fontweight='bold')

# Metric 4: Confusion matrix (normalized)
sns.heatmap(display_cm_norm, annot=True, fmt='.2%', cmap='Greens',
            xticklabels=display_classes, yticklabels=display_classes,
            ax=axes[1, 0], cbar=False)
axes[1, 0].set_title('Normalized Confusion Matrix', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Predicted'); axes[1, 0].set_ylabel('True')

# Metric 5: Class distribution
class_counts_short = dict(list(class_counts.items())[:10])
axes[1, 1].bar(list(class_counts_short.keys()), list(class_counts_short.values()),
               color=plt.cm.viridis(np.linspace(0, 1, len(class_counts_short))))
axes[1, 1].set_title('Top 10 Classes Distribution', fontsize=12, fontweight='bold')
axes[1, 1].set_yscale('log')
axes[1, 1].set_xlabel('Class'); axes[1, 1].set_ylabel('Count (log scale)')

# Metric 6: Training summary
epochs = len(history['train_loss'])
axes[1, 2].text(0.1, 0.9, f'Total Epochs: {epochs}', fontsize=11, fontweight='bold')
axes[1, 2].text(0.1, 0.8, f'Total Classes: {NUM_CLASSES}', fontsize=11)
axes[1, 2].text(0.1, 0.7, f'Total Samples: {len(test_ds)}', fontsize=11)
axes[1, 2].text(0.1, 0.6, f'Best Val F1: {best_f1:.3f}', fontsize=11)
axes[1, 2].text(0.1, 0.5, f'Test Accuracy: {test_acc:.3f}', fontsize=11)
axes[1, 2].text(0.1, 0.4, f'Datasets: {len(all_datasets)}', fontsize=11)
axes[1, 2].axis('off')

plt.tight_layout()
plt.savefig(PLTDIR / '20_summary_metrics.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Plot 20: Summary Metrics")

print("\n" + "="*60)
print("✅ ALL 20 PLOTS GENERATED SUCCESSFULLY!")
print(f"📁 Location: {PLTDIR}")
print("="*60)
```
